## Có 2 cách tiếp cận để vẽ một Plot

### Cách 1: Sử dụng lệnh Matplotlib cơ bản (giao diện `plt.*`)

| Lệnh | Tác dụng |
|---|---|
| `plt.figure()` | tạo một figure |
| `plt.plot(x, y)` | vẽ đồ thị đường y theo x |
| `plt.xlabel()`, `plt.ylabel()`, `plt.title()` | đặt tên trục x, y, tiêu đề |
| `plt.grid()` | hiển thị lưới |
| `plt.legend()` | hiển thị chú thích |
| `plt.savefig()` | lưu hình về máy |
| `plt.show()` | hiển thị hình |
| `plt.clf()` | xoá hình hiện tại |

**Ví dụ:** vẽ đồ thị hàm số $y = x^3$:

```python
x = np.arange(-5, 5, 0.01)
y = x ** 3

plt.title('Hàm y = x^3')
plt.xlabel('Trục x')
plt.ylabel('Trục y')
plt.grid()
plt.plot(x, y, label='y = x^3')
plt.legend()
plt.show()
```

Vẽ nhiều hình trên 1 Figure bằng `plt.subplot(nrows, ncols, plot_number)`:

```python
plt.subplot(1, 2, 1)   # 1 hàng, 2 cột, vẽ ở vị trí 1
plt.plot(x, y, 'red')

plt.subplot(1, 2, 2)   # vẽ ở vị trí 2
plt.plot(x, y, 'green')

plt.tight_layout()     # tránh các subplot đè lên nhau
plt.show()
```

### Cách 2: Phương thức hướng đối tượng (khuyến nghị)

Thành phần chính của **figure** là các **axes**. Figure chỉ là khung chứa; **axes** mới thật sự là nơi các hình được vẽ lên. Vì vậy **không có** những cách gọi kiểu `fig.plot()` hay `fig.xlabel()`.

Cách phổ biến nhất là tạo đồng thời Figure và Axes bằng `plt.subplots()`. Với axes, các hàm đặt tên đổi thành `ax.set_xlabel()`, `ax.set_ylabel()`, `ax.set_title()`:

```python
fig, ax = plt.subplots(figsize=(8, 4))     # figsize: kích thước (inch)

x = np.linspace(0, 2, 100)
ax.plot(x, x, label='linear')
ax.plot(x, x**2, label='quadratic')
ax.plot(x, x**3, label='cubic')

ax.set_xlabel('Trục x')
ax.set_ylabel('Trục y')
ax.set_title('Ba hàm số')
ax.legend()
plt.show()
```

Có thể đặt axes ở vị trí tuỳ ý bằng `fig.add_axes([trái, dưới, rộng, cao])` (toạ độ tương đối 0–1) — ví dụ để vẽ hình phóng to lồng trong hình lớn:

```python
fig = plt.figure(figsize=(7, 4))
ax = fig.add_axes([0.1, 0.1, 0.85, 0.85])
x = np.linspace(-5, 5, 300)
ax.plot(x, x ** 3, 'red')
ax.set_title('Hình lớn')

inset = fig.add_axes([0.2, 0.55, 0.3, 0.3], facecolor='#BECAFB')   # hình nhỏ lồng bên trong
mask = np.abs(x) < 1
inset.plot(x[mask], x[mask] ** 3, 'red')
inset.set_title('Phóng to quanh 0', fontsize=9)
plt.show()
```

Lưới nhiều axes: `plt.subplots(nrows, ncols)` trả về **mảng** axes — rất tiện khi so sánh nhiều model cạnh nhau:

```python
fig, axes = plt.subplots(2, 3, figsize=(12, 6), sharex=True)
x = np.linspace(-3, 3, 200)
funcs = {"sigmoid": 1 / (1 + np.exp(-x)), "tanh": np.tanh(x), "ReLU": np.maximum(0, x),
         "x²": x**2, "|x|": np.abs(x), "sin": np.sin(x)}
for ax, (name, values) in zip(axes.ravel(), funcs.items()):
    ax.plot(x, values)
    ax.set_title(name)
    ax.grid(alpha=0.3)
fig.suptitle("Một số hàm sẽ gặp trong khoá học")
fig.tight_layout()
plt.show()
```

### Một số thuộc tính khác của Figure
1. `figsize` — kích thước Figure (đơn vị inch).
2. `dpi` — độ phân giải (mặc định 100). `dpi` càng cao thì chất lượng ảnh càng cao.
3. Lưu Figure bằng `fig.savefig(đường_dẫn, dpi=..., bbox_inches="tight")`.

```python
import tempfile
from pathlib import Path

import matplotlib.image as mpimg

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(x, np.sin(x))
out_path = Path(tempfile.gettempdir()) / "my_figure.png"   # lưu vào thư mục tạm để không làm bẩn repo
fig.savefig(out_path, dpi=120, bbox_inches="tight")
plt.close(fig)                                              # đóng figure để giải phóng bộ nhớ

# Kiểm tra Figure đã lưu
plt.imshow(mpimg.imread(out_path))
plt.axis("off")
plt.show()
```

Xoá / đóng figure: `fig.clf()` hoặc `plt.clf()` xoá nội dung; `plt.close(fig)` đóng hẳn (quan trọng khi vẽ hàng trăm hình trong vòng lặp).

## Chỉnh trục toạ độ (Spines và Ticks)

**Spine** là đường viền ghi nhận ranh giới khu vực dữ liệu, nối các dấu tick trên trục. Ví dụ: khảo sát đồ thị hàm số $y = \sin(x)$ với trục toạ độ đặt ở giữa như sách giáo khoa:

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
x = np.arange(-10., 10., 0.2)
y = np.sin(x)

axes[0].plot(x, y)
axes[0].set_title("Mặc định")

ax = axes[1]
ax.spines['top'].set_color('none')            # ẩn đường biên trên và phải
ax.spines['right'].set_color('none')
ax.xaxis.set_ticks_position('bottom')
ax.spines['bottom'].set_position(('data', 0))  # đưa trục x về y = 0
ax.yaxis.set_ticks_position('left')
ax.spines['left'].set_position('center')       # đưa trục y vào giữa
ax.plot(x, y)
ax.set_title("Trục ở giữa")
plt.show()
```

## Vẽ đồ thị kiểu điểm, marker và linestyle

Các `marker` thông dụng:

<img src="https://imgur.com/nOixtOT.png" style="width:60%;height:60%;">

Một số kiểu `linestyle`:

<img src="https://imgur.com/YhFLaeY.png" style="width:40%;height:40%;">

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 3.5))

t = np.arange(0., 10., 0.5)
axes[0].plot(t, t**2, 'X')
axes[0].set_title("Chỉ điểm: marker 'X'")

xs, ys = [1, 2, 3, 5], [4, 5, 6, 7]
axes[1].plot(xs, ys, linestyle='--', marker='o', color='b')   # dạng đầy đủ
axes[1].plot(xs, [v - 1 for v in ys], '--ro')                 # dạng viết tắt: nét đứt, màu đỏ, marker tròn
axes[1].set_title("Vừa điểm vừa đường")

xx = np.linspace(-np.pi, np.pi, 30)
axes[2].plot(xx, np.sin(xx), '-gD', markevery=[0, 5, 12, 19])
axes[2].set_title("Đánh dấu một vài điểm")
plt.show()
```

## Các loại đồ thị dùng nhiều nhất trong ML *(bổ sung)*

| Đồ thị | Hàm | Dùng để |
|---|---|---|
| Scatter tô màu theo lớp | `ax.scatter(x, y, c=labels)` | xem phân bố các lớp, decision boundary |
| Histogram | `ax.hist(values, bins=30)` | phân phối của một feature, của residual |
| Bar | `ax.bar(names, values)` | so sánh metric giữa các model, feature importance |
| Line (log-scale) | `ax.plot(...)`, `ax.set_yscale("log")` | loss theo epoch |
| Image | `ax.imshow(img)` | ảnh, confusion matrix, heatmap tương quan |
| Contour | `ax.contourf(xx, yy, zz)` | loss surface, decision boundary |

```python
from ml_course.datasets import load_digits_data, make_classification_2d

rng = np.random.default_rng(0)
fig, axes = plt.subplots(2, 3, figsize=(15, 8))

# 1) Scatter theo lớp
X, y = make_classification_2d("moons", n=300, noise=0.15)
axes[0, 0].scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", s=15, edgecolor="k", linewidth=0.3)
axes[0, 0].set_title("Scatter tô màu theo lớp")

# 2) Histogram của 2 nhóm
axes[0, 1].hist(rng.normal(0, 1, 1000), bins=30, alpha=0.6, label="nhóm A")
axes[0, 1].hist(rng.normal(1.5, 0.8, 1000), bins=30, alpha=0.6, label="nhóm B")
axes[0, 1].legend()
axes[0, 1].set_title("Histogram")

# 3) Bar: so sánh model
axes[0, 2].bar(["baseline", "logreg", "tree", "forest"], [0.62, 0.81, 0.78, 0.86], color="#2563eb")
axes[0, 2].set_ylim(0.5, 0.9)
axes[0, 2].set_title("Accuracy theo model")

# 4) Loss theo epoch (log scale)
epochs = np.arange(1, 101)
axes[1, 0].plot(epochs, 2.0 * epochs ** -0.7, label="train")
axes[1, 0].plot(epochs, 2.0 * epochs ** -0.5 + 0.05, label="validation")
axes[1, 0].set_yscale("log")
axes[1, 0].set_xlabel("epoch")
axes[1, 0].legend()
axes[1, 0].set_title("Loss curve (log scale)")

# 5) Ảnh: chữ số viết tay 8×8
_, labels, images = load_digits_data()
axes[1, 1].imshow(images[0], cmap="gray_r")
axes[1, 1].set_title(f"Ảnh digit, nhãn = {labels[0]}")
axes[1, 1].axis("off")

# 6) Contour: một hàm loss 2 tham số
w, b = np.meshgrid(np.linspace(-3, 3, 100), np.linspace(-3, 3, 100))
loss = (w - 1) ** 2 + 3 * (b + 0.5) ** 2
cs = axes[1, 2].contourf(w, b, loss, levels=30, cmap="viridis")
fig.colorbar(cs, ax=axes[1, 2])
axes[1, 2].plot(1, -0.5, "r*", ms=15)
axes[1, 2].set_xlabel("w")
axes[1, 2].set_ylabel("b")
axes[1, 2].set_title("Loss surface — điểm tối ưu (*)")

fig.tight_layout()
plt.show()
```
