**Bài 8.** Hàm `mse` dưới đây cho kết quả sai khi `y_pred` là vector cột, và hàm `add_ratio` gây `SettingWithCopyWarning` (đôi khi không ghi được dữ liệu). Sửa cả hai.

Đầu ra kỳ vọng: `mse([1, 2, 3], [[1], [2], [4]]) = 0.3333`

```python
def mse(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    ### START CODE HERE ### (≈ 1 dòng)
    y_pred = np.asarray(y_pred, dtype=float)          # 🐛
    ### END CODE HERE ###
    return np.mean((y_true - y_pred) ** 2)


def add_ratio(df):
    ### START CODE HERE ### (≈ 2 dòng)
    long_term = df[df["tenure_months"] > 24]          # 🐛
    long_term["ratio"] = long_term["total_charges"] / long_term["monthly_charges"]
    ### END CODE HERE ###
    return long_term
```

```python
check_close(mse([1, 2, 3], [[1], [2], [4]]), 1 / 3, tol=1e-9, message="mse với y_pred dạng cột")
churn = pd.read_csv(repo_path("data", "sample", "customer_churn.csv"))
import warnings
with warnings.catch_warnings():
    warnings.simplefilter("error")      # biến warning thành lỗi để kiểm tra
    out = add_ratio(churn)
check_true("ratio" in out.columns and out["tenure_months"].min() > 24, "add_ratio chạy không có warning")
```
