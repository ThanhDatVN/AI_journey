## Phần 5 — Mini Project

**Mini project — Học thứ tự sự kiện.** Phiên gồm 10–30 sự kiện (mã 0–5 ngẫu nhiên), mỗi phiên có đúng một sự kiện 6 và một sự kiện 7;
nhãn = 1 nếu 7 xuất hiện **sau** 6. Train trên `make_sessions(2000, seed=10)`, test trên `make_sessions(1000, seed=11)`. Yêu cầu
accuracy test `acc` ≥ 0.95.

```python
import torch
from torch import nn
from torch.nn.utils.rnn import pack_padded_sequence

def make_sessions(n, seed):
    r = np.random.default_rng(seed)
    seqs, labels = [], []
    for _ in range(n):
        L = r.integers(10, 31)
        s = [int(v) for v in r.integers(0, 6, L)]
        p1, p2 = sorted(r.choice(L + 2, 2, replace=False))
        sus = r.random() < 0.5
        a, b = (6, 7) if sus else (7, 6)
        s.insert(p1, a)
        s.insert(p2, b)
        seqs.append(s)
        labels.append(int(sus))
    return seqs, np.array(labels)

tr_seqs, y_tr = make_sessions(2000, seed=10)
te_seqs, y_te = make_sessions(1000, seed=11)
torch.manual_seed(0)
### START CODE HERE ### (≈ 25 dòng)
Xtr, Ltr = None, None
Xte, Lte = None, None
ytr = None
model = None
opt = None
acc = None
### END CODE HERE ###
print(f"accuracy test = {acc:.3f}")
```
