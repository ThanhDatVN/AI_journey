**Bài 10.** **Ước lượng ARMA(1,1) bằng tổng bình phương có điều kiện (CSS).** Với $y_t = c + \phi y_{t-1} + \varepsilon_t + \theta\varepsilon_{t-1}$, cho tham số, tính đệ quy phần dư $e_t = y_t - c - \phi y_{t-1} - \theta e_{t-1}$ (với $e_0 = 0$, bắt đầu từ $t = 1$), rồi tối thiểu $\sum_t e_t^2$ bằng `scipy.optimize.minimize` (L-BFGS-B, ràng buộc $|\phi|, |\theta| \le 0.99$).

- Kiểm tra trên ARMA(1,1) mô phỏng ($c = 1$, $\phi = 0.7$, $\theta = 0.5$, $n = 3000$) và MA(1) ($\theta = -0.6$).
