## 2. Quy ước của khoá học *(bổ sung)*

| Ký hiệu | Ý nghĩa | Code |
|---|---|---|
| $n$, $d$ | số mẫu, số feature | `X.shape == (n, d)` |
| $\mathbf x^{(i)}$ | mẫu thứ $i$ (chỉ số **trên**, trong ngoặc — không phải luỹ thừa!) | `X[i]` |
| $x_j$ | feature thứ $j$ (chỉ số **dưới**) | `X[:, j]` |
| $\mathbf w, b$ hoặc $\boldsymbol\theta$ | tham số | `w`, `b` |
| $\hat y$ | dự đoán ("y mũ") | `y_hat` |
| $L(\boldsymbol\theta)$, $J(\boldsymbol\theta)$ | hàm mất mát | `loss` |
| $\nabla_{\mathbf w} L$ | gradient | `grad_w` |
| $\eta$ (AIVN: $\alpha$) | learning rate | `lr` |

> ⚠️ Tài liệu của Andrew Ng và khoá AIVN dùng $m$ cho số mẫu và $\alpha$ cho learning rate; scikit-learn lại dùng `alpha` cho hệ số regularization. Đừng nhầm!

## 3. Từ công thức đến code

Kỹ năng quan trọng nhất: nhìn công thức → viết được code vectorized. Một số mẫu thường gặp:

| Công thức | NumPy |
|---|---|
| $\sum_{i=1}^n x_i$ | `x.sum()` |
| $\frac{1}{n}\sum_{i=1}^n (y^{(i)} - \hat y^{(i)})^2$ | `np.mean((y - y_hat) ** 2)` |
| $\mathbf w^\top\mathbf x = \sum_j w_j x_j$ | `w @ x` |
| $\hat{\mathbf y} = X\mathbf w + b$ | `X @ w + b` |
| $\lVert\mathbf w\rVert_2^2$ | `w @ w` hoặc `np.sum(w ** 2)` |
| $\lVert\mathbf w\rVert_1$ | `np.abs(w).sum()` |
| $\prod_{i} p_i$ | `np.prod(p)` |
| $\arg\max_k p_k$ | `p.argmax()` |
| $\mathbb{1}[y = \hat y]$ | `(y == y_hat)` |
| $\frac{1}{n}\sum_i \mathbb{1}[y^{(i)} = \hat y^{(i)}]$ (accuracy) | `np.mean(y == y_hat)` |
| $\mu_j = \frac{1}{n}\sum_i x^{(i)}_j$ | `X.mean(axis=0)` |
| $\sigma(z) = \frac{1}{1+e^{-z}}$ | `1 / (1 + np.exp(-z))` |

```python
import numpy as np

rng = np.random.default_rng(0)
n, d = 6, 3
X = rng.normal(size=(n, d))
w, b = np.array([0.5, -1.0, 2.0]), 0.1
y = rng.integers(0, 2, n)

z = X @ w + b                                  # z^{(i)} = w^T x^{(i)} + b
p = 1 / (1 + np.exp(-z))                       # σ(z)
y_hat = (p >= 0.5).astype(int)                 # ŷ = 1[σ(z) ≥ 0.5]
accuracy = np.mean(y == y_hat)                 # (1/n) Σ 1[y = ŷ]
log_loss = -np.mean(y * np.log(p) + (1 - y) * np.log(1 - p))
l2_penalty = w @ w                             # ||w||²

print("z     =", z.round(3))
print("p     =", p.round(3))
print("ŷ     =", y_hat, "| y =", y)
print(f"accuracy = {accuracy:.3f}, log-loss = {log_loss:.3f}, ||w||² = {l2_penalty:.2f}")
```

## 4. Đọc thử một công thức "đáng sợ"

Hàm mất mát của **Ridge Regression** (chương 06):

$$J(\mathbf w, b) = \frac{1}{2n}\sum_{i=1}^{n}\left(\mathbf w^\top\mathbf x^{(i)} + b - y^{(i)}\right)^2 + \frac{\lambda}{2}\lVert\mathbf w\rVert_2^2$$

Đọc từ trong ra ngoài:
1. $\mathbf w^\top\mathbf x^{(i)} + b$ — dự đoán cho mẫu $i$;
2. trừ $y^{(i)}$, bình phương — sai số bình phương của mẫu $i$;
3. $\frac{1}{2n}\sum_i$ — trung bình (nhân ½ cho gọn đạo hàm);
4. $\frac{\lambda}{2}\lVert\mathbf w\rVert_2^2$ — phạt trọng số lớn.

```python
lam = 0.1
y_reg = rng.normal(size=n)
J = np.sum((X @ w + b - y_reg) ** 2) / (2 * n) + lam / 2 * (w @ w)
print(f"J = {J:.4f}")
```
