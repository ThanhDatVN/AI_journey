## 1. Notebook là gì?

Jupyter Notebook là tài liệu gồm các **cell**:

- **Code cell**: chứa code Python, chạy bằng `Shift + Enter`, output hiện ngay bên dưới.
- **Markdown cell**: chứa văn bản, công thức, hình ảnh.

Code được chạy bởi một **kernel** (một tiến trình Python chạy nền). Mọi biến bạn tạo ra sống trong kernel cho tới khi bạn *restart* nó.

### Hai chế độ và phím tắt quan trọng

| Chế độ | Vào bằng | Phím tắt hay dùng |
|---|---|---|
| **Command mode** (viền xanh dương) | `Esc` | `A`/`B` thêm cell trên/dưới · `DD` xoá cell · `M` → Markdown · `Y` → Code · `Z` hoàn tác xoá |
| **Edit mode** (viền xanh lá) | `Enter` | `Shift+Enter` chạy & sang cell sau · `Ctrl+Enter` chạy tại chỗ · `Tab` gợi ý code · `Shift+Tab` xem docstring |

> 💡 Trên Google Colab phím tắt có tiền tố `Ctrl + M` (ví dụ `Ctrl+M B` thêm cell bên dưới).

## 2. Bẫy thứ tự thực thi (hidden state)

Notebook cho phép chạy cell **theo bất kỳ thứ tự nào**. Điều này rất tiện nhưng cũng rất nguy hiểm: kết quả bạn thấy có thể phụ thuộc vào một cell đã bị sửa hoặc xoá.

Thử nghiệm: chạy 3 cell dưới đây theo thứ tự 1 → 2 → 3, sau đó sửa cell 1 thành `learning_rate = 0.5` nhưng **chỉ chạy lại cell 3**.

````python
# Cell 1
learning_rate = 0.1
````

````python
# Cell 2
step = learning_rate * 10
````

````python
# Cell 3
print("step =", step)  # vẫn là 1.0 dù bạn đã đổi learning_rate → hidden state!
````

## 3. Magic commands

Dòng bắt đầu bằng `%` là **line magic**, `%%` là **cell magic** (áp dụng cho cả cell). Một số magic hay dùng trong ML:

| Magic | Tác dụng |
|---|---|
| `%timeit expr` | đo thời gian chạy (lặp nhiều lần, rất chính xác) |
| `%time expr` | đo thời gian một lần chạy |
| `%matplotlib inline` | hiển thị hình ngay trong notebook (mặc định ở bản mới) |
| `%load_ext autoreload` + `%autoreload 2` | tự nạp lại module `.py` khi bạn sửa — cực hữu ích khi phát triển `src/` |
| `%pwd`, `%cd` | xem / đổi thư mục hiện tại |
| `!pip install ...` | chạy lệnh shell (dấu `!`) |

````python
import numpy as np

x = np.random.rand(100_000)
%timeit sum(x)        # vòng lặp Python
%timeit np.sum(x)     # vectorized — nhanh hơn hàng chục lần (xem chương 01)
````

## 5. Công thức toán với LaTeX

Đây là kỹ năng bạn sẽ dùng **liên tục** để ghi chú công thức.

- Công thức *inline*: đặt giữa `$...$` → giả thuyết hồi quy tuyến tính $h_\theta(x) = \theta_0 + \theta_1 x$.
- Công thức *khối*: đặt giữa `$$...$$`:

$$
J(\theta) = \frac{1}{2m}\sum_{i=1}^{m}\left(h_\theta(x^{(i)}) - y^{(i)}\right)^2
$$

| Ý nghĩa | LaTeX | Hiển thị |
|---|---|---|
| chỉ số dưới / trên | `x_i`, `x^{(i)}` | $x_i$, $x^{(i)}$ |
| phân số, căn | `\frac{a}{b}`, `\sqrt{x}` | $\frac{a}{b}$, $\sqrt{x}$ |
| tổng, tích | `\sum_{i=1}^{n}`, `\prod_i` | $\sum_{i=1}^{n}$, $\prod_i$ |
| đạo hàm riêng | `\frac{\partial J}{\partial \theta_j}` | $\frac{\partial J}{\partial \theta_j}$ |
| vector, ma trận | `\mathbf{x}`, `X^\top` | $\mathbf{x}$, $X^\top$ |
| chuẩn | `\lVert w \rVert_2^2` | $\lVert w \rVert_2^2$ |
| chữ Hy Lạp | `\alpha, \beta, \lambda, \sigma, \mu` | $\alpha, \beta, \lambda, \sigma, \mu$ |
| argmin | `\arg\min_\theta` | $\arg\min_\theta$ |
| kỳ vọng | `\mathbb{E}[X]` | $\mathbb{E}[X]$ |

Ví dụ hàm sigmoid và cross-entropy (sẽ gặp ở chương 07):

$$
\sigma(z) = \frac{1}{1 + e^{-z}}, \qquad
L = -\frac{1}{m}\sum_{i=1}^{m}\Big[y^{(i)}\log \hat y^{(i)} + (1-y^{(i)})\log(1-\hat y^{(i)})\Big]
$$

## 6. Hiển thị phong phú từ code

Code cũng có thể sinh ra Markdown/LaTeX/HTML — tiện để in kết quả đẹp.

````python
from IPython.display import HTML, Latex, Markdown, display

theta = np.array([1.5, -0.7])
display(Markdown(f"**Tham số học được:** $\\theta_0 = {theta[0]:.2f}$, $\\theta_1 = {theta[1]:.2f}$"))
display(Latex(r"$$\hat{y} = \theta_0 + \theta_1 x$$"))
display(HTML("<span style='color:#16a34a'>✔ Hiển thị HTML được luôn</span>"))
````

## 7. Notebook vs file `.py` — khi nào dùng gì?

| | Notebook `.ipynb` | Script / module `.py` |

| Phù hợp | khám phá dữ liệu, thử nghiệm, học, báo cáo | code tái sử dụng, pipeline, production, test |
| Điểm yếu | hidden state, khó review diff, khó test | không có output trực quan |
