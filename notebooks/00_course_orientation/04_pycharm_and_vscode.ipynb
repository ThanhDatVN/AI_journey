## IV. VS Code *(bổ sung)*

VS Code là editor miễn phí, nhẹ, hỗ trợ Python và **Jupyter notebook trực tiếp** — rất phù hợp với khoá học này.

### Thử debug: tìm lỗi trong hàm dưới đây

Hàm sau được viết để tính trung bình của các số dương, nhưng kết quả sai. Hãy dùng `print` hoặc debugger (đặt breakpoint trong cell) để tìm lỗi.

````python
def mean_of_positive(values):
    total, count = 0, 0
    for v in values:
        if v > 0:
            total += v
        count += 1  # 🐛 đặt breakpoint ở đây và quan sát `count`
    return total / count


print(mean_of_positive([4, -2, 6, -8]))  # kỳ vọng 5.0
````

<details><summary>💡 Gợi ý</summary>

`count += 1` nằm **ngoài** khối `if` nên đếm cả số âm. Thụt lề nó vào trong `if`.
</details>

Một công cụ debug rất hữu ích ngay trong notebook là magic `%debug`: sau khi một cell bị lỗi, chạy `%debug` ở cell kế tiếp để mở debugger tại đúng vị trí lỗi (gõ `p tên_biến` để in, `q` để thoát).
