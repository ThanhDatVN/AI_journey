## 1. Các kỹ năng Search Google nâng cao

| Cú pháp | Tác dụng | Ví dụ |
|---|---|---|
| `"..."` | tìm **chính xác** cụm từ | `"ValueError: Input contains NaN"` |
| `site:` | chỉ tìm trong một website | `site:scikit-learn.org StandardScaler` |
| `intitle:` | từ khoá phải có trong tiêu đề trang | `intitle:"gradient boosting" tutorial` |

## 3. Đọc thông báo lỗi Python *(bổ sung)*

Kỹ năng quan trọng nhất khi lập trình ML: **đọc traceback từ dưới lên**.

- **Dòng cuối** = loại lỗi + thông điệp → thường đã nói rõ vấn đề.
- **Các dòng phía trên** = chuỗi lời gọi hàm; tìm dòng đầu tiên thuộc **code của bạn** (không phải thư viện).

Chạy cell sau và đọc lỗi:

```python
import numpy as np

X = np.random.rand(100, 3)   # 100 mẫu, 3 feature
w = np.random.rand(4)        # 🐛 sai số chiều
try:
    y_hat = X @ w
except ValueError as e:
    print("ValueError:", e)
```

Thông điệp *"matmul: Input operand 1 has a mismatch in its core dimension 0 … (size 4 is different from 3)"* cho biết: `X` có 3 cột nhưng `w` có 4 phần tử. **Lỗi shape** là lỗi phổ biến nhất trong ML — chương 01 có hẳn một bài về cách debug shape.

| Lỗi hay gặp | Nguyên nhân thường gặp |
|---|---|
| `ValueError: shapes ... not aligned` | sai chiều ma trận / quên reshape |
| `ValueError: Input contains NaN` | dữ liệu có missing value chưa xử lý |
| `KeyError: 'col'` | sai tên cột pandas (thừa khoảng trắng, sai hoa/thường) |
| `ModuleNotFoundError` | chưa cài thư viện / sai kernel |
| `ConvergenceWarning` | optimizer chưa hội tụ — thường do chưa scale dữ liệu |
