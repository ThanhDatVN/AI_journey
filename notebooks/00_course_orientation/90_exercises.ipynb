## L4 — Debug

**Bài 5.** Hàm dưới đây tính độ chính xác (accuracy) nhưng cho kết quả sai. Sửa hàm (đừng đổi tên hàm).

Đầu ra kỳ vọng: `accuracy([1, 0, 1, 1], [1, 1, 1, 0]) = 0.5`

```python
def accuracy(y_true, y_pred):
    correct = 0
    ### START CODE HERE ### (≈ 4 dòng)
    for t, p in zip(y_true, y_pred):
        if t == p:
            correct += 1
        return correct / len(y_true)   # 🐛 có lỗi ở dòng này
    ### END CODE HERE ###
```

```python
check_close(accuracy([1, 0, 1, 1], [1, 1, 1, 0]), 0.5, message="accuracy([1,0,1,1], [1,1,1,0])")
check_close(accuracy([0, 0, 0], [0, 0, 0]), 1.0, message="accuracy khi đúng hết")
```
