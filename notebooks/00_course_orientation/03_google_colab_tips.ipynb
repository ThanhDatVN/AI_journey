## 6. Dung lượng còn lại, bộ nhớ chính và bộ nhớ GPU

Cấu hình thay đổi theo thời gian; hãy tự kiểm tra bằng code. Đổi runtime sang GPU tại `Runtime → Change runtime type → GPU`.

````python
import os
import shutil

total, used, free = shutil.disk_usage("/")
print(f"Disk: {free / 1e9:.1f} GB trống / {total / 1e9:.1f} GB")
print(f"CPU cores: {os.cpu_count()}")

try:
    import torch

    if torch.cuda.is_available():
        props = torch.cuda.get_device_properties(0)
        print(f"GPU: {props.name} — {props.total_memory / 1e9:.1f} GB")
    else:
        print("Không có GPU (CPU là đủ cho phần lớn khoá học đến chương 26).")
except ImportError:
    print("Chưa cài PyTorch — chỉ cần từ chương 27.")
````

## 7. Lưu Model / checkpoint

Để model đang train không bị mất khi mất kết nối, hãy **lưu checkpoint** định kỳ và có khả năng **train tiếp** từ checkpoint. Ví dụ với PyTorch (sẽ học kỹ ở chương 27); thư mục lưu là Google Drive nếu đang ở Colab, ngược lại là thư mục tạm:

````python
import tempfile
from pathlib import Path

ckpt_dir = Path("/content/drive/MyDrive/ml-course-checkpoints") if IN_COLAB else Path(tempfile.gettempdir()) / "ml-course-checkpoints"
ckpt_dir.mkdir(parents=True, exist_ok=True)
ckpt_path = ckpt_dir / "demo.pt"

try:
    import torch
    from torch import nn

    model = nn.Linear(3, 1)
    optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
    start_epoch = 0
    if ckpt_path.exists():  # resume nếu đã có checkpoint
        state = torch.load(ckpt_path)
        model.load_state_dict(state["model"])
        optimizer.load_state_dict(state["optimizer"])
        start_epoch = state["epoch"] + 1
    X, y = torch.randn(64, 3), torch.randn(64, 1)
    for epoch in range(start_epoch, start_epoch + 3):
        loss = nn.functional.mse_loss(model(X), y)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        torch.save({"model": model.state_dict(), "optimizer": optimizer.state_dict(), "epoch": epoch}, ckpt_path)
        print(f"epoch {epoch}: loss={loss.item():.4f} → đã lưu {ckpt_path.name}")
except ImportError:
    print("Cần PyTorch để chạy ví dụ này.")
````

Với scikit-learn, dùng `joblib.dump(model, path)` / `joblib.load(path)` (chương 39).
