## 1. Triết lý: học Machine Learning thật sự là gì?

Rất nhiều khoá học dừng ở:

```python
model.fit(X, y)
model.predict(X_new)
```

Khoá học này đi xa hơn. Với **mỗi** model, bạn phải trả lời được:

> Nó hoạt động thế nào? Giả định gì? Khi nào dùng, khi nào **không** nên dùng? Điểm mạnh/yếu? Hyperparameter ảnh hưởng ra sao? **Vì sao nó thất bại?**

Kiến thức được xây theo **4 tầng**:

```text
LEVEL 1 — USE         Biết gọi model
        ↓
LEVEL 2 — UNDERSTAND  Biết model hoạt động thế nào (toán + code from scratch)
        ↓
LEVEL 3 — ANALYZE     Biết tại sao model tốt/xấu (diagnostics, failure cases)
        ↓
LEVEL 4 — DESIGN      Biết chọn và thiết kế giải pháp ML cho bài toán thực tế
```

Mục tiêu cuối cùng: nhìn vào một dataset và **tự suy luận** được, ví dụ:

> *Đây là supervised classification. Dataset mất cân bằng → không dùng accuracy, cần stratified CV. Baseline: Logistic Regression, sau đó thử tree ensemble. Feature X có nguy cơ leakage. Xác suất dự đoán cần calibration. Cuối cùng kiểm tra SHAP và distribution shift.*

## 4. Hệ thống bài tập L0 → L7

| Level | Nội dung | Ví dụ (Decision Tree) |
|---|---|---|
| L0 | Nhớ khái niệm | Entropy là gì? |
| L1 | Giải thích bằng lời | Vì sao Decision Tree không cần standardization? |
| L2 | Tính tay | class 0: 6, class 1: 4 → entropy = ? |
| L3 | Code | Tự implement `gini(y)` |
| L4 | Debug | Train acc 100%, test acc 71% — vấn đề gì? |
| L5 | Phân tích model | Ảnh hưởng của `max_depth`, `min_samples_leaf` |
| L6 | Bài toán thực tế | Xây pipeline dự đoán churn hoàn chỉnh |
| L7 | Advanced / research | So sánh các chiến lược pruning |

## 6. Kiểm tra môi trường

Cell dưới đây kiểm tra các thư viện cần cho từng phase. Thư viện **core** là bắt buộc; các thư viện khác chỉ cần khi học tới chương tương ứng (cài bằng `uv sync --extra <tên>` hoặc `pip install -e ".[<tên>]"`).

````python
import importlib.metadata
import platform

REQUIREMENTS = {                                                        # tên gói khi cài (pip/uv)
    "core (bắt buộc)": ["numpy", "pandas", "scipy", "matplotlib", "scikit-learn", "pyyaml"],
    "stats — ch.25": ["statsmodels"],
    "boosting — ch.12, 24": ["xgboost", "lightgbm", "catboost"],
    "tuning — ch.16": ["optuna"],
    "viz — ch.21": ["umap-learn"],
    "dl — ch.27–35": ["torch", "torchvision"],
    "tf — ch.27 bài 05 (tuỳ chọn)": ["tensorflow"],
    "xai — ch.36": ["shap"],
    "mlops — ch.39–41": ["mlflow", "fastapi", "onnxruntime"],
}

print(f"Python {platform.python_version()} — {platform.system()}")
missing_core = []
for group, packages in REQUIREMENTS.items():
    status = []
    for name in packages:
        try:
            status.append(f"✅ {name} {importlib.metadata.version(name)}")   # chỉ đọc metadata — không import thư viện nặng
        except importlib.metadata.PackageNotFoundError:
            status.append(f"⬜ {name}")
            if group.startswith("core"):
                missing_core.append(name)
    print(f"\n{group}:\n   " + "   ".join(status))

print("\n" + ("🎉 Môi trường core đã sẵn sàng!" if not missing_core else f"❗ Thiếu thư viện core: {missing_core}"))
````

## 7. Cách học hiệu quả (dựa trên khoa học về học tập)

1. **Active recall — tự nhớ lại thay vì đọc lại.** Sau mỗi bài, đóng notebook và viết ra 3 ý chính. Checkpoint chính là active recall.
2. **Code lại, đừng chỉ chạy lại.** Chạy `Shift + Enter` qua notebook *không* phải là học. Hãy gõ lại phần from scratch mà không nhìn.
3. **Kỹ thuật Feynman.** Giải thích một khái niệm (vd: bias-variance) bằng lời đơn giản cho người không học ML. Chỗ bạn ấp úng là chỗ bạn chưa hiểu.
4. **Spaced repetition.** Ôn lại `questions.md` của chương cũ sau 1 ngày, 1 tuần, 1 tháng.
5. **Interleaving.** Làm bài tập trộn nhiều chương (`exercises/` ở root) — buộc bạn phải *chọn* phương pháp, giống thực tế.
6. **Thay đổi và quan sát.** Mọi notebook đều có tham số — hãy đổi `learning_rate`, `max_depth`, … và *dự đoán trước* điều gì sẽ xảy ra rồi mới chạy.
