## Phần 3 — Coding Exercise

### Task c01 — Min-max scaler from scratch

Hoàn thành `MinMaxScalerManual`: `fit` học `min_`, `max_` theo cột; `transform` trả về $(x - min)/(max - min)$, cột hằng chia cho 1.

```python
class MinMaxScalerManual:
    def fit(self, X):
        X = np.asarray(X, dtype=float)
        ### START CODE HERE ### (≈ 3 dòng)
        self.min_ = X.min(axis=0)
        self.max_ = X.max(axis=0)
        self.range_ = np.where(self.max_ - self.min_ == 0, 1.0, self.max_ - self.min_)
        ### END CODE HERE ###
        return self

    def transform(self, X):
        ### START CODE HERE ### (≈ 1 dòng)
        Z = (np.asarray(X, dtype=float) - self.min_) / self.range_
        ### END CODE HERE ###
        return Z
```

```python
with cp.task("c01"):
    from sklearn.preprocessing import MinMaxScaler
    X = np.c_[np.arange(10.0), np.full(10, 3.0)]
    s = MinMaxScalerManual().fit(X)
    check_close(s.transform(X), MinMaxScaler().fit_transform(X), message="khớp sklearn")
    check_close(s.transform([[20.0, 3.0]]), [[20 / 9, 0.0]], message="giá trị ngoài khoảng train")
```

## Phần 4 — Mini Project

**Mini project — Pipeline dự đoán giá nhà (nối chương 03 + 04).** Với `house_prices_dirty.csv` (đã bỏ trùng lặp, chuẩn hoá `district`), target là **log** giá.
Mốc tham khảo (MAE trên log-giá, 5-fold CV): dự đoán bằng trung bình ≈ 0.43; pipeline cơ bản ≈ 0.22.
1. Sửa lỗi gõ diện tích (chương 03): nếu `area_m2 / bedrooms > 150` thì chia `area_m2` cho 10.
2. Thay `area_m2` bằng `log(area_m2)` (giá tăng theo diện tích nhưng không tuyến tính).
3. `Pipeline`: ColumnTransformer (số: median + indicator + scale; `district`: one-hot `handle_unknown="ignore"`; `has_garden`: passthrough) + `Ridge(alpha=1.0)`.
Đạt **MAE < 0.12** (biến `cv_mae`).

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

df = pd.read_csv("../../data/sample/house_prices_dirty.csv").drop_duplicates()
df["district"] = df["district"].str.strip().str.lower().str.replace("q.", "quận", regex=False).str.title()
X, y = df.drop(columns="price_billion_vnd"), np.log(df["price_billion_vnd"])
num = ["area_m2", "bedrooms", "bathrooms", "age_years", "distance_to_center_km"]
### START CODE HERE ### (≈ 11 dòng)
X = X.copy()
bad = X["area_m2"] / X["bedrooms"] > 150
X.loc[bad, "area_m2"] = X.loc[bad, "area_m2"] / 10
X["area_m2"] = np.log(X["area_m2"])
pre = ColumnTransformer([
    ("num", Pipeline([("imp", SimpleImputer(strategy="median", add_indicator=True)), ("sc", StandardScaler())]), num),
    ("cat", OneHotEncoder(handle_unknown="ignore"), ["district"]),
    ("bool", "passthrough", ["has_garden"]),
])
pipe = Pipeline([("pre", pre), ("model", Ridge(alpha=1.0))])
cv_mae = -cross_val_score(pipe, X, y, cv=5, scoring="neg_mean_absolute_error").mean()
### END CODE HERE ###
print("CV MAE:", cv_mae)
```

```python
with cp.task("project"):
    check_less(cv_mae, 0.12, "MAE log-giá < 0.12")
```
