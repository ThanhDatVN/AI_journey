### Task c01 — Ngưỡng chia tốt nhất theo Gini

Viết `gini(y)` (y nhãn 0/1) và `best_split(x, y)`: thử mọi ngưỡng là trung điểm giữa hai giá trị liên tiếp (sau khi sắp xếp, khác nhau)
của x; trả về `(threshold, weighted_gini)` nhỏ nhất, với weighted_gini = (n_trái·gini_trái + n_phải·gini_phải)/n.

```python
def gini(y):
    ### START CODE HERE ### (≈ 3 dòng)
    y = np.asarray(y)
    if len(y) == 0:
        return 0.0
    p = y.mean()
    return 1 - p ** 2 - (1 - p) ** 2
    ### END CODE HERE ###


def best_split(x, y):
    ### START CODE HERE ### (≈ 8 dòng)
    x, y = np.asarray(x, float), np.asarray(y)
    values = np.unique(x)
    best = (None, np.inf)
    for t in (values[:-1] + values[1:]) / 2:
        left, right = y[x <= t], y[x > t]
        score = (len(left) * gini(left) + len(right) * gini(right)) / len(y)
        if score < best[1]:
            best = (t, score)
    return best
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    check_close(gini([0, 0, 1, 1]), 0.5, message="gini cân bằng")
    check_close(gini([1, 1, 1]), 0.0, message="gini thuần khiết")
    t, s = best_split([1, 2, 3, 10, 11, 12], [0, 0, 0, 1, 1, 1])
    check_close([t, s], [6.5, 0.0], message="tách hoàn hảo")
    t2, s2 = best_split([1, 2, 3, 4], [0, 1, 0, 1])
    check_close(s2, min((1 * 0 + 3 * gini([1, 0, 1])) / 4, (2 * 0.5 + 2 * 0.5) / 4, (3 * gini([0, 1, 0]) + 0) / 4), message="ngưỡng tốt nhất khi không tách hoàn hảo")
```

### Task c02 — kNN từ đầu

Viết `knn_predict(X_train, y_train, X_new, k)`: khoảng cách Euclid, bỏ phiếu đa số trong k láng giềng gần nhất
(hoà thì chọn nhãn nhỏ nhất — dùng `np.bincount(...).argmax()`).

```python
def knn_predict(X_train, y_train, X_new, k):
    ### START CODE HERE ### (≈ 4 dòng)
    X_train, X_new, y_train = np.asarray(X_train, float), np.asarray(X_new, float), np.asarray(y_train)
    d = ((X_new[:, None, :] - X_train[None, :, :]) ** 2).sum(-1)
    nearest = np.argsort(d, axis=1, kind="stable")[:, :k]
    return np.array([np.bincount(y_train[idx]).argmax() for idx in nearest])
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    Xtr = np.array([[0, 0], [0, 1], [1, 0], [5, 5], [5, 6], [6, 5]])
    ytr = np.array([0, 0, 0, 1, 1, 1])
    check_equal(list(knn_predict(Xtr, ytr, np.array([[0.2, 0.2], [5.5, 5.5]]), 3)), [0, 1], "hai cụm rõ ràng")
    from sklearn.neighbors import KNeighborsClassifier
    rng = np.random.default_rng(0)
    Xa, ya, Xb = rng.normal(size=(100, 2)), rng.integers(0, 2, 100), rng.normal(size=(30, 2))
    check_equal(list(knn_predict(Xa, ya, Xb, 5)), list(KNeighborsClassifier(5).fit(Xa, ya).predict(Xb)), "khớp scikit-learn")
```

khả năng giải thích lý do từ chối — ở tín dụng, nếu hiệu năng tương đương thì mô hình giải thích được thường được ưu tiên.

## Phần 5 — Mini Project

**Mini project — Dự đoán rời bỏ.** Với `customer_churn.csv` (chia 75/25 phân tầng, `random_state=1`): xây pipeline (chuẩn hoá cột số, one-hot cột
phân loại) và **chọn một trong hai** mô hình logistic hoặc random forest bằng **5-fold CV trên train** (ROC-AUC). Huấn luyện mô hình được chọn trên
toàn bộ train, đánh giá test. Lưu `cv_auc` (dict tên → AUC CV), `chosen` (tên), `test_auc`. Yêu cầu: AUC test > 0.80.

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import StratifiedKFold, cross_val_score, train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from ml_course.utils import repo_path

churn = pd.read_csv(repo_path("data", "sample", "customer_churn.csv"))
NUMERIC = ["tenure_months", "monthly_charges", "total_charges", "support_calls", "n_products"]
CATEGORICAL = ["contract", "internet_service", "payment_method", "senior"]
train, test = train_test_split(churn, test_size=0.25, random_state=1, stratify=churn.churn)
### START CODE HERE ### (≈ 10 dòng)
def pipeline(estimator):
    pre = ColumnTransformer([("num", StandardScaler(), NUMERIC), ("cat", OneHotEncoder(handle_unknown="ignore"), CATEGORICAL)])
    return make_pipeline(pre, estimator)


models = {"logistic": pipeline(LogisticRegression(max_iter=1000)),
          "forest": pipeline(RandomForestClassifier(300, min_samples_leaf=10, random_state=0, n_jobs=-1))}
cv = StratifiedKFold(5, shuffle=True, random_state=0)
cv_auc = {name: cross_val_score(m, train[NUMERIC + CATEGORICAL], train.churn, cv=cv, scoring="roc_auc").mean() for name, m in models.items()}
chosen = max(cv_auc, key=cv_auc.get)
final = models[chosen].fit(train[NUMERIC + CATEGORICAL], train.churn)
test_auc = roc_auc_score(test.churn, final.predict_proba(test[NUMERIC + CATEGORICAL])[:, 1])
### END CODE HERE ###
print(cv_auc, "→", chosen, f"| AUC test = {test_auc:.3f}")
```
