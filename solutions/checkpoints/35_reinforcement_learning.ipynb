## Phần 3 — Coding Exercise

### Task c01 — Return từ mỗi bước

Viết `discounted_returns(rewards, gamma)` trả về mảng G với G[t] = r[t] + γ·r[t+1] + γ²·r[t+2] + ... (tính ngược từ cuối, O(T)).

```python
def discounted_returns(rewards, gamma):
    ### START CODE HERE ### (≈ 5 dòng)
    G, running = np.zeros(len(rewards)), 0.0
    for t in reversed(range(len(rewards))):
        running = rewards[t] + gamma * running
        G[t] = running
    return G
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    check_close(discounted_returns([0, 0, 1], 0.9), [0.81, 0.9, 1.0], message="thưởng ở cuối")
    check_close(discounted_returns([1, 1, 1, 1], 1.0), [4, 3, 2, 1], message="γ = 1")
    check_close(discounted_returns([], 0.9).shape, (0,), message="lượt rỗng")
```

### Task c02 — Bellman backup

MDP nhỏ cho dưới dạng `P[s][a]` = danh sách (xác suất, s_tiếp, phần_thưởng, kết_thúc). Viết `bellman_backup(P, V, gamma)` trả về mảng Q
(n_states, n_actions) với Q[s, a] = Σ p·(r + γ·V[s'] nếu chưa kết thúc).

```python
def bellman_backup(P, V, gamma):
    ### START CODE HERE ### (≈ 3 dòng)
    n_s, n_a = len(P), len(P[0])
    return np.array([[sum(p * (r + gamma * V[s2] * (not done)) for p, s2, r, done in P[s][a]) for a in range(n_a)] for s in range(n_s)])
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    P = {0: {0: [(1.0, 1, 0.0, False)], 1: [(0.5, 0, 1.0, False), (0.5, 1, 0.0, True)]},
         1: {0: [(1.0, 1, 0.0, True)], 1: [(1.0, 0, 2.0, False)]}}
    Q = bellman_backup(P, np.array([1.0, 3.0]), 0.9)
    check_close(Q, [[2.7, 0.5 * (1 + 0.9)], [0.0, 2 + 0.9]], message="Q sau một backup")
    V = np.zeros(2)
    for _ in range(500):
        V = bellman_backup(P, V, 0.9).max(1)
    check_close(V, bellman_backup(P, V, 0.9).max(1), tol=1e-9, message="lặp tới điểm bất động (Bellman tối ưu)")
```

## Phần 5 — Mini Project

**Mini project — Q-learning trên băng trơn.** Môi trường `GridWorld("4x4", slip=2/3)` (FrozenLake trơn), γ = 0.99. Cài Q-learning dạng bảng
(ε-greedy giảm dần), **không** dùng `env.P`. Đánh giá chính sách tham lam cuối cùng trên 2 000 lượt (`env.reset(seed=50_000 + i)`), lưu tỉ lệ
thắng vào `win_rate`. Yêu cầu `win_rate` ≥ 0.74 (chính sách tối ưu đạt khoảng 0.81).

```python
from ml_course.envs import GridWorld

env = GridWorld("4x4", slip=2 / 3)
rng = np.random.default_rng(0)
### START CODE HERE ### (≈ 20 dòng)
Q = np.zeros((env.n_states, env.n_actions))
episodes = 20000
for ep in range(episodes):
    eps = max(0.05, 1 - ep / (episodes / 2))
    s, done = env.reset(seed=ep), False
    while not done:
        a = int(rng.integers(4)) if rng.random() < eps else int(rng.choice(np.flatnonzero(Q[s] == Q[s].max())))
        s2, r, done = env.step(a)
        target = r if done else r + 0.99 * Q[s2].max()
        Q[s, a] += 0.1 * (target - Q[s, a])
        s = s2
policy = Q.argmax(1)
wins = 0
for i in range(2000):
    s, done = env.reset(seed=50_000 + i), False
    while not done:
        s, r, done = env.step(policy[s])
    wins += r
win_rate = wins / 2000
### END CODE HERE ###
print(f"tỉ lệ thắng = {win_rate:.3f}")
```

```python
with cp.task("project"):
    check_greater(win_rate, 0.74, "tỉ lệ thắng ≥ 0.74")
```
