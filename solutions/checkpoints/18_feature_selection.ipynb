## Phần 3 — Coding Exercise

### Task c01 — Soft-thresholding

Viết `soft_threshold(beta, lam)` = sign(β)·max(|β| − λ, 0) (vector hoá). Đây là nghiệm Lasso khi feature trực chuẩn.

```python
def soft_threshold(beta, lam):
    ### START CODE HERE ### (≈ 1 dòng)
    out = np.sign(beta) * np.maximum(np.abs(beta) - lam, 0)
    ### END CODE HERE ###
    return out
```

```python
with cp.task("c01"):
    check_close(soft_threshold(np.array([1.5, -0.3, -2.0, 0.4]), 0.4), [1.1, 0.0, -1.6, 0.0], message="soft threshold")
    check_true(np.all(soft_threshold(np.array([0.1, -0.2]), 1.0) == 0), "hệ số nhỏ → 0")
```

### Task c02 — Chọn top-k theo điểm

Viết `select_top_k(scores, names, k)` trả về list tên k feature có điểm cao nhất, theo thứ tự điểm giảm dần.

```python
def select_top_k(scores, names, k):
    ### START CODE HERE ### (≈ 2 dòng)
    order = np.argsort(-np.asarray(scores), kind="stable")[:k]
    top = [names[i] for i in order]
    ### END CODE HERE ###
    return top
```

```python
with cp.task("c02"):
    check_equal(select_top_k([0.1, 0.9, 0.5, 0.7], ["a", "b", "c", "d"], 2), ["b", "d"], "top-2")
    check_equal(select_top_k([3, 1, 2], ["x", "y", "z"], 3), ["x", "z", "y"], "đủ k")
```

## Phần 5 — Mini Project

**Mini project — Chọn feature cho KNN.** Dùng `make_friedman1(n_samples=1000, n_features=30, noise=1.0, random_state=7)`.
Chia train/test 75/25 (`random_state=0`). Xây Pipeline: StandardScaler → **bước chọn feature** (tuỳ chọn: SelectKBest với mutual_info,
SelectFromModel, SFS...) → `KNeighborsRegressor(10)`; chọn tham số bằng CV trên train. Đạt **R² test ≥ 0.80** (biến `test_r2`)
— so với KNN trên 30 feature (`baseline_r2`).

```python
from functools import partial
from sklearn.datasets import make_friedman1
from sklearn.feature_selection import SelectKBest, mutual_info_regression
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X, y = make_friedman1(n_samples=1000, n_features=30, noise=1.0, random_state=7)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, random_state=0)
### START CODE HERE ### (≈ 8 dòng)
baseline_r2 = make_pipeline(StandardScaler(), KNeighborsRegressor(10)).fit(X_tr, y_tr).score(X_te, y_te)
mi = partial(mutual_info_regression, n_neighbors=10, random_state=0)
grid = GridSearchCV(make_pipeline(StandardScaler(), SelectKBest(mi), KNeighborsRegressor(10)),
                    {"selectkbest__k": [3, 4, 5, 6, 8, 10]}, cv=5, scoring="r2").fit(X_tr, y_tr)
test_r2 = grid.score(X_te, y_te)
### END CODE HERE ###
print(f"R² test: KNN 30 feature = {baseline_r2:.3f} | KNN + chọn feature = {test_r2:.3f} (k = {grid.best_params_['selectkbest__k']})")
```

```python
with cp.task("project"):
    check_greater(test_r2, 0.80, "R² test ≥ 0.80")
```
