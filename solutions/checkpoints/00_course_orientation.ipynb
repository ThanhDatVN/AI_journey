### Task c01 — Kiểm tra môi trường

Viết hàm `library_versions(names)` nhận danh sách tên module và trả về **dict** `{tên: phiên_bản}`.
Module chưa cài thì giá trị là `None`. Gợi ý: `importlib.import_module`, thuộc tính `__version__`, bắt `ImportError`.

```python
import importlib

def library_versions(names):
    versions = {}
    ### START CODE HERE ### (≈ 5 dòng)
    for name in names:
        try:
            versions[name] = importlib.import_module(name).__version__
        except ImportError:
            versions[name] = None
    ### END CODE HERE ###
    return versions

library_versions(["numpy", "pandas", "sklearn", "matplotlib"])
```

```python
with cp.task("c01"):
    v = library_versions(["numpy", "sklearn", "this_module_does_not_exist"])
    check_equal(type(v), dict, "trả về dict")
    check_true(v["numpy"] is not None and v["sklearn"] is not None, "đã cài numpy và scikit-learn")
    check_equal(v["this_module_does_not_exist"], None, "module chưa cài → None")
```
