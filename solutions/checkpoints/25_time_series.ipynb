## Phần 3 — Coding Exercise

### Task c01 — Seasonal naive và MASE

Viết `seasonal_naive(train, h, m)` (lặp lại m giá trị cuối) và `mase(y_true, y_pred, y_train, m)`.

```python
def seasonal_naive(train, h, m):
    ### START CODE HERE ### (≈ 1 dòng)
    return np.tile(np.asarray(train, float)[-m:], h // m + 1)[:h]
    ### END CODE HERE ###

def mase(y_true, y_pred, y_train, m):
    ### START CODE HERE ### (≈ 2 dòng)
    y_train = np.asarray(y_train, float)
    return np.mean(np.abs(np.asarray(y_true) - np.asarray(y_pred))) / np.mean(np.abs(y_train[m:] - y_train[:-m]))
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    check_close(seasonal_naive([1, 2, 3, 4, 5, 6, 7, 8], 5, 3), [6, 7, 8, 6, 7], message="seasonal naive")
    check_close(mase([14, 16], [13, 13], [10, 12, 14, 11, 13, 15], 1), 2 / 2.2, message="MASE")
    y = np.tile([1.0, 2, 3, 4, 5, 6, 7], 10)
    check_close(mase(y[-7:], seasonal_naive(y[:-7], 7, 7), y[:-7] + np.arange(63) * 0.1, 7), 0.0, message="chuỗi tuần hoàn hoàn hảo → MASE 0")
```

### Task c02 — Feature trễ không rò rỉ

Viết `lagged_rolling_mean(series, window, horizon)` = trung bình `window` giá trị mới nhất **đã biết** khi dự báo
`horizon` bước tới (pandas Series cùng index, NaN ở đầu).

```python
def lagged_rolling_mean(series, window, horizon):
    ### START CODE HERE ### (≈ 1 dòng)
    return series.shift(horizon).rolling(window).mean()
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    import pandas as pd
    s = pd.Series(np.arange(30, dtype=float))
    r = lagged_rolling_mean(s, 7, 1)
    check_close(r.iloc[7], np.mean(np.arange(0, 7)), message="ngày 7 dùng ngày 0..6")
    r14 = lagged_rolling_mean(s, 7, 14)
    check_close(r14.iloc[20], np.mean(np.arange(0, 7)), message="tầm 14: ngày 20 dùng ngày 0..6")
    check_true(r14.iloc[:20].isna().all(), "không đủ dữ liệu → NaN")
```

## Phần 5 — Mini Project

**Mini project — Dự báo 28 ngày cho một cửa hàng mới.** Dữ liệu `make_sales_series(n_days=900, start="2022-06-01", seed=11)`.
Dự báo 28 ngày cuối chỉ dùng dữ liệu trước đó (và lịch khuyến mãi/lễ của 28 ngày cuối — đã biết trước). Yêu cầu `holdout_mase` < 0.75
(thang đo: seasonal naive m = 7 trên phần dữ liệu train).

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from ml_course.datasets import make_sales_series

data = make_sales_series(n_days=900, start="2022-06-01", seed=11).set_index("date").asfreq("D")
train, test = data.iloc[:-28], data.iloc[-28:]
### START CODE HERE ### (≈ 14 dòng)
def feats(frame):
    F = pd.DataFrame(index=frame.index)
    F["t"] = (frame.index - data.index[0]).days / 365.25
    for d in range(7):
        F[f"dow_{d}"] = (frame.index.dayofweek == d).astype(float)
    doy = frame.index.dayofyear.to_numpy()
    for k in [1, 2, 3]:
        F[f"sin_{k}"] = np.sin(2 * np.pi * k * doy / 365.25)
        F[f"cos_{k}"] = np.cos(2 * np.pi * k * doy / 365.25)
    F["tet"] = np.exp(-0.5 * ((doy - 30) / 7) ** 2)
    F["promo"], F["holiday"] = frame["promo"].to_numpy(float), frame["is_holiday"].to_numpy(float)
    return F
model = LinearRegression().fit(feats(train), np.log(train["sales"]))
forecast = np.exp(model.predict(feats(test)))
scale = np.mean(np.abs(train["sales"].to_numpy()[7:] - train["sales"].to_numpy()[:-7]))
holdout_mase = np.mean(np.abs(test["sales"].to_numpy() - forecast)) / scale
### END CODE HERE ###
print(f"holdout MASE = {holdout_mase:.3f}")
```

```python
with cp.task("project"):
    check_less(holdout_mase, 0.75, "MASE holdout < 0.75")
```
