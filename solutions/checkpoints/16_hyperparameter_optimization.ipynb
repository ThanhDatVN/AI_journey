## Phần 3 — Coding Exercise

### Task c01 — Expected Improvement

Viết `expected_improvement(mu, sigma, best, xi=0.0)` cho bài toán **tối đa hoá** (mu, sigma là mảng). Chặn sigma ≥ 1e-12.

```python
from scipy.stats import norm


def expected_improvement(mu, sigma, best, xi=0.0):
    ### START CODE HERE ### (≈ 3 dòng)
    sigma = np.maximum(sigma, 1e-12)
    z = (mu - best - xi) / sigma
    ei = (mu - best - xi) * norm.cdf(z) + sigma * norm.pdf(z)
    ### END CODE HERE ###
    return ei
```

```python
with cp.task("c01"):
    from scipy.stats import norm
    ei = expected_improvement(np.array([0.9, 0.8, 0.85]), np.array([0.1, 0.3, 0.0]), best=0.85)
    check_close(ei[0], 0.05 * norm.cdf(0.5) + 0.1 * norm.pdf(0.5), message="điểm 1")
    check_true(ei[1] > ei[0], "bất định lớn → EI lớn hơn")
    check_close(ei[2], 0.0, tol=1e-9, message="σ = 0 và μ = best → EI = 0")
```

### Task c02 — Một vòng successive halving

Viết `halving_step(configs, scores, eta)` trả về list các cấu hình được giữ: `max(1, len(configs) // eta)` cấu hình có điểm cao nhất,
theo thứ tự điểm giảm dần.

```python
def halving_step(configs, scores, eta=3):
    ### START CODE HERE ### (≈ 2 dòng)
    keep = np.argsort(scores)[::-1][: max(1, len(configs) // eta)]
    kept = [configs[i] for i in keep]
    ### END CODE HERE ###
    return kept
```

```python
with cp.task("c02"):
    cfgs = list("abcdefghi")
    sc = [0.1, 0.9, 0.3, 0.8, 0.2, 0.7, 0.4, 0.6, 0.5]
    check_equal(halving_step(cfgs, sc, 3), ["b", "d", "f"], "giữ 3 cấu hình tốt nhất")
    check_equal(halving_step(["x", "y"], [0.2, 0.1], 3), ["x"], "luôn giữ ít nhất 1")
```

## Phần 5 — Mini Project

**Mini project — Tuning có kỷ luật.** Đọc `customer_churn.csv` (`repo_path("data", "sample", "customer_churn.csv")`), one-hot.
Tách test 25% (`random_state=0`, `stratify`). Trên phần dev: tune `HistGradientBoostingClassifier` bằng `RandomizedSearchCV`
(≥ 15 cấu hình, 3-fold, `roc_auc`) với `learning_rate` **log-uniform**, `max_leaf_nodes`, `min_samples_leaf`, `l2_regularization` (log-uniform),
early stopping bật. Báo cáo `baseline_auc` (mặc định) và `tuned_auc` trên test. Yêu cầu `tuned_auc ≥ 0.80`.

```python
import pandas as pd
from scipy.stats import loguniform, randint
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import RandomizedSearchCV, train_test_split
from ml_course.utils import repo_path

df = pd.read_csv(repo_path("data", "sample", "customer_churn.csv"))
X = pd.get_dummies(df.drop(columns=["customer_id", "churn"]), dtype=float)
y = df["churn"].to_numpy()
### START CODE HERE ### (≈ 10 dòng)
X_dev, X_te, y_dev, y_te = train_test_split(X, y, test_size=0.25, random_state=0, stratify=y)
baseline_auc = roc_auc_score(y_te, HistGradientBoostingClassifier(random_state=0).fit(X_dev, y_dev).predict_proba(X_te)[:, 1])
space = {"learning_rate": loguniform(0.005, 0.3), "max_leaf_nodes": randint(3, 48), "min_samples_leaf": randint(10, 300),
         "l2_regularization": loguniform(1e-3, 10)}
search = RandomizedSearchCV(HistGradientBoostingClassifier(max_iter=1000, early_stopping=True, n_iter_no_change=20, random_state=0),
                            space, n_iter=15, cv=3, scoring="roc_auc", random_state=0).fit(X_dev, y_dev)
tuned_auc = roc_auc_score(y_te, search.predict_proba(X_te)[:, 1])
### END CODE HERE ###
print(f"ROC-AUC test: mặc định = {baseline_auc:.4f} | đã tune = {tuned_auc:.4f}")
```

```python
with cp.task("project"):
    check_greater(tuned_auc, 0.80, "ROC-AUC test (đã tune) ≥ 0.80")
```
