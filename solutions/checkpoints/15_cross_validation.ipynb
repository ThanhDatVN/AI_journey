## Phần 3 — Coding Exercise

### Task c01 — K-Fold indices

Viết `kfold_splits(n, k, seed)` trả về list `(train_idx, val_idx)`: xáo trộn `np.arange(n)` bằng `np.random.default_rng(seed).permutation`,
chia bằng `np.array_split` thành k phần, fold i là validation.

```python
def kfold_splits(n, k=5, seed=0):
    ### START CODE HERE ### (≈ 5 dòng)
    idx = np.random.default_rng(seed).permutation(n)
    folds = np.array_split(idx, k)
    out = []
    for i in range(k):
        out.append((np.concatenate([folds[j] for j in range(k) if j != i]), folds[i]))
    ### END CODE HERE ###
    return out
```

```python
with cp.task("c01"):
    splits = kfold_splits(53, 5)
    check_equal([len(v) for _, v in splits], [11, 11, 11, 10, 10], "kích thước fold")
    check_equal(sorted(np.concatenate([v for _, v in splits]).tolist()), list(range(53)), "mỗi mẫu validation đúng 1 lần")
    check_true(all(len(np.intersect1d(t, v)) == 0 for t, v in splits), "train ∩ val = ∅")
```

### Task c02 — Cross-validation thủ công

Viết `manual_cv(make_model, X, y, splits, scorer)` trả về mảng điểm số: với mỗi `(train, val)` trong `splits`, tạo model **mới** bằng
`make_model()`, fit trên train, tính `scorer(y[val], model.predict(X[val]))`.

```python
def manual_cv(make_model, X, y, splits, scorer):
    scores = []
    ### START CODE HERE ### (≈ 3 dòng)
    for tr, va in splits:
        model = make_model().fit(X[tr], y[tr])
        scores.append(scorer(y[va], model.predict(X[va])))
    ### END CODE HERE ###
    return np.array(scores)
```

```python
with cp.task("c02"):
    from sklearn.linear_model import LinearRegression
    from sklearn.metrics import mean_squared_error
    from sklearn.model_selection import KFold, cross_val_score
    rng = np.random.default_rng(0)
    X = rng.normal(size=(100, 3))
    y = X @ np.array([1.0, -2.0, 0.5]) + rng.normal(0, 0.1, 100)
    cv = KFold(5, shuffle=True, random_state=0)
    ours = manual_cv(LinearRegression, X, y, list(cv.split(X)), mean_squared_error)
    ref = -cross_val_score(LinearRegression(), X, y, cv=cv, scoring="neg_mean_squared_error")
    check_close(ours, ref, message="khớp cross_val_score")
```

## Phần 5 — Mini Project

**Mini project — Đánh giá trung thực với dữ liệu nhóm.** Dữ liệu tổng hợp: 80 cửa hàng × 30 ngày, feature là đặc điểm cửa hàng
(cố định) + ngày; target là doanh thu. Mục tiêu: dự đoán cho **cửa hàng mới**. (1) Tính MAE bằng KFold ngẫu nhiên (`mae_random`) và
GroupKFold theo cửa hàng (`mae_group`) cho `RandomForestRegressor(n_estimators=100, random_state=0)`; (2) tính MAE thật trên 40 cửa hàng mới
(`mae_new`). Yêu cầu: `mae_group` gần `mae_new` hơn `mae_random`.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import GroupKFold, KFold, cross_val_score

def make_stores(n_stores, seed):
    r = np.random.default_rng(seed)
    traits = r.normal(size=(n_stores, 6))                    # đặc điểm cửa hàng (dân số, diện tích, ...) — chỉ 2 cái có ích
    store_effect = r.normal(0, 3, n_stores)                  # hiệu ứng riêng không quan sát được
    g = np.repeat(np.arange(n_stores), 30)
    day = np.tile(np.arange(30), n_stores)
    X = np.c_[traits[g], day % 7]
    y = 20 + 2 * traits[g, 0] - 1.5 * traits[g, 1] + store_effect[g] + 1.5 * (day % 7 >= 5) + r.normal(0, 1, len(g))
    return X, y, g

X, y, groups = make_stores(80, 0)
X_new, y_new, _ = make_stores(40, 1)
### START CODE HERE ### (≈ 5 dòng)
model = RandomForestRegressor(n_estimators=100, random_state=0)
mae_random = -cross_val_score(model, X, y, cv=KFold(5, shuffle=True, random_state=0), scoring="neg_mean_absolute_error").mean()
mae_group = -cross_val_score(model, X, y, cv=GroupKFold(5), groups=groups, scoring="neg_mean_absolute_error").mean()
mae_new = mean_absolute_error(y_new, model.fit(X, y).predict(X_new))
### END CODE HERE ###
print(f"MAE: KFold ngẫu nhiên = {mae_random:.2f} | GroupKFold = {mae_group:.2f} | cửa hàng mới = {mae_new:.2f}")
```

```python
with cp.task("project"):
    check_less(abs(mae_group - mae_new), abs(mae_random - mae_new), "GroupKFold gần thực tế hơn")
    check_less(mae_random, mae_new, "KFold ngẫu nhiên lạc quan")
```
