## Phần 3 — Coding Exercise

### Task c01 — KL của Gaussian chéo với prior chuẩn

Viết `kl_diag_gaussian(mu, logvar)` (NumPy, mảng (B, k)) trả về KL(N(μ, diag σ²) ‖ N(0, I)) cho từng mẫu — mảng (B,).

```python
def kl_diag_gaussian(mu, logvar):
    ### START CODE HERE ### (≈ 1 dòng)
    return -0.5 * np.sum(1 + logvar - mu**2 - np.exp(logvar), axis=1)
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    check_close(kl_diag_gaussian(np.zeros((2, 3)), np.zeros((2, 3))), [0.0, 0.0], message="trùng prior → 0")
    check_close(kl_diag_gaussian(np.array([[2.0]]), np.array([[0.0]])), [2.0], message="μ = 2, σ = 1")
    check_close(kl_diag_gaussian(np.array([[1.0]]), np.array([[np.log(0.25)]])), [0.5 * (0.25 + 1 - 1 - np.log(0.25))], message="μ = 1, σ = 0.5")
    check_true(np.all(kl_diag_gaussian(np.random.default_rng(0).normal(size=(50, 4)), np.random.default_rng(1).normal(size=(50, 4))) >= 0), "KL ≥ 0")
```

### Task c02 — Loss NT-Xent

Viết `nt_xent_np(z1, z2, tau)` (NumPy): z1, z2 shape (N, d) là hai góc nhìn; chuẩn hoá L2, ma trận cosine của 2N vector chia τ,
bỏ đường chéo; cặp dương của i là i ± N; trả về trung bình cross-entropy trên 2N vector. Ổn định số (trừ max).

```python
def nt_xent_np(z1, z2, tau=0.2):
    ### START CODE HERE ### (≈ 8 dòng)
    z = np.concatenate([z1, z2])
    z = z / np.linalg.norm(z, axis=1, keepdims=True)
    sim = z @ z.T / tau
    np.fill_diagonal(sim, -np.inf)
    n = len(z1)
    pos = np.r_[np.arange(n, 2 * n), np.arange(0, n)]
    sim = sim - sim.max(axis=1, keepdims=True)
    log_prob = sim[np.arange(2 * n), pos] - np.log(np.exp(sim).sum(axis=1))
    return -log_prob.mean()
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    e = np.array([[1.0, 0.0], [0.0, 1.0]])
    check_close(nt_xent_np(e, e, tau=1.0), np.log(1 + 2 / np.e), message="ví dụ tính tay")
    r = np.random.default_rng(0)
    same = r.normal(size=(8, 5))
    check_less(nt_xent_np(same, same, 0.1), nt_xent_np(same, r.normal(size=(8, 5)), 0.1), "góc nhìn khớp → loss nhỏ hơn")
    check_true(np.isfinite(nt_xent_np(same * 100, same, 0.01)), "ổn định số")
```

## Phần 5 — Mini Project

**Mini project — SimCLR với ít nhãn.** Train SimCLR (encoder conv nhỏ) trên 10 000 ảnh **không nhãn** `make_shapes(10000, seed=0)`
(800 bước, batch 256), rồi linear probe (LogisticRegression trên đặc trưng encoder) với **100 ảnh có nhãn** đầu tiên của tập đó.
Đánh giá trên `make_shapes(2000, seed=1)`; lưu accuracy vào `acc`. Yêu cầu `acc` ≥ 0.75 (pixel thô + LogReg chỉ khoảng 0.35).

```python
import math

import torch
import torch.nn.functional as F
from sklearn.linear_model import LogisticRegression
from torch import nn

from ml_course.datasets import make_shapes

device = "cuda" if torch.cuda.is_available() else "cpu"
X, y = make_shapes(10000, seed=0)
X_test, y_test = make_shapes(2000, seed=1)
X, X_test = torch.tensor(X[:, None]), torch.tensor(X_test[:, None])
### START CODE HERE ### (≈ 40 dòng)
def augment(x, g):
    B = x.shape[0]
    angle = (torch.rand(B, generator=g) * 2 - 1) * math.pi
    offset = (torch.rand(2, B, generator=g) * 2 - 1) * 0.3
    flip = torch.where(torch.rand(B, generator=g) < 0.5, -1.0, 1.0)
    cos, sin = angle.cos(), angle.sin()
    theta = torch.stack([torch.stack([cos * flip, -sin, offset[0]], 1), torch.stack([sin * flip, cos, offset[1]], 1)], 1)
    x = F.grid_sample(x, F.affine_grid(theta, x.shape, align_corners=False), align_corners=False)
    return (x * (0.6 + 0.8 * torch.rand(B, 1, 1, 1, generator=g)) + 0.15 * torch.randn(x.shape, generator=g)).clamp(0, 1)

def nt_xent(z1, z2, tau=0.2):
    z = F.normalize(torch.cat([z1, z2]), dim=1)
    sim = z @ z.T / tau
    sim.fill_diagonal_(float("-inf"))
    n = len(z1)
    return F.cross_entropy(sim, torch.cat([torch.arange(n, 2 * n), torch.arange(0, n)]).to(z.device))

def block(i, o):
    return [nn.Conv2d(i, o, 3, padding=1), nn.BatchNorm2d(o), nn.ReLU()]

torch.manual_seed(0)
encoder = nn.Sequential(*block(1, 32), nn.MaxPool2d(2), *block(32, 64), nn.MaxPool2d(2), *block(64, 128),
                        nn.AdaptiveAvgPool2d(1), nn.Flatten()).to(device)
head = nn.Sequential(nn.Linear(128, 128), nn.ReLU(), nn.Linear(128, 64)).to(device)
opt = torch.optim.Adam(list(encoder.parameters()) + list(head.parameters()), lr=1e-3)
g = torch.Generator().manual_seed(0)
for _ in range(800):
    xb = X[torch.randint(0, len(X), (256,), generator=g)]
    opt.zero_grad()
    nt_xent(head(encoder(augment(xb, g).to(device))), head(encoder(augment(xb, g).to(device)))).backward()
    opt.step()
encoder.eval()
with torch.no_grad():
    f_train = encoder(X[:100].to(device)).cpu().numpy()
    f_test = encoder(X_test.to(device)).cpu().numpy()
acc = (LogisticRegression(max_iter=5000).fit(f_train, y[:100]).predict(f_test) == y_test).mean()
### END CODE HERE ###
print(f"linear probe 100 nhãn: accuracy = {acc:.3f}")
```
