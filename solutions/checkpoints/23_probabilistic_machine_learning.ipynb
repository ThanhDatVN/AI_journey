## Phần 3 — Coding Exercise

### Task c01 — Gaussian có điều kiện (2 biến)

Viết `conditional_1d(mu, Sigma, x2)` trả về `(mean, var)` của x₁ | x₂ cho Gaussian 2 chiều.

```python
def conditional_1d(mu, Sigma, x2):
    ### START CODE HERE ### (≈ 2 dòng)
    mean = mu[0] + Sigma[0][1] / Sigma[1][1] * (x2 - mu[1])
    var = Sigma[0][0] - Sigma[0][1] ** 2 / Sigma[1][1]
    ### END CODE HERE ###
    return mean, var
```

```python
with cp.task("c01"):
    m, v = conditional_1d([1, 2], [[4, 2], [2, 3]], 5)
    check_close(m, 3.0, message="kỳ vọng")
    check_close(v, 8 / 3, message="phương sai")
    m0, v0 = conditional_1d([0, 0], [[2, 0], [0, 1]], 10)
    check_close([m0, v0], [0.0, 2.0], message="độc lập → không đổi")
```

### Task c02 — Responsibility của GMM

Viết `responsibilities(x, weights, means, stds)` cho GMM 1D: trả về ma trận (n, K) với r_ik, tính trong không gian log (dùng `scipy.special.logsumexp`).

```python
from scipy.special import logsumexp
from scipy.stats import norm

def responsibilities(x, weights, means, stds):
    x = np.asarray(x, float)
    ### START CODE HERE ### (≈ 2 dòng)
    log_p = np.log(weights)[None] + norm.logpdf(x[:, None], np.asarray(means)[None], np.asarray(stds)[None])
    return np.exp(log_p - logsumexp(log_p, axis=1, keepdims=True))
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    R = responsibilities([1.0, 1.5, 100.0], [0.5, 0.5], [0, 2], [1, 1])
    check_close(R.sum(1), np.ones(3), message="mỗi hàng tổng = 1")
    check_close(R[0, 0], 0.5, message="điểm cách đều hai tâm")
    check_close(R[1, 0], 1 / (1 + np.exp(1.0)), tol=1e-9, message="x = 1.5")
    check_true(np.isfinite(R).all() and R[2, 1] > 0.999, "ổn định số khi x rất xa")
```

## Phần 5 — Mini Project

**Mini project — Gaussian Process cho chuỗi có xu hướng và mùa vụ.** Dữ liệu: 80 quan sát của y = 0.4x + 2·sin(2πx/3) + nhiễu (σ = 0.3)
trên x ∈ [0, 12]. Dự báo x ∈ (12, 18]. Chọn kernel (sklearn) sao cho: RMSE ngoại suy so với hàm thật < 0.5 (biến `rmse_future`)
**và** độ phủ khoảng 95% trên 500 điểm mới có nhiễu trong (12, 18] ≥ 0.85 (biến `coverage`).

```python
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel, DotProduct, ExpSineSquared, WhiteKernel

rng = np.random.default_rng(0)
f = lambda x: 0.4 * x + 2 * np.sin(2 * np.pi * x / 3)
x_tr = np.sort(rng.uniform(0, 12, 80))
y_tr = f(x_tr) + rng.normal(0, 0.3, 80)
x_new = rng.uniform(12, 18, 500)
y_new = f(x_new) + rng.normal(0, 0.3, 500)
### START CODE HERE ### (≈ 6 dòng)
kernel = ConstantKernel(0.1) * DotProduct(1.0) + ConstantKernel(1.0) * ExpSineSquared(1.0, 3.0) + WhiteKernel(0.1)
gp = GaussianProcessRegressor(kernel, normalize_y=True, n_restarts_optimizer=3, random_state=0).fit(x_tr[:, None], y_tr)
grid = np.linspace(12, 18, 200)
rmse_future = np.sqrt(np.mean((gp.predict(grid[:, None]) - f(grid)) ** 2))
m, s = gp.predict(x_new[:, None], return_std=True)
coverage = np.mean(np.abs(y_new - m) < 1.96 * s)
### END CODE HERE ###
print(gp.kernel_)
print(f"RMSE ngoại suy = {rmse_future:.3f} | độ phủ 95% = {coverage:.3f}")
```

```python
with cp.task("project"):
    check_less(rmse_future, 0.5, "RMSE ngoại suy < 0.5")
    check_greater(coverage, 0.85, "độ phủ khoảng 95% ≥ 0.85")
```
