## Phần 3 — Coding Exercise

### Task c01 — Bỏ dấu tiếng Việt

Viết `strip_accents(s)`: chữ thường, bỏ mọi dấu thanh và dấu mũ/móc (dùng `unicodedata.normalize("NFD", ...)` và bỏ ký tự loại "Mn"),
chuyển "đ" thành "d". Phải đúng cả khi đầu vào ở dạng NFC hay NFD.

```python
import unicodedata

def strip_accents(s):
    ### START CODE HERE ### (≈ 2 dòng)
    s = unicodedata.normalize("NFD", s.lower()).replace("đ", "d")
    return "".join(c for c in s if unicodedata.category(c) != "Mn")
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    check_equal(strip_accents("Đường Lê Lợi"), "duong le loi", "đ và dấu")
    check_equal(strip_accents("Hội An, Phú Quốc"), "hoi an, phu quoc", "dấu thanh + mũ")
    check_equal(strip_accents(unicodedata.normalize("NFD", "Việt Nam")), "viet nam", "đầu vào NFD")
    check_equal(strip_accents("abc 123"), "abc 123", "không đổi chữ không dấu")
```

### Task c02 — Recall@k và MRR

Viết `retrieval_metrics(scores, gold, k)`: `scores` mảng (n_truy_vấn, n_tài_liệu), `gold[i]` chỉ số tài liệu đúng của truy vấn i.
Hạng của tài liệu đúng = 1 + số tài liệu có điểm **lớn hơn hẳn** nó. Trả về `(recall_at_k, mrr)`.

```python
def retrieval_metrics(scores, gold, k):
    ### START CODE HERE ### (≈ 2 dòng)
    ranks = np.array([(s > s[g]).sum() + 1 for s, g in zip(scores, gold)])
    return np.mean(ranks <= k), np.mean(1 / ranks)
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    S = np.array([[0.9, 0.1, 0.3], [0.2, 0.5, 0.8], [0.1, 0.2, 0.3]])
    r, m = retrieval_metrics(S, [0, 1, 0], k=2)
    check_close(r, 2 / 3, message="Recall@2: hạng 1, 2, 3")
    check_close(m, (1 + 1 / 2 + 1 / 3) / 3, message="MRR")
    check_close(retrieval_metrics(S, [0, 2, 2], k=1), (1.0, 1.0), message="mọi truy vấn đúng ở hạng 1")
```

## Phần 5 — Mini Project

**Mini project — Phân loại chủ đề tin tức.** Dữ liệu `make_news_vi(4000, seed=0)`, chia `train_test_split(test_size=0.3, random_state=0,
stratify=topic)`. Xây pipeline vectorizer + mô hình tuyến tính, lưu macro-F1 trên tập test vào `macro_f1`. Yêu cầu `macro_f1` ≥ 0.95
(gợi ý: unigram chưa đủ).

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import f1_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.svm import LinearSVC

from ml_course.datasets import make_news_vi

news = make_news_vi(4000, seed=0)
train, test = train_test_split(news, test_size=0.3, random_state=0, stratify=news.topic)
### START CODE HERE ### (≈ 3 dòng)
model = make_pipeline(TfidfVectorizer(ngram_range=(1, 2)), LinearSVC(C=1.0))
model.fit(train.text, train.topic)
macro_f1 = f1_score(test.topic, model.predict(test.text), average="macro")
### END CODE HERE ###
print(f"macro-F1 = {macro_f1:.3f}")
```

```python
with cp.task("project"):
    check_greater(macro_f1, 0.95, "macro-F1 ≥ 0.95")
```
