## Phần 3 — Coding Exercise

### Task c01 — Bootstrap và OOB

Viết `bootstrap_indices(n, rng)` trả về `(idx, oob_mask)`: `idx` là n chỉ số rút có hoàn lại (dùng `rng.integers`),
`oob_mask` là mảng bool độ dài n, True tại các chỉ số KHÔNG có trong `idx`.

```python
def bootstrap_indices(n, rng):
    ### START CODE HERE ### (≈ 3 dòng)
    idx = rng.integers(0, n, n)
    oob_mask = np.ones(n, dtype=bool)
    oob_mask[idx] = False
    ### END CODE HERE ###
    return idx, oob_mask
```

```python
with cp.task("c01"):
    rng = np.random.default_rng(0)
    idx, oob = bootstrap_indices(10000, rng)
    check_equal(len(idx), 10000, "đủ n chỉ số")
    check_close(oob.mean(), np.exp(-1), tol=0.01, message="tỉ lệ OOB ≈ 1/e")
    check_true(not np.any(oob[idx]), "chỉ số trong mẫu không được đánh dấu OOB")
```

### Task c02 — Tổng hợp dự đoán của rừng

Cho `probas` là mảng shape `(B, n, C)` — xác suất mỗi lớp của B cây cho n mẫu. Viết `forest_predict(probas, classes)`
trả về `(proba, pred)`: `proba` là trung bình theo cây (soft voting), `pred` là nhãn trong `classes` có xác suất lớn nhất.

```python
def forest_predict(probas, classes):
    ### START CODE HERE ### (≈ 2 dòng)
    proba = probas.mean(axis=0)
    pred = np.asarray(classes)[np.argmax(proba, axis=1)]
    ### END CODE HERE ###
    return proba, pred
```

```python
with cp.task("c02"):
    probas = np.array([[[0.9, 0.1], [0.2, 0.8]], [[0.4, 0.6], [0.3, 0.7]], [[0.5, 0.5], [0.9, 0.1]]])
    proba, pred = forest_predict(probas, ["không", "có"])
    check_close(proba, [[0.6, 0.4], [1.4 / 3, 1.6 / 3]], message="trung bình xác suất")
    check_equal(list(pred), ["không", "có"], "nhãn dự đoán")
```

## Phần 5 — Mini Project

**Mini project — Random Forest dự đoán giá nhà.** Đọc `house_prices.csv` (`repo_path("data", "sample", "house_prices.csv")`),
target `price_billion_vnd`, one-hot `district`. Chia train/test 75/25 (`random_state=0`). Chọn `max_features` và `min_samples_leaf`
(bằng OOB hoặc CV), train `RandomForestRegressor` (≥ 200 cây). Đạt **MAE test ≤ 0.52** (biến `test_mae`, tỉ VND).

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
from ml_course.utils import repo_path

df = pd.read_csv(repo_path("data", "sample", "house_prices.csv"))
### START CODE HERE ### (≈ 10 dòng)
X = pd.get_dummies(df.drop(columns=["price_billion_vnd"]), columns=["district"], dtype=float)
y = df["price_billion_vnd"]
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, random_state=0)
best, best_score = None, -np.inf
for mf in [0.3, 0.5, 0.8]:
    for leaf in [1, 3, 5]:
        m = RandomForestRegressor(n_estimators=200, max_features=mf, min_samples_leaf=leaf, oob_score=True, random_state=0).fit(X_tr, y_tr)
        if m.oob_score_ > best_score:
            best, best_score = m, m.oob_score_
test_mae = mean_absolute_error(y_te, best.predict(X_te))
### END CODE HERE ###
print("OOB R² tốt nhất:", round(best_score, 4), "| MAE test:", round(test_mae, 3))
```

```python
with cp.task("project"):
    check_less(test_mae, 0.52, "MAE test ≤ 0.52 tỉ")
```
