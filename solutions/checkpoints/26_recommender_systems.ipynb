## Phần 3 — Coding Exercise

### Task c01 — Trung bình giảm chấn

Viết `damped_mean(sums, counts, mu, m)` trả về mảng (sums + m·mu) / (counts + m).

```python
def damped_mean(sums, counts, mu, m):
    ### START CODE HERE ### (≈ 1 dòng)
    return (np.asarray(sums, float) + m * mu) / (np.asarray(counts, float) + m)
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    check_close(damped_mean([10.0, 420.0], [2, 100], 3.8, 10), [4.0, 458 / 110], message="ví dụ")
    check_close(damped_mean([0.0], [0], 3.5, 5), [3.5], message="item chưa có rating → μ")
```

### Task c02 — Top-k loại item đã thấy

Viết `top_k(scores, seen_items, k)` trả về chỉ số k item điểm cao nhất (giảm dần), bỏ qua các item trong `seen_items`.

```python
def top_k(scores, seen_items, k):
    ### START CODE HERE ### (≈ 3 dòng)
    s = np.asarray(scores, float).copy()
    s[list(seen_items)] = -np.inf
    return np.argsort(-s)[:k]
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    check_equal(list(top_k([0.1, 0.9, 0.5, 0.7], {1}, 2)), [3, 2], "loại item đã thấy")
    check_equal(list(top_k([3, 2, 1], set(), 3)), [0, 1, 2], "không có item đã thấy")
```

## Phần 5 — Mini Project

**Mini project — Gợi ý top-10 vượt popularity.** Dùng `ratings.csv` với cách chia `split_per_user` (20% mỗi user, seed 0) và giao thức
của chương (loại phim đã thấy trong train, "đúng" = phim trong test có rating ≥ 4). Xây mô hình top-N bất kỳ đạt precision@10
`p10_model` ≥ 0.11 (popularity ≈ 0.099).

```python
import pandas as pd
from ml_course.metrics import precision_at_k
from ml_course.utils import repo_path

ratings = pd.read_csv(repo_path("data", "sample", "ratings.csv"))
n_users, n_items = ratings.user_id.max() + 1, ratings.item_id.max() + 1
rng = np.random.default_rng(0)
test_mask = np.zeros(len(ratings), bool)
for _, g in ratings.groupby("user_id"):
    if len(g) >= 5:
        test_mask[rng.choice(g.index, max(1, int(0.2 * len(g))), replace=False)] = True
train, test = ratings[~test_mask], ratings[test_mask]
seen = {u: set(g.item_id) for u, g in train.groupby("user_id")}
liked = {u: [i for i, r in zip(g.item_id, g.rating) if r >= 4] for u, g in test.groupby("user_id")}
### START CODE HERE ### (≈ 5 dòng)
B = np.zeros((n_users, n_items))
B[train.user_id, train.item_id] = 1.0
U, s, Vt = np.linalg.svd(B, full_matrices=False)
scores = (U[:, :5] * s[:5]) @ Vt[:5]                               # PureSVD k = 5
### END CODE HERE ###
p = []
for u, rel in liked.items():
    if rel:
        sc = scores[u].copy()
        sc[list(seen.get(u, []))] = -np.inf
        p.append(precision_at_k(np.argsort(-sc)[:10], rel, 10))
p10_model = float(np.mean(p))
print(f"precision@10 = {p10_model:.4f}")
```

```python
with cp.task("project"):
    check_greater(p10_model, 0.11, "precision@10 ≥ 0.11")
```
