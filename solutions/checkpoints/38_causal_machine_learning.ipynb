## Phần 3 — Coding Exercise

### Task c01 — Chênh lệch trung bình và sai số chuẩn (A/B test)

Viết `ab_effect(y, t)` trả về `(diff, se)`: chênh lệch trung bình nhóm t = 1 trừ nhóm t = 0, và sai số chuẩn
sqrt(var1/n1 + var0/n0) (phương sai mẫu ddof = 1).

```python
def ab_effect(y, t):
    ### START CODE HERE ### (≈ 4 dòng)
    y, t = np.asarray(y, float), np.asarray(t)
    y1, y0 = y[t == 1], y[t == 0]
    diff = y1.mean() - y0.mean()
    return diff, np.sqrt(y1.var(ddof=1) / len(y1) + y0.var(ddof=1) / len(y0))
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    d, s = ab_effect([3, 5, 1, 1], [1, 1, 0, 0])
    check_close(d, 3.0, message="chênh lệch")
    check_close(s, np.sqrt(2 / 2 + 0 / 2), message="sai số chuẩn")
    r = np.random.default_rng(0)
    t = (r.random(20000) < 0.5).astype(int)
    y = 2 * t + r.normal(size=20000)
    d2, s2 = ab_effect(y, t)
    check_less(abs(d2 - 2), 4 * s2, "ước lượng gần hiệu ứng thật")
```

### Task c02 — IPW (Hajek)

Viết `ipw_hajek(y, t, e)`: trung bình có trọng số t/e của y trong nhóm điều trị trừ trung bình có trọng số (1 − t)/(1 − e) trong nhóm đối chứng.

```python
def ipw_hajek(y, t, e):
    ### START CODE HERE ### (≈ 3 dòng)
    y, t, e = (np.asarray(v, float) for v in (y, t, e))
    w1, w0 = t / e, (1 - t) / (1 - e)
    return (w1 * y).sum() / w1.sum() - (w0 * y).sum() / w0.sum()
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    check_close(ipw_hajek([10, 6, 4, 2], [1, 1, 0, 0], [0.5, 0.8, 0.5, 0.2]), (20 + 7.5) / (2 + 1.25) - (8 + 2.5) / (2 + 1.25), message="ví dụ nhỏ")
    r = np.random.default_rng(1)
    x = r.normal(size=50000)
    e = 1 / (1 + np.exp(-x))
    t = (r.random(50000) < e).astype(int)
    y = 3 * t + 2 * x + r.normal(size=50000)
    check_less(abs(ipw_hajek(y, t, e) - 3), 0.1, "IPW với điểm xu hướng đúng ≈ ATE")
    check_greater(y[t == 1].mean() - y[t == 0].mean() - 3, 1.0, "so sánh ngây thơ bị chệch")
```

## Phần 5 — Mini Project

**Mini project — Ước lượng tác động từ dữ liệu quan sát.** Dữ liệu `make_causal_marketing(20000, seed=5)` (coupon không ngẫu nhiên). Dùng AIPW
với cross-fitting (5 fold, GBM cho kết quả, logistic cho xu hướng, cắt e vào [0.01, 0.99]) trên các đặc trưng `age`, `income_mvnd`,
`past_purchases`. Lưu ước lượng vào `ate_hat`, sai số chuẩn vào `se_hat`. Yêu cầu: |ate_hat − ATE thật| < 8 và so sánh ngây thơ lệch > 60.

```python
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import KFold

from ml_course.datasets import make_causal_marketing

d = make_causal_marketing(20000, seed=5)
X, T, Y = d[["age", "income_mvnd", "past_purchases"]].values, d.coupon.values, d.spend_kvnd.values
true_ate = d.true_effect.mean()
naive = Y[T == 1].mean() - Y[T == 0].mean()
### START CODE HERE ### (≈ 12 dòng)
mu1, mu0, e = np.zeros(len(Y)), np.zeros(len(Y)), np.zeros(len(Y))
for tr, te in KFold(5, shuffle=True, random_state=0).split(X):
    mu1[te] = HistGradientBoostingRegressor(random_state=0).fit(X[tr][T[tr] == 1], Y[tr][T[tr] == 1]).predict(X[te])
    mu0[te] = HistGradientBoostingRegressor(random_state=0).fit(X[tr][T[tr] == 0], Y[tr][T[tr] == 0]).predict(X[te])
    e[te] = np.clip(LogisticRegression(max_iter=1000).fit(X[tr], T[tr]).predict_proba(X[te])[:, 1], 0.01, 0.99)
scores = mu1 - mu0 + T * (Y - mu1) / e - (1 - T) * (Y - mu0) / (1 - e)
ate_hat, se_hat = scores.mean(), scores.std() / np.sqrt(len(Y))
### END CODE HERE ###
print(f"ngây thơ = {naive:.1f} | AIPW = {ate_hat:.1f} ± {1.96 * se_hat:.1f} | ATE thật = {true_ate:.1f}")
```

```python
with cp.task("project"):
    check_less(abs(ate_hat - true_ate), 8, "AIPW gần ATE thật")
    check_greater(abs(naive - true_ate), 60, "so sánh ngây thơ bị chệch")
    check_less(se_hat, 5, "sai số chuẩn hợp lý")
```
