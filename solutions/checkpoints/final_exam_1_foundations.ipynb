### Task c01 — Chuẩn hoá không rò rỉ

Viết `fit_standardizer(X_train)` trả về `(mean, std)` theo cột (std với ddof = 0; cột hằng → std = 1) và
`transform(X, mean, std)` → (X − mean)/std. Chỉ dùng thống kê của train.

```python
def fit_standardizer(X_train):
    ### START CODE HERE ### (≈ 3 dòng)
    X_train = np.asarray(X_train, float)
    mean, std = X_train.mean(axis=0), X_train.std(axis=0)
    return mean, np.where(std == 0, 1.0, std)
    ### END CODE HERE ###


def transform(X, mean, std):
    ### START CODE HERE ### (≈ 1 dòng)
    return (np.asarray(X, float) - mean) / std
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    Xtr = np.array([[1.0, 10, 5], [3.0, 30, 5]])
    mean, std = fit_standardizer(Xtr)
    check_close(mean, [2, 20, 5], message="trung bình theo cột")
    check_close(std, [1, 10, 1], message="độ lệch chuẩn (cột hằng → 1)")
    check_close(transform(np.array([[2.0, 40, 5]]), mean, std), [[0, 2, 0]], message="biến đổi dữ liệu mới bằng thống kê train")
```

### Task c02 — Gradient descent cho hồi quy tuyến tính

Viết `fit_linear_gd(X, y, lr=0.1, n_iter=2000)`: thêm cột 1 cho hệ số chặn, khởi tạo trọng số 0, lặp
w ← w − lr · (2/n) Xᵀ(Xw − y). Trả về w (phần tử đầu là hệ số chặn).

```python
def fit_linear_gd(X, y, lr=0.1, n_iter=2000):
    ### START CODE HERE ### (≈ 5 dòng)
    Xb = np.column_stack([np.ones(len(X)), np.asarray(X, float)])
    y = np.asarray(y, float)
    w = np.zeros(Xb.shape[1])
    for _ in range(n_iter):
        w -= lr * 2 / len(y) * Xb.T @ (Xb @ w - y)
    return w
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    rng = np.random.default_rng(0)
    X = rng.normal(size=(200, 3))
    y = 1.5 + X @ np.array([2.0, -1.0, 0.5]) + rng.normal(0, 0.1, 200)
    w = fit_linear_gd(X, y)
    w_ls = np.linalg.lstsq(np.column_stack([np.ones(200), X]), y, rcond=None)[0]
    check_close(w, w_ls, tol=1e-4, message="hội tụ về nghiệm bình phương tối thiểu")
    check_close(fit_linear_gd(np.array([[0.0], [1.0], [2.0]]), np.array([1.0, 3.0, 5.0])), [1, 2], tol=1e-3, message="ví dụ nhỏ y = 1 + 2x")
```

```python
import pandas as pd
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_percentage_error
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import FunctionTransformer, OneHotEncoder, StandardScaler

from ml_course.utils import repo_path

houses = pd.read_csv(repo_path("data", "sample", "house_prices.csv"))
NUMERIC = ["area_m2", "bedrooms", "bathrooms", "age_years", "distance_to_center_km"]
CATEGORICAL = ["district", "has_garden"]
train, test = train_test_split(houses, test_size=0.2, random_state=0)
### START CODE HERE ### (≈ 6 dòng)
pre = ColumnTransformer([("num", make_pipeline(FunctionTransformer(np.log1p), StandardScaler()), NUMERIC),
                         ("cat", OneHotEncoder(handle_unknown="ignore"), CATEGORICAL)])
model = TransformedTargetRegressor(make_pipeline(pre, Ridge(1.0)), func=np.log, inverse_func=np.exp)
model.fit(train[NUMERIC + CATEGORICAL], train.price_billion_vnd)
pred = model.predict(test[NUMERIC + CATEGORICAL])
mape = mean_absolute_percentage_error(test.price_billion_vnd, pred)
### END CODE HERE ###
print(f"MAPE test = {mape:.1%}")
```

```python
with cp.task("project"):
    check_equal(len(pred), len(test), "dự đoán cho mọi căn trong test")
    check_less(mape, 0.12, "MAPE test < 12%")
    check_true((np.asarray(pred) > 0).all(), "giá dự đoán dương")
```
