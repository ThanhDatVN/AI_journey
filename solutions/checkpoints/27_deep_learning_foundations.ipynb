## Phần 3 — Coding Exercise

### Task c01 — Softmax ổn định

Viết `stable_softmax(Z)` theo hàng (trừ max mỗi hàng trước exp).

```python
def stable_softmax(Z):
    ### START CODE HERE ### (≈ 2 dòng)
    E = np.exp(Z - Z.max(axis=1, keepdims=True))
    return E / E.sum(axis=1, keepdims=True)
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    P = stable_softmax(np.array([[1.0, 2.0, 3.0], [1000.0, 1000.0, 1000.0]]))
    check_close(P.sum(1), [1.0, 1.0], message="tổng mỗi hàng = 1")
    check_close(P[1], [1 / 3] * 3, message="ổn định với logit lớn")
    check_close(P[0], np.exp([1, 2, 3]) / np.exp([1, 2, 3]).sum(), message="giá trị đúng")
```

### Task c02 — Backward của ReLU và lớp tuyến tính

Viết `relu_linear_backward(d_out, X, W, Z)`: mạng con H = ReLU(Z), Z = XW + b, nhận d_out = ∂L/∂H; trả về (dX, dW, db).

```python
def relu_linear_backward(d_out, X, W, Z):
    ### START CODE HERE ### (≈ 2 dòng)
    dZ = d_out * (Z > 0)
    return dZ @ W.T, X.T @ dZ, dZ.sum(axis=0)
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    rng = np.random.default_rng(0)
    X, W, b = rng.normal(size=(5, 3)), rng.normal(size=(3, 4)), rng.normal(size=4)
    Z = X @ W + b
    d_out = rng.normal(size=(5, 4))
    dX, dW, db = relu_linear_backward(d_out, X, W, Z)
    L = lambda W_: np.sum(d_out * np.maximum(0, X @ W_ + b))
    E = np.zeros_like(W); E[1, 2] = 1e-6
    check_close(dW[1, 2], (L(W + E) - L(W - E)) / 2e-6, tol=1e-5, message="dW khớp sai phân hữu hạn")
    check_close(db, (d_out * (Z > 0)).sum(0), message="db")
    check_true(dX.shape == X.shape and dW.shape == W.shape, "đúng shape")
```

## Phần 5 — Mini Project

**Mini project — MLP PyTorch gọn nhẹ cho chữ số.** `load_digits()`, chia train/test 75/25 (`random_state=0`, stratify), chuẩn hoá theo train.
Xây và huấn luyện MLP sao cho accuracy test `test_acc` ≥ 0.96 **và** số tham số `n_params` ≤ 10 000.

```python
import torch
from torch import nn
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X, y = load_digits(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, random_state=0, stratify=y)
sc = StandardScaler().fit(X_tr)
Xt_tr, Xt_te = torch.tensor(sc.transform(X_tr), dtype=torch.float32), torch.tensor(sc.transform(X_te), dtype=torch.float32)
yt_tr, yt_te = torch.tensor(y_tr), torch.tensor(y_te)
torch.manual_seed(0)
### START CODE HERE ### (≈ 14 dòng)
model = nn.Sequential(nn.Linear(64, 64), nn.ReLU(), nn.Linear(64, 32), nn.ReLU(), nn.Linear(32, 10))
opt = torch.optim.Adam(model.parameters(), lr=1e-3)
loss_fn = nn.CrossEntropyLoss()
for epoch in range(60):
    model.train()
    perm = torch.randperm(len(Xt_tr))
    for s in range(0, len(perm), 32):
        idx = perm[s:s + 32]
        opt.zero_grad()
        loss_fn(model(Xt_tr[idx]), yt_tr[idx]).backward()
        opt.step()
model.eval()
with torch.no_grad():
    test_acc = (model(Xt_te).argmax(1) == yt_te).float().mean().item()
n_params = sum(p.numel() for p in model.parameters())
### END CODE HERE ###
print(f"tham số = {n_params} | accuracy test = {test_acc:.4f}")
```

```python
with cp.task("project"):
    check_less(n_params, 10001, "≤ 10 000 tham số")
    check_greater(test_acc, 0.96, "accuracy test ≥ 0.96")
```
