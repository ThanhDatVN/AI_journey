### Task c01 — Population Stability Index

Viết `psi(reference, current, bins=10, eps=1e-4)`: bin theo phân vị của reference (hai đầu mở ra ±∞), tỉ lệ mỗi bin kẹp
tối thiểu `eps`, trả về Σ (c − r)·ln(c/r).

```python
def psi(reference, current, bins=10, eps=1e-4):
    ### START CODE HERE ### (≈ 6 dòng)
    edges = np.unique(np.quantile(reference, np.linspace(0, 1, bins + 1)))
    edges[0], edges[-1] = -np.inf, np.inf
    r = np.histogram(reference, edges)[0] / len(reference)
    c = np.histogram(current, edges)[0] / len(current)
    r, c = np.clip(r, eps, None), np.clip(c, eps, None)
    return float(np.sum((c - r) * np.log(c / r)))
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    rng = np.random.default_rng(0)
    ref = rng.normal(size=20000)
    check_close(psi(ref, ref), 0.0, message="giống hệt → 0")
    check_less(psi(ref, rng.normal(size=20000)), 0.01, "cùng phân phối")
    check_greater(psi(ref, rng.normal(1, 1, 20000)), 0.25, "lệch 1σ")
    check_true(np.isfinite(psi(ref, np.full(500, 100.0))), "bin rỗng vẫn hữu hạn")
    check_close(psi(np.arange(4.0), np.array([0.0, 1, 1, 2, 2, 3, 3, 3, 3, 3]), bins=4),
                sum((c - 0.25) * np.log(c / 0.25) for c in [0.1, 0.2, 0.2, 0.5]), message="ví dụ nhỏ 4 bin")
```

### Task c02 — Gán biến thể ổn định

Viết `assign(user_id, weights, salt)`: ánh xạ `salt + ":" + user_id` qua SHA-256 thành số u ∈ [0, 1)
(lấy 15 ký tự hex đầu chia 16**15), rồi trả về biến thể đầu tiên mà tổng trọng số tích luỹ > u.
`weights` là dict có thứ tự, ví dụ {"v2": 0.1, "v1": 0.9}.

```python
import hashlib


def assign(user_id, weights, salt="exp"):
    ### START CODE HERE ### (≈ 7 dòng)
    u = int(hashlib.sha256(f"{salt}:{user_id}".encode()).hexdigest()[:15], 16) / 16 ** 15
    cumulative = 0.0
    for variant, w in weights.items():
        cumulative += w
        if u < cumulative:
            return variant
    return variant
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    users = [f"U{i}" for i in range(20000)]
    share = np.mean([assign(u, {"v2": 0.1, "v1": 0.9}) == "v2" for u in users])
    check_less(abs(share - 0.1), 0.01, "tỉ lệ ≈ 10%")
    check_true(all(assign(u, {"v2": 0.1, "v1": 0.9}) == assign(u, {"v2": 0.1, "v1": 0.9}) for u in users[:500]), "ổn định")
    check_true(all(assign(u, {"v2": 0.25, "v1": 0.75}) == "v2" for u in users if assign(u, {"v2": 0.1, "v1": 0.9}) == "v2"), "tăng tỉ lệ: bucket lồng nhau")
    expected_u = int(hashlib.sha256(b"exp:U0").hexdigest()[:15], 16) / 16 ** 15
    check_equal(assign("U0", {"a": 0.5, "b": 0.5}), "a" if expected_u < 0.5 else "b", "đúng công thức băm")
    other = np.mean([assign(u, {"v2": 0.1, "v1": 0.9}, salt="other") == "v2" for u in users if assign(u, {"v2": 0.1, "v1": 0.9}) == "v2"])
    check_less(abs(other - 0.1), 0.03, "salt khác → độc lập")
```

## Phần 5 — Mini Project

**Mini project — Giám sát trôi theo tuần.** Mô hình churn huấn luyện trên `make_customer_churn(6000, seed=0)`. Cho 6 tuần production
(hàm `week(w)` đã viết sẵn: tuần 4–6 giá cước tăng 30%). Viết `weekly_report(reference, model, weeks)` trả về DataFrame (index = tuần) với cột
`psi_monthly_charges`, `psi_contract`, `pred_mean`, `drift_alert` (True nếu PSI lớn nhất trong hai cột > 0.25). Dùng `psi` (biến số, bin theo phân vị
reference) và `psi_categorical` bạn tự viết. Yêu cầu: cảnh báo đúng ở tuần 4–6 và không cảnh báo ở tuần 1–3.

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from ml_course.datasets import make_customer_churn

FEATURES = ["tenure_months", "contract", "internet_service", "monthly_charges", "total_charges",
            "payment_method", "support_calls", "senior", "n_products"]
reference = make_customer_churn(6000, seed=0)
model = make_pipeline(ColumnTransformer([("num", StandardScaler(), ["tenure_months", "monthly_charges", "total_charges", "support_calls", "n_products"]),
                                         ("cat", OneHotEncoder(handle_unknown="ignore"), ["contract", "internet_service", "payment_method", "senior"])]),
                      LogisticRegression(max_iter=1000)).fit(reference[FEATURES], reference.churn)


def week(w):
    df = make_customer_churn(2000, seed=50 + w)
    if w >= 4:
        df["monthly_charges"] = df.monthly_charges * 1.3
    return df


weeks = {w: week(w) for w in range(1, 7)}
### START CODE HERE ### (≈ 22 dòng)
def psi(reference, current, bins=10, eps=1e-4):
    edges = np.unique(np.quantile(reference, np.linspace(0, 1, bins + 1)))
    edges[0], edges[-1] = -np.inf, np.inf
    r = np.clip(np.histogram(reference, edges)[0] / len(reference), eps, None)
    c = np.clip(np.histogram(current, edges)[0] / len(current), eps, None)
    return float(np.sum((c - r) * np.log(c / r)))


def psi_categorical(reference, current, eps=1e-4):
    reference, current = np.asarray(reference), np.asarray(current)
    cats = sorted(set(reference) | set(current), key=str)
    r = np.clip([np.mean(reference == k) for k in cats], eps, None)
    c = np.clip([np.mean(current == k) for k in cats], eps, None)
    return float(np.sum((c - r) * np.log(c / r)))


def weekly_report(reference, model, weeks):
    rows = {}
    for w, df in weeks.items():
        row = {"psi_monthly_charges": psi(reference.monthly_charges, df.monthly_charges),
               "psi_contract": psi_categorical(reference.contract, df.contract),
               "pred_mean": float(model.predict_proba(df[FEATURES])[:, 1].mean())}
        row["drift_alert"] = max(row["psi_monthly_charges"], row["psi_contract"]) > 0.25
        rows[w] = row
    return pd.DataFrame(rows).T
### END CODE HERE ###
report = weekly_report(reference, model, weeks)
print(report)
```
