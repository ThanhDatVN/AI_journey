## Phần 2 — Coding Exercise

### Task c01 — Phát hiện outlier bằng IQR

Viết `iqr_outlier_mask(s, k=1.5)` nhận `pd.Series`, trả về Series boolean: True nếu giá trị nằm ngoài $[Q_1 - k\cdot IQR,\; Q_3 + k\cdot IQR]$.

```python
import pandas as pd

def iqr_outlier_mask(s, k=1.5):
    ### START CODE HERE ### (≈ 3 dòng)
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    mask = (s < q1 - k * iqr) | (s > q3 + k * iqr)
    ### END CODE HERE ###
    return mask
```

```python
with cp.task("c01"):
    s = pd.Series([1, 2, 3, 4, 5, 6, 7, 100, -50])
    m = iqr_outlier_mask(s)
    check_equal(m.tolist(), [False] * 7 + [True, True], "phát hiện 100 và -50")
    check_equal(int(iqr_outlier_mask(s, k=100).sum()), 0, "k rất lớn → không có outlier")
```

### Task c02 — Cramér's V

Viết `cramers_v(x, y)` (hai Series phân loại) dùng `scipy.stats.chi2_contingency` trên `pd.crosstab(x, y)`:
$V = \sqrt{\chi^2 / (n\,(\min(r, c) - 1))}$. Dùng `correction=False`.

```python
import pandas as pd
from scipy import stats

def cramers_v(x, y):
    ### START CODE HERE ### (≈ 4 dòng)
    table = pd.crosstab(x, y)
    chi2 = stats.chi2_contingency(table, correction=False)[0]
    n = table.to_numpy().sum()
    v = np.sqrt(chi2 / (n * (min(table.shape) - 1)))
    ### END CODE HERE ###
    return v
```

```python
with cp.task("c02"):
    a = pd.Series(["x", "x", "y", "y"] * 50)
    check_close(cramers_v(a, a), 1.0, message="liên hệ hoàn hảo → V = 1")
    rng = np.random.default_rng(0)
    u = pd.Series(rng.choice(["p", "q"], 5000)); v = pd.Series(rng.choice(["r", "s", "t"], 5000))
    check_less(cramers_v(u, v), 0.05, "độc lập → V ≈ 0")
```

## Phần 4 — Mini Project

**Mini project — Làm sạch dữ liệu giá nhà.** Viết `clean_houses(df)` nhận DataFrame từ `house_prices_dirty.csv` và trả về DataFrame đã:
1. xoá dòng trùng lặp;
2. chuẩn hoá `district` về đúng 5 giá trị `{"Quận 1", "Quận 3", "Quận 7", "Bình Thạnh", "Thủ Đức"}`;
3. sửa diện tích gõ thừa số 0: nếu `area_m2 / bedrooms > 150` thì chia `area_m2` cho 10.

```python
import pandas as pd

def clean_houses(df):
    df = df.copy()
    ### START CODE HERE ### (≈ 6 dòng)
    df = df.drop_duplicates().reset_index(drop=True)
    df["district"] = df["district"].map(
        lambda s: " ".join(w.capitalize() for w in s.strip().lower().replace("q.", "quận").split()))
    bad = df["area_m2"] / df["bedrooms"] > 150
    df.loc[bad, "area_m2"] = df.loc[bad, "area_m2"] / 10
    ### END CODE HERE ###
    return df

raw = pd.read_csv("../../data/sample/house_prices_dirty.csv")
clean = clean_houses(raw)
clean.shape
```

```python
with cp.task("project"):
    check_equal(int(clean.duplicated().sum()), 0, "không còn trùng lặp")
    check_equal(set(clean["district"]), {"Quận 1", "Quận 3", "Quận 7", "Bình Thạnh", "Thủ Đức"}, "5 quận chuẩn")
    check_true(bool((clean["area_m2"] / clean["bedrooms"]).max() <= 150), "đã sửa diện tích vô lý")
    check_equal(len(clean), len(raw.drop_duplicates()), "không xoá nhầm hàng")
```
