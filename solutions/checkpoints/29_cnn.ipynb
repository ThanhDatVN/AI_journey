## Phần 3 — Coding Exercise

### Task c01 — Kích thước đầu ra conv

Viết `conv_out_size(h, k, s, p)` trả về kích thước đầu ra của tích chập theo một chiều.

```python
def conv_out_size(h, k, s=1, p=0):
    ### START CODE HERE ### (≈ 1 dòng)
    return (h + 2 * p - k) // s + 1
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    check_equal(conv_out_size(28, 3, 1, 1), 28, "padding same")
    check_equal(conv_out_size(64, 3, 2, 1), 32, "stride 2")
    check_equal(conv_out_size(13, 3), 11, "không padding")
    check_equal(conv_out_size(7, 3, 2, 0), 3, "làm tròn xuống")
```

### Task c02 — Tích chập 2D một kênh

Viết `conv2d_single(x, k)` (stride 1, không padding) cho ảnh 2D `x` và kernel 2D `k` (cross-correlation).

```python
def conv2d_single(x, k):
    kh, kw = k.shape
    H, W = x.shape[0] - kh + 1, x.shape[1] - kw + 1
    out = np.zeros((H, W))
    ### START CODE HERE ### (≈ 3 dòng)
    for i in range(H):
        for j in range(W):
            out[i, j] = np.sum(x[i:i + kh, j:j + kw] * k)
    ### END CODE HERE ###
    return out
```

```python
with cp.task("c02"):
    x = np.arange(16.0).reshape(4, 4)
    k = np.array([[1.0, 0.0], [0.0, -1.0]])
    check_close(conv2d_single(x, k), np.full((3, 3), -5.0), message="x[i,j] − x[i+1,j+1] = −5")
    rng = np.random.default_rng(0)
    x2, k2 = rng.normal(size=(6, 5)), rng.normal(size=(3, 3))
    from scipy.signal import correlate2d
    check_close(conv2d_single(x2, k2), correlate2d(x2, k2, mode="valid"), message="khớp scipy correlate2d")
```

## Phần 5 — Mini Project

**Mini project — Tổng quát hoá ra vị trí mới.** Train **chỉ** với `make_shapes(3000, region="left", seed=40)`; đánh giá trên
`make_shapes(1000, region="right", seed=41)` (hình ở vị trí chưa từng thấy). Yêu cầu accuracy trên nửa phải `right_acc` ≥ 0.75.
Gợi ý: kiến trúc nào bất biến với vị trí? Hoặc augmentation nào?

```python
import torch
from torch import nn
from ml_course.datasets import make_shapes

device = "cuda" if torch.cuda.is_available() else "cpu"
torch.backends.cudnn.deterministic = True
to_t = lambda a: torch.tensor(a[:, None], dtype=torch.float32)
XL, yL = make_shapes(3000, region="left", seed=40)
XR, yR = make_shapes(1000, region="right", seed=41)
XL, XR, yL, yR = to_t(XL), to_t(XR), torch.tensor(yL), torch.tensor(yR)
torch.manual_seed(0)
### START CODE HERE ### (≈ 18 dòng)
model = nn.Sequential(nn.Conv2d(1, 16, 3, padding=1), nn.BatchNorm2d(16), nn.ReLU(), nn.MaxPool2d(2),
                      nn.Conv2d(16, 32, 3, padding=1), nn.BatchNorm2d(32), nn.ReLU(), nn.MaxPool2d(2),
                      nn.Conv2d(32, 64, 3, padding=1), nn.BatchNorm2d(64), nn.ReLU(),
                      nn.AdaptiveAvgPool2d(1), nn.Flatten(), nn.Linear(64, 5)).to(device)
opt = torch.optim.Adam(model.parameters(), lr=1e-3)
g = torch.Generator().manual_seed(0)
for epoch in range(15):
    model.train()
    perm = torch.randperm(len(XL), generator=g)
    for s in range(0, len(perm), 64):
        idx = perm[s:s + 64]
        opt.zero_grad()
        nn.functional.cross_entropy(model(XL[idx].to(device)), yL[idx].to(device)).backward()
        opt.step()
model.eval()
with torch.no_grad():
    right_acc = (model(XR.to(device)).argmax(1).cpu() == yR).float().mean().item()
### END CODE HERE ###
print(f"accuracy trên nửa phải (vị trí chưa từng thấy) = {right_acc:.3f}")
```

```python
with cp.task("project"):
    check_greater(right_acc, 0.75, "tổng quát hoá ra vị trí mới ≥ 0.75")
```
