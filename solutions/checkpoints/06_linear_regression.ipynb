## Phần 3 — Coding Exercise

### Task c01 — Gradient descent vector hoá

Viết `gd_linear(X, y, alpha, num_iters)` (X **đã có** cột bias) trả về `(theta, J_history)`,
với $\theta := \theta - \frac{\alpha}{m}X^\top(X\theta - y)$ và $J = \frac{1}{2m}\lVert X\theta - y\rVert^2$ ghi sau mỗi vòng.

```python
def gd_linear(X, y, alpha=0.1, num_iters=500):
    m = len(y)
    theta = np.zeros(X.shape[1])
    J_history = []
    for _ in range(num_iters):
        ### START CODE HERE ### (≈ 3 dòng)
        theta = theta - alpha / m * X.T @ (X @ theta - y)
        err = X @ theta - y
        J_history.append(err @ err / (2 * m))
        ### END CODE HERE ###
    return theta, J_history
```

```python
with cp.task("c01"):
    rng = np.random.default_rng(1)
    Z = rng.normal(size=(200, 3))
    X = np.c_[np.ones(200), Z]
    y = X @ np.array([1.0, 2.0, -3.0, 0.5]) + rng.normal(0, 0.1, 200)
    theta, J = gd_linear(X, y, alpha=0.1, num_iters=1000)
    check_close(theta, np.linalg.lstsq(X, y, rcond=None)[0], tol=1e-3, message="hội tụ về nghiệm tối ưu")
    check_true(all(np.diff(J) <= 1e-12), "J giảm mỗi vòng lặp")
```

### Task c02 — Ridge nghiệm đóng (không phạt bias)

Viết `ridge_closed(X, y, lam)` (X **đã có** cột bias ở đầu) giải $(X^\top X + \lambda I')\theta = X^\top y$, với $I'$ là ma trận đơn vị có phần tử $[0, 0] = 0$. Dùng `np.linalg.solve`.

```python
def ridge_closed(X, y, lam):
    ### START CODE HERE ### (≈ 3 dòng)
    I = np.eye(X.shape[1])
    I[0, 0] = 0.0
    theta = np.linalg.solve(X.T @ X + lam * I, X.T @ y)
    ### END CODE HERE ###
    return theta
```

```python
with cp.task("c02"):
    from sklearn.linear_model import Ridge
    rng = np.random.default_rng(2)
    Z = rng.normal(size=(100, 4))
    y = Z @ np.array([1.0, 0.0, -2.0, 3.0]) + 5 + rng.normal(size=100)
    theta = ridge_closed(np.c_[np.ones(100), Z], y, 10.0)
    ref = Ridge(alpha=10.0).fit(Z, y)
    check_close(theta[1:], ref.coef_, tol=1e-8, message="hệ số khớp sklearn Ridge")
    check_close(theta[0], ref.intercept_, tol=1e-8, message="bias không bị phạt")
```

Dữ liệu chỉ có 10 điểm. So sánh bằng **Leave-One-Out CV** (`LeaveOneOut()`, scoring MAE) ba model:
(a) Linear Regression; (b) đa thức bậc 4 (Pipeline: PolynomialFeatures → StandardScaler → LinearRegression);
(c) **log-linear**: Linear Regression trên log(Salary) rồi `exp` lại (dùng `TransformedTargetRegressor(func=np.log, inverse_func=np.exp)`).
Lưu MAE LOOCV vào dict `mae` với các khoá `"linear"`, `"poly4"`, `"loglinear"`. Model tốt nhất phải có MAE < 60 000.

```python
import pandas as pd
from sklearn.compose import TransformedTargetRegressor
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import LeaveOneOut, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler

data = pd.read_csv("../../data/sample/Position_Salaries.csv")
X, y = data[["Level"]].to_numpy(float), data["Salary"].to_numpy(float)
models = {
    "linear": LinearRegression(),
    "poly4": make_pipeline(PolynomialFeatures(4), StandardScaler(), LinearRegression()),
    "loglinear": TransformedTargetRegressor(regressor=LinearRegression(), func=np.log, inverse_func=np.exp),
}
mae = {}
### START CODE HERE ### (≈ 2 dòng)
for name, model in models.items():
    mae[name] = -cross_val_score(model, X, y, cv=LeaveOneOut(), scoring="neg_mean_absolute_error").mean()
### END CODE HERE ###
print({k: round(v) for k, v in mae.items()})
```
