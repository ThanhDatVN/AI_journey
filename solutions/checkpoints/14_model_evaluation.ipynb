## Phần 3 — Coding Exercise

### Task c01 — Các metric từ confusion matrix

Viết `binary_metrics(y_true, y_pred)` trả về dict với khoá `precision`, `recall`, `specificity`, `f1` (nhãn 0/1; mẫu số 0 → trả 0.0).

```python
def binary_metrics(y_true, y_pred):
    ### START CODE HERE ### (≈ 8 dòng)
    tp = np.sum((y_true == 1) & (y_pred == 1))
    fp = np.sum((y_true == 0) & (y_pred == 1))
    fn = np.sum((y_true == 1) & (y_pred == 0))
    tn = np.sum((y_true == 0) & (y_pred == 0))
    precision = tp / (tp + fp) if tp + fp else 0.0
    recall = tp / (tp + fn) if tp + fn else 0.0
    specificity = tn / (tn + fp) if tn + fp else 0.0
    f1 = 2 * precision * recall / (precision + recall) if precision + recall else 0.0
    ### END CODE HERE ###
    return {"precision": precision, "recall": recall, "specificity": specificity, "f1": f1}
```

```python
with cp.task("c01"):
    from sklearn import metrics as skm
    rng = np.random.default_rng(1)
    yt, yp = rng.integers(0, 2, 300), rng.integers(0, 2, 300)
    m = binary_metrics(yt, yp)
    check_close(m["precision"], skm.precision_score(yt, yp), message="precision")
    check_close(m["recall"], skm.recall_score(yt, yp), message="recall")
    check_close(m["f1"], skm.f1_score(yt, yp), message="F1")
    check_close(m["specificity"], skm.recall_score(1 - yt, 1 - yp), message="specificity")
    check_equal(binary_metrics(np.array([0, 0]), np.array([0, 0]))["precision"], 0.0, "mẫu số 0")
```

### Task c02 — ROC-AUC bằng đếm cặp

Viết `auc_pairwise(y, s)` = tỉ lệ cặp (positive, negative) mà positive có điểm cao hơn, cặp hoà tính 0.5.

```python
def auc_pairwise(y, s):
    ### START CODE HERE ### (≈ 3 dòng)
    sp, sn = s[y == 1], s[y == 0]
    diff = sp[:, None] - sn[None, :]
    auc = np.mean(diff > 0) + 0.5 * np.mean(diff == 0)
    ### END CODE HERE ###
    return auc
```

```python
with cp.task("c02"):
    from sklearn import metrics as skm
    rng = np.random.default_rng(2)
    y = rng.integers(0, 2, 400)
    s = np.round(rng.normal(y * 0.7, 1), 1)
    check_close(auc_pairwise(y, s), skm.roc_auc_score(y, s), tol=1e-12, message="khớp sklearn (có hoà)")
```

## Phần 5 — Mini Project

**Mini project — Báo cáo đánh giá model churn.** Đọc `customer_churn.csv` (`repo_path("data", "sample", "customer_churn.csv")`).
Chia train/valid/test 60/20/20 (`random_state=0`, `stratify`). Train Logistic Regression (one-hot + scale). Trên **valid**, chọn ngưỡng
tối thiểu chi phí với C_FP = 1 (gọi điện giữ chân), C_FN = 5 (mất khách). Báo cáo trên **test**: ROC-AUC (`test_auc`),
chi phí trung bình mỗi khách tại ngưỡng đã chọn (`test_cost_per_customer`) và so với hai chiến lược "gọi tất cả" (`cost_all`), "không gọi ai" (`cost_none`).
Yêu cầu: `test_auc ≥ 0.80` và chi phí của model **thấp hơn cả hai** chiến lược đơn giản.

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from ml_course.utils import repo_path

df = pd.read_csv(repo_path("data", "sample", "customer_churn.csv"))
X = pd.get_dummies(df.drop(columns=["customer_id", "churn"]), dtype=float)
y = df["churn"].to_numpy()
### START CODE HERE ### (≈ 12 dòng)
X_tmp, X_te, y_tmp, y_te = train_test_split(X, y, test_size=0.2, random_state=0, stratify=y)
X_tr, X_va, y_tr, y_va = train_test_split(X_tmp, y_tmp, test_size=0.25, random_state=0, stratify=y_tmp)
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=3000)).fit(X_tr, y_tr)
p_va, p_te = model.predict_proba(X_va)[:, 1], model.predict_proba(X_te)[:, 1]
cost = lambda yy, pred: (1 * np.sum((pred == 1) & (yy == 0)) + 5 * np.sum((pred == 0) & (yy == 1))) / len(yy)
ts = np.linspace(0.01, 0.99, 99)
best_t = ts[int(np.argmin([cost(y_va, (p_va >= t).astype(int)) for t in ts]))]
test_auc = roc_auc_score(y_te, p_te)
test_cost_per_customer = cost(y_te, (p_te >= best_t).astype(int))
cost_all = cost(y_te, np.ones_like(y_te))
cost_none = cost(y_te, np.zeros_like(y_te))
### END CODE HERE ###
print(f"ngưỡng = {best_t:.2f} | ROC-AUC = {test_auc:.3f} | chi phí/khách: model = {test_cost_per_customer:.3f}, gọi tất cả = {cost_all:.3f}, không gọi = {cost_none:.3f}")
```

```python
with cp.task("project"):
    check_greater(test_auc, 0.80, "ROC-AUC test ≥ 0.80")
    check_less(test_cost_per_customer, min(cost_all, cost_none), "model rẻ hơn cả hai chiến lược đơn giản")
```
