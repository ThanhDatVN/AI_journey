## Phần 3 — Coding Exercise

### Task c01 — Scaled dot-product attention có mask

Viết `attention(Q, K, V, mask=None)` (NumPy): softmax(QKᵀ/√d_k)V; `mask` bool (n_q, n_k), `False` = bị che (điểm −∞).
Trả về `(output, weights)`. Softmax phải ổn định số.

```python
def attention(Q, K, V, mask=None):
    ### START CODE HERE ### (≈ 6 dòng)
    scores = Q @ K.T / np.sqrt(Q.shape[1])
    if mask is not None:
        scores = np.where(mask, scores, -np.inf)
    scores = scores - scores.max(axis=1, keepdims=True)
    w = np.exp(scores) / np.exp(scores).sum(axis=1, keepdims=True)
    return w @ V, w
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    r = np.random.default_rng(0)
    Q, K, V = r.normal(size=(3, 4)), r.normal(size=(5, 4)), r.normal(size=(5, 2))
    out, w = attention(Q, K, V)
    check_close(w.sum(1), np.ones(3), message="mỗi hàng tổng 1")
    s = Q @ K.T / 2.0
    ref = np.exp(s) / np.exp(s).sum(1, keepdims=True)
    check_close(out, ref @ V, message="khớp công thức")
    m = np.tril(np.ones((3, 5), dtype=bool))
    _, wm = attention(Q, K, V, m)
    check_close(wm[0, 1:], np.zeros(4), message="vị trí bị che có trọng số 0")
    check_true(np.isfinite(attention(Q * 1000, K, V)[0]).all(), "ổn định số")
```

### Task c02 — Mã hoá vị trí sinusoidal

Viết `sinusoidal_encoding(T, d)` (NumPy, d chẵn): PE[p, 2i] = sin(p / 10000^(2i/d)), PE[p, 2i+1] = cos(p / 10000^(2i/d)).

```python
def sinusoidal_encoding(T, d):
    ### START CODE HERE ### (≈ 5 dòng)
    pos = np.arange(T)[:, None]
    freq = 1.0 / 10000 ** (np.arange(0, d, 2) / d)
    pe = np.zeros((T, d))
    pe[:, 0::2], pe[:, 1::2] = np.sin(pos * freq), np.cos(pos * freq)
    return pe
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    pe = sinusoidal_encoding(50, 16)
    check_equal(pe.shape, (50, 16), "shape")
    check_close(pe[0], np.tile([0.0, 1.0], 8), message="vị trí 0: sin = 0, cos = 1")
    check_close(pe[3, 2], np.sin(3 / 10000 ** (2 / 16)), message="PE[3, 2]")
    check_close(pe[10] @ pe[17], pe[30] @ pe[37], message="tích vô hướng chỉ phụ thuộc khoảng cách")
```

## Phần 5 — Mini Project

**Mini project — Sắp xếp bằng Transformer.** Huấn luyện một encoder Transformer (có mã hoá vị trí) nhận 12 chữ số và xuất dãy đã
sắp xếp tăng dần (mỗi vị trí một lớp 0–9). Đánh giá trên 2 000 chuỗi validation sinh bởi `torch.Generator().manual_seed(123)`;
lưu tỉ lệ khớp **cả chuỗi** vào `exact`. Yêu cầu `exact` ≥ 0.9 (gợi ý: AdamW + warmup + cosine như bài 03).

```python
import math

import torch
from torch import nn

device = "cuda" if torch.cuda.is_available() else "cpu"
torch.manual_seed(0)
T = 12
### START CODE HERE ### (≈ 25 dòng)
class SortNet(nn.Module):
    def __init__(self, d=64, n_heads=4, n_layers=2):
        super().__init__()
        self.emb = nn.Embedding(10, d)
        self.pos = nn.Parameter(torch.randn(T, d) * 0.02)
        layer = nn.TransformerEncoderLayer(d, n_heads, 4 * d, dropout=0.0, batch_first=True, norm_first=True)
        self.encoder = nn.TransformerEncoder(layer, n_layers, enable_nested_tensor=False)
        self.ln = nn.LayerNorm(d)
        self.head = nn.Linear(d, 10)

    def forward(self, x):
        return self.head(self.ln(self.encoder(self.emb(x) + self.pos)))

model = SortNet().to(device)
steps = 2000
opt = torch.optim.AdamW(model.parameters(), lr=1e-3)
sched = torch.optim.lr_scheduler.LambdaLR(opt, lambda s: min(1.0, (s + 1) / 100) * 0.5 * (1 + math.cos(math.pi * s / steps)))
g = torch.Generator().manual_seed(0)
for _ in range(steps):
    x = torch.randint(0, 10, (128, T), generator=g)
    opt.zero_grad()
    nn.functional.cross_entropy(model(x.to(device)).reshape(-1, 10), x.sort(1).values.to(device).reshape(-1)).backward()
    opt.step()
    sched.step()                                                    # warmup 100 bước + cosine: ổn định hơn LR cố định
model.eval()
x_val = torch.randint(0, 10, (2000, T), generator=torch.Generator().manual_seed(123))
with torch.no_grad():
    pred = model(x_val.to(device)).argmax(-1).cpu()
exact = (pred == x_val.sort(1).values).all(1).float().mean().item()
### END CODE HERE ###
print(f"khớp cả chuỗi = {exact:.3f}")
```

```python
with cp.task("project"):
    check_greater(exact, 0.9, "khớp cả chuỗi ≥ 0.9")
```
