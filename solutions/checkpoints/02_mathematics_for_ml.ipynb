### Task c01 — Gradient của logistic loss

Với $p = \sigma(X\mathbf w)$, loss $L(\mathbf w) = -\frac1n\sum_i [y_i\log p_i + (1-y_i)\log(1-p_i)]$.
Viết `logistic_grad(w, X, y)` trả về $\nabla_{\mathbf w}L = \frac1n X^\top(p - y)$ (vectorized).

```python
def logistic_grad(w, X, y):
    ### START CODE HERE ### (≈ 2 dòng)
    p = 1 / (1 + np.exp(-(X @ w)))
    grad = X.T @ (p - y) / len(y)
    ### END CODE HERE ###
    return grad
```

```python
with cp.task("c01"):
    from ml_course.utils import numerical_gradient, relative_error
    rng = np.random.default_rng(1)
    X, y, w = rng.normal(size=(30, 4)), rng.integers(0, 2, 30), rng.normal(size=4)
    def loss(w):
        p = 1 / (1 + np.exp(-(X @ w)))
        return -np.mean(y * np.log(p) + (1 - y) * np.log(1 - p))
    check_shape(logistic_grad(w, X, y), (4,))
    check_less(relative_error(logistic_grad(w, X, y), numerical_gradient(loss, w.copy())), 1e-7, "gradient check")
```

### Task c02 — Khoảng tin cậy bootstrap

Viết `bootstrap_ci(x, stat, n_boot, seed)` trả về tuple `(low, high)` — khoảng tin cậy 95% percentile bootstrap của thống kê `stat` (một hàm, vd `np.median`).

```python
def bootstrap_ci(x, stat=np.mean, n_boot=2000, seed=0):
    rng = np.random.default_rng(seed)
    x = np.asarray(x)
    ### START CODE HERE ### (≈ 3 dòng)
    idx = rng.integers(0, len(x), size=(n_boot, len(x)))
    values = np.array([stat(x[i]) for i in idx])
    ci = (np.quantile(values, 0.025), np.quantile(values, 0.975))
    ### END CODE HERE ###
    return ci
```

```python
with cp.task("c02"):
    data = np.random.default_rng(3).normal(10, 2, 400)
    lo, hi = bootstrap_ci(data, np.median)
    check_true(lo < np.median(data) < hi, "CI chứa median mẫu")
    check_true(0.1 < hi - lo < 1.0, "độ rộng CI hợp lý")
```

## Phần 4 — Mini Project

**Mini project — Linear regression bằng gradient descent thuần NumPy.** Dữ liệu có 2 feature với thang đo rất khác nhau.
1. Chuẩn hoá feature (`z = (x − mean) / std` theo cột).
2. Cài đặt GD cho $\hat y = X\mathbf w + b$ với MSE (gradient tự viết).
3. Chạy 500 vòng với learning rate phù hợp; MSE cuối trên dữ liệu train phải **< 0.3** (nhiễu thật có phương sai 0.25).

```python
rng = np.random.default_rng(7)
X_raw = np.column_stack([rng.uniform(20, 200, 500), rng.uniform(0, 5, 500)])
y = 0.03 * X_raw[:, 0] + 1.5 * X_raw[:, 1] + 4 + rng.normal(0, 0.5, 500)

### START CODE HERE ### (≈ 10 dòng)
Z = (X_raw - X_raw.mean(axis=0)) / X_raw.std(axis=0)
w, b, lr = np.zeros(2), 0.0, 0.1
for _ in range(500):
    err = Z @ w + b - y
    w -= lr * 2 * Z.T @ err / len(y)
    b -= lr * 2 * err.mean()
final_mse = np.mean((Z @ w + b - y) ** 2)
### END CODE HERE ###
print("MSE cuối:", final_mse)
```

```python
with cp.task("project"):
    check_less(final_mse, 0.3, "MSE train < 0.3")
```
