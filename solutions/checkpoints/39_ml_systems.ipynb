### Task c01 — Khoảng tin cậy của chênh lệch theo cặp

Viết `paired_ci(a, b, confidence=0.95)` trả về `(mean_diff, lo, hi)`: d = a − b (cùng fold), trung bình d và khoảng tin cậy
theo phân phối t với n − 1 bậc tự do: mean ± t_{(1+confidence)/2, n−1} · sd(d, ddof = 1)/√n.

```python
from scipy import stats


def paired_ci(a, b, confidence=0.95):
    ### START CODE HERE ### (≈ 4 dòng)
    d = np.asarray(a, float) - np.asarray(b, float)
    half = stats.t.ppf((1 + confidence) / 2, len(d) - 1) * d.std(ddof=1) / np.sqrt(len(d))
    return d.mean(), d.mean() - half, d.mean() + half
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    m, lo, hi = paired_ci([0.81, 0.79, 0.83, 0.80, 0.82], [0.80, 0.79, 0.82, 0.79, 0.80])
    check_close(m, 0.01, message="trung bình chênh lệch")
    check_close(hi - m, stats.t.ppf(0.975, 4) * np.sqrt(0.00005) / np.sqrt(5), message="nửa độ rộng KTC")
    check_close(m - lo, hi - m, message="KTC đối xứng")
    m2, lo2, hi2 = paired_ci([1, 2, 3, 4], [1, 2, 3, 4])
    check_close([m2, lo2, hi2], [0, 0, 0], message="hai mô hình giống hệt")
    check_greater(paired_ci([0.81, 0.79, 0.83, 0.80, 0.82], [0.80, 0.79, 0.82, 0.79, 0.80], 0.99)[2], hi, "độ tin cậy cao hơn → KTC rộng hơn")
```

### Task c02 — As-of join (nối đặc trưng đúng thời điểm)

Viết `as_of_join(labels, features)`: `labels` có cột `customer_id, ts, label`; `features` có cột `customer_id, ts, value`. Với mỗi dòng
của `labels`, lấy `value` của bản ghi đặc trưng **mới nhất có ts nhỏ hơn hẳn** `ts` của nhãn, cùng `customer_id` (không có → NaN).
Trả về DataFrame `labels` (giữ nguyên thứ tự dòng ban đầu) thêm cột `value`. Gợi ý: `pd.merge_asof(..., by=..., allow_exact_matches=False)`.

```python
import pandas as pd


def as_of_join(labels, features):
    ### START CODE HERE ### (≈ 5 dòng)
    left = labels.reset_index().sort_values("ts")
    right = features.sort_values("ts")
    merged = pd.merge_asof(left, right, on="ts", by="customer_id", allow_exact_matches=False, direction="backward")
    return merged.sort_values("index").set_index("index").rename_axis(None)
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    feats = pd.DataFrame({"customer_id": ["a", "a", "a", "b"], "ts": [1, 5, 9, 3], "value": [10.0, 50.0, 90.0, 30.0]})
    labs = pd.DataFrame({"customer_id": ["a", "b", "a", "a", "b"], "ts": [6, 2, 9, 12, 4], "label": [0, 1, 1, 0, 0]})
    out = as_of_join(labs, feats)
    check_equal(list(out.columns[:3]), ["customer_id", "ts", "label"], "giữ các cột của labels")
    check_equal(list(out.label), [0, 1, 1, 0, 0], "giữ thứ tự dòng ban đầu")
    check_close(out.value.values, [50.0, np.nan, 50.0, 90.0, 30.0], message="giá trị mới nhất có ts < ts nhãn (không lấy cùng thời điểm, không lấy tương lai)")
```

## Phần 5 — Mini Project

**Mini project — Gói mô hình có kiểm chứng.** Với `make_customer_churn(5000, seed=3)` (chia 70/30, `random_state=0`, stratify): huấn luyện
pipeline `ColumnTransformer` (chuẩn hoá cột số, one-hot cột danh mục) + `LogisticRegression`; lưu vào thư mục `package_dir` gồm `model.joblib`
và `metadata.json` với các khoá `features` (danh sách cột), `test_roc_auc`, `sklearn_version`, `model_sha256` (sha256 của file model.joblib).
Nạp lại thành `reloaded` sau khi **kiểm tra hash**. Yêu cầu: AUC test > 0.78; dự đoán của `reloaded` khớp mô hình gốc.

```python
import hashlib
import json
import tempfile
from pathlib import Path

import joblib
import sklearn
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from ml_course.datasets import make_customer_churn

data = make_customer_churn(5000, seed=3)
FEATURES = ["tenure_months", "contract", "internet_service", "monthly_charges", "total_charges",
            "payment_method", "support_calls", "senior", "n_products"]
NUMERIC = ["tenure_months", "monthly_charges", "total_charges", "support_calls", "n_products"]
CATEGORICAL = ["contract", "internet_service", "payment_method", "senior"]
X_tr, X_te, y_tr, y_te = train_test_split(data[FEATURES], data.churn, test_size=0.3, random_state=0, stratify=data.churn)
package_dir = Path(tempfile.mkdtemp()) / "churn_v1"
### START CODE HERE ### (≈ 14 dòng)
pre = ColumnTransformer([("num", StandardScaler(), NUMERIC), ("cat", OneHotEncoder(handle_unknown="ignore"), CATEGORICAL)])
model = make_pipeline(pre, LogisticRegression(max_iter=1000)).fit(X_tr, y_tr)
package_dir.mkdir(parents=True)
joblib.dump(model, package_dir / "model.joblib")
meta = {"features": FEATURES, "test_roc_auc": float(roc_auc_score(y_te, model.predict_proba(X_te)[:, 1])),
        "sklearn_version": sklearn.__version__,
        "model_sha256": hashlib.sha256((package_dir / "model.joblib").read_bytes()).hexdigest()}
(package_dir / "metadata.json").write_text(json.dumps(meta, indent=2), encoding="utf-8")

loaded_meta = json.loads((package_dir / "metadata.json").read_text(encoding="utf-8"))
if hashlib.sha256((package_dir / "model.joblib").read_bytes()).hexdigest() != loaded_meta["model_sha256"]:
    raise RuntimeError("hash không khớp")
reloaded = joblib.load(package_dir / "model.joblib")
### END CODE HERE ###
print(json.dumps({k: v for k, v in meta.items() if k != "features"}, indent=2))
```
