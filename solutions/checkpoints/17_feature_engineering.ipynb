### Task c01 — Target encoding có smoothing

Viết `smooth_encode(train_cat, train_y, apply_cat, m)` trả về mảng encoding cho `apply_cat`:
(tổng y + m·prior)/(số mẫu + m) với prior = trung bình train_y; category chưa thấy → prior.

```python
def smooth_encode(train_cat, train_y, apply_cat, m=10):
    ### START CODE HERE ### (≈ 4 dòng)
    prior = np.mean(train_y)
    stats = pd.DataFrame({"c": train_cat, "y": train_y}).groupby("c")["y"].agg(["sum", "count"])
    enc = (stats["sum"] + m * prior) / (stats["count"] + m)
    out = pd.Series(apply_cat).map(enc).fillna(prior).to_numpy()
    ### END CODE HERE ###
    return out
```

```python
with cp.task("c01"):
    import pandas as pd
    cats = np.array(["a"] * 100 + ["b"] * 4)
    ys = np.r_[np.r_[np.ones(30), np.zeros(70)], np.ones(4)]
    out = smooth_encode(cats, ys, np.array(["a", "b", "z"]), m=10)
    prior = ys.mean()
    check_close(out[0], (30 + 10 * prior) / 110, message="category a")
    check_close(out[1], (4 + 10 * prior) / 14, message="category b")
    check_close(out[2], prior, message="category mới → prior")
```

### Task c02 — Feature quá khứ không rò rỉ

Cho DataFrame `tx` đã sắp xếp theo thời gian với cột `customer`, `amount`. Viết `past_mean_amount(tx)` trả về Series:
trung bình `amount` của **các giao dịch trước** của cùng khách hàng (NaN nếu chưa có).

```python
def past_mean_amount(tx):
    ### START CODE HERE ### (≈ 2 dòng)
    g = tx.groupby("customer")["amount"]
    out = (g.cumsum() - tx["amount"]) / g.cumcount().replace(0, np.nan)
    ### END CODE HERE ###
    return out
```

```python
with cp.task("c02"):
    import pandas as pd
    tx = pd.DataFrame({"customer": ["A", "B", "A", "A", "B"], "amount": [10.0, 5.0, 20.0, 60.0, 7.0]})
    out = past_mean_amount(tx)
    check_true(np.isnan(out.iloc[0]) and np.isnan(out.iloc[1]), "giao dịch đầu tiên: NaN")
    check_close(out.iloc[2:].to_numpy(), [10.0, 15.0, 5.0], message="chỉ dùng quá khứ")
```

## Phần 5 — Mini Project

**Mini project — Feature engineering cho dự đoán giá nhà.** Đọc `house_prices_dirty.csv` (`repo_path("data", "sample", "house_prices_dirty.csv")`),
target `price_billion_vnd`. Chia train/test 80/20 (`random_state=0`). Xây **một Pipeline** nhận dữ liệu thô: làm sạch (tên quận, missing),
feature miền (log diện tích, diện tích/phòng ngủ...), encoding, model **tuyến tính** (Ridge, có thể log target). Đạt **MAE test ≤ 0.60** (biến `test_mae`, tỉ VND).

```python
import pandas as pd
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.impute import SimpleImputer
from sklearn.linear_model import RidgeCV
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from ml_course.utils import repo_path

df = pd.read_csv(repo_path("data", "sample", "house_prices_dirty.csv")).drop_duplicates()
X, y = df.drop(columns="price_billion_vnd"), df["price_billion_vnd"]
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2, random_state=0)
### START CODE HERE ### (≈ 25 dòng)
class Clean(TransformerMixin, BaseEstimator):
    def fit(self, X, y=None):
        self.median_area_ = X["area_m2"].median()
        return self

    def transform(self, X):
        X = X.copy()
        X["district"] = X["district"].astype(str).str.strip().str.replace("Q.", "Quận", regex=False).str.title()
        big = X["area_m2"] > 6 * self.median_area_
        X.loc[big, "area_m2"] = X.loc[big, "area_m2"] / 10
        X["has_garden"] = X["has_garden"].astype(float)
        X["log_area"] = np.log(X["area_m2"])
        X["area_per_bedroom"] = X["area_m2"] / X["bedrooms"]
        return X

cols = ["log_area", "area_per_bedroom", "bedrooms", "bathrooms", "age_years", "distance_to_center_km", "has_garden"]
pre = ColumnTransformer([("num", make_pipeline(SimpleImputer(strategy="median"), StandardScaler()), cols),
                         ("cat", OneHotEncoder(handle_unknown="ignore"), ["district"])])
pipe = Pipeline([("clean", Clean()), ("pre", pre),
                 ("model", TransformedTargetRegressor(RidgeCV(alphas=np.logspace(-3, 3, 13)), func=np.log, inverse_func=np.exp))])
pipe.fit(X_tr, y_tr)
test_mae = mean_absolute_error(y_te, pipe.predict(X_te))
### END CODE HERE ###
print(f"MAE test = {test_mae:.3f} tỉ")
```

```python
with cp.task("project"):
    check_less(test_mae, 0.60, "MAE test ≤ 0.60 tỉ")
```
