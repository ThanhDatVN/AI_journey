## Phần 3 — Coding Exercise

### Task c01 — IoU hai hộp

Viết `iou(a, b)` cho hai hộp dạng [x1, y1, x2, y2] (list hoặc tuple); trả về 0 nếu không giao nhau.

```python
def iou(a, b):
    ### START CODE HERE ### (≈ 5 dòng)
    w = max(0.0, min(a[2], b[2]) - max(a[0], b[0]))
    h = max(0.0, min(a[3], b[3]) - max(a[1], b[1]))
    inter = w * h
    union = (a[2] - a[0]) * (a[3] - a[1]) + (b[2] - b[0]) * (b[3] - b[1]) - inter
    return inter / union
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    check_close(iou([0, 0, 10, 10], [0, 0, 10, 10]), 1.0, message="trùng nhau")
    check_close(iou([0, 0, 10, 10], [5, 5, 15, 15]), 25 / 175, message="chồng 1/4")
    check_close(iou([0, 0, 10, 10], [20, 20, 30, 30]), 0.0, message="không giao")
    check_close(iou([0, 0, 4, 4], [2, 0, 6, 4]), 1 / 3, message="nửa chồng")
```

### Task c02 — Dice và IoU của mask

Viết `mask_scores(pred, true)` cho hai mảng bool cùng shape, trả về `(dice, iou)`; nếu cả hai rỗng trả về `(1.0, 1.0)`.

```python
def mask_scores(pred, true):
    ### START CODE HERE ### (≈ 5 dòng)
    inter = np.logical_and(pred, true).sum()
    total = pred.sum() + true.sum()
    if total == 0:
        return 1.0, 1.0
    return 2 * inter / total, inter / (total - inter)
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    a = np.zeros((10, 10), bool); a[:5] = True
    b = np.zeros((10, 10), bool); b[2:7] = True
    d, j = mask_scores(a, b)
    check_close(d, 2 * 30 / 100, message="Dice")
    check_close(j, 30 / 70, message="IoU")
    check_close(mask_scores(np.zeros((3, 3), bool), np.zeros((3, 3), bool)), (1.0, 1.0), message="cả hai rỗng")
    check_close(d, 2 * j / (1 + j), message="Dice = 2·IoU/(1 + IoU)")
```

## Phần 5 — Mini Project

**Mini project — Phân loại hình khi ít dữ liệu.** Chỉ có 300 ảnh train (`make_shapes(300, seed=0)`), test trên `make_shapes(1000, seed=1)`.
Dùng CNN + data augmentation phù hợp để đạt accuracy test `acc` ≥ 0.88 (không augmentation chỉ khoảng 0.73).

```python
import torch
import torch.nn.functional as F
from torch import nn

from ml_course.datasets import make_shapes

device = "cuda" if torch.cuda.is_available() else "cpu"
X, y = make_shapes(300, seed=0)
X_test, y_test = make_shapes(1000, seed=1)
X, X_test = torch.tensor(X[:, None]), torch.tensor(X_test[:, None])
y, y_test = torch.tensor(y), torch.tensor(y_test)
### START CODE HERE ### (≈ 25 dòng)
def shift_flip(x, g, max_shift=0.15):
    B = x.shape[0]
    shift = (torch.rand(2, B, generator=g) * 2 - 1) * max_shift * 2
    sx = torch.where(torch.rand(B, generator=g) < 0.5, -1.0, 1.0)
    zeros, ones = torch.zeros(B), torch.ones(B)
    theta = torch.stack([torch.stack([sx, zeros, shift[0]], 1), torch.stack([zeros, ones, shift[1]], 1)], 1)
    return F.grid_sample(x, F.affine_grid(theta, x.shape, align_corners=False), align_corners=False)

torch.manual_seed(0)
model = nn.Sequential(nn.Conv2d(1, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
                      nn.Conv2d(16, 32, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
                      nn.Conv2d(32, 64, 3, padding=1), nn.ReLU(), nn.Flatten(), nn.Linear(64 * 7 * 7, 5)).to(device)
opt = torch.optim.Adam(model.parameters(), lr=1e-3)
g = torch.Generator().manual_seed(0)
for _ in range(1500):
    model.train()
    idx = torch.randint(0, len(X), (64,), generator=g)
    opt.zero_grad()
    F.cross_entropy(model(shift_flip(X[idx], g).to(device)), y[idx].to(device)).backward()
    opt.step()
model.eval()
with torch.no_grad():
    acc = (model(X_test.to(device)).argmax(1).cpu() == y_test).float().mean().item()
### END CODE HERE ###
print(f"accuracy test = {acc:.3f}")
```

```python
with cp.task("project"):
    check_greater(acc, 0.88, "accuracy test ≥ 0.88")
```
