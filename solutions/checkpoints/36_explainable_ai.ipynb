### Task c01 — Shapley từ bảng giá trị

Viết `shapley_from_values(values, d)`: `values` là dict ánh xạ tuple các chỉ số đã sắp xếp (tập con S ⊆ {0..d−1}) → v(S).
Trả về mảng φ độ dài d theo công thức Shapley.

```python
from itertools import combinations
from math import factorial

def shapley_from_values(values, d):
    ### START CODE HERE ### (≈ 8 dòng)
    phi = np.zeros(d)
    for j in range(d):
        others = [i for i in range(d) if i != j]
        for k in range(d):
            for S in combinations(others, k):
                w = factorial(k) * factorial(d - k - 1) / factorial(d)
                phi[j] += w * (values[tuple(sorted(S + (j,)))] - values[S])
    return phi
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    v2 = {(): 0, (0,): 4, (1,): 2, (0, 1): 10}
    check_close(shapley_from_values(v2, 2), [6, 4], message="2 người chơi")
    v3 = {S: float(len(S) ** 2) for k in range(4) for S in combinations(range(3), k)}
    phi3 = shapley_from_values(v3, 3)
    check_close(phi3, [3, 3, 3], message="đối xứng: trò chơi chỉ phụ thuộc |S|")
    check_close(phi3.sum(), v3[(0, 1, 2)] - v3[()], message="hiệu quả")
```

### Task c02 — Partial dependence

Viết `pdp_values(predict, X, feature, grid)` (X là DataFrame): với mỗi giá trị v trong grid, đặt cột `feature` = v cho mọi dòng, trả về
trung bình `predict` — mảng độ dài len(grid).

```python
import pandas as pd

def pdp_values(predict, X, feature, grid):
    ### START CODE HERE ### (≈ 1 dòng)
    return np.array([predict(X.assign(**{feature: v})).mean() for v in grid])
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    Xd = pd.DataFrame({"a": [1.0, 2.0, 3.0], "b": [0.0, 1.0, 2.0]})
    lin = lambda D: 2 * D["a"].values + D["b"].values
    check_close(pdp_values(lin, Xd, "a", [0, 1, 5]), [1.0, 3.0, 11.0], message="mô hình tuyến tính: PDP = 2v + E[b]")
    inter = lambda D: D["a"].values * D["b"].values
    check_close(pdp_values(inter, Xd, "a", [2]), [2.0], message="tương tác: 2·E[b]")
```

## Phần 5 — Mini Project

**Mini project — Dùng giải thích để gỡ lỗi.** Bảng churn có thêm 3 cột "mới": `survey_score`, `app_sessions`, `retention_call` — một trong
số đó là **rò rỉ**. Train `HistGradientBoostingClassifier` (cho sẵn), tính permutation importance (ROC-AUC) trên tập test, xác định cột
rò rỉ (lưu tên vào `leak_feature`), train lại không có nó và lưu ROC-AUC test vào `auc_clean`.

```python
import pandas as pd
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.inspection import permutation_importance
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import train_test_split

from ml_course.datasets import make_customer_churn

raw = make_customer_churn(20000, seed=0)
r = np.random.default_rng(1)
y = raw.churn.values
D = pd.DataFrame({
    "tenure": raw.tenure_months, "monthly": raw.monthly_charges, "support_calls": raw.support_calls,
    "month_to_month": (raw.contract == "month-to-month").astype(int), "fiber": (raw.internet_service == "fiber").astype(int),
    "survey_score": r.normal(7, 1.5, len(raw)) - 0.3 * y,
    "app_sessions": r.poisson(8, len(raw)),
    "retention_call": np.where(y == 1, r.random(len(raw)) < 0.6, r.random(len(raw)) < 0.03).astype(int),
}).astype(float)
D_train, D_test, y_train, y_test = train_test_split(D, y, test_size=0.3, random_state=0, stratify=y)
### START CODE HERE ### (≈ 8 dòng)
model = HistGradientBoostingClassifier(max_iter=300, learning_rate=0.05, random_state=0).fit(D_train, y_train)
imp = permutation_importance(model, D_test, y_test, scoring="roc_auc", n_repeats=5, random_state=0)
ranking = pd.Series(imp.importances_mean, index=D.columns).sort_values(ascending=False)
print(ranking.round(3))
leak_feature = ranking.index[0]
keep = [c for c in D.columns if c != leak_feature]
clean = HistGradientBoostingClassifier(max_iter=300, learning_rate=0.05, random_state=0).fit(D_train[keep], y_train)
auc_clean = roc_auc_score(y_test, clean.predict_proba(D_test[keep])[:, 1])
### END CODE HERE ###
print(f"cột rò rỉ: {leak_feature} | ROC-AUC sau khi bỏ = {auc_clean:.3f}")
```
