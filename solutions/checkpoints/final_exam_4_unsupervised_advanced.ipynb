## Phần 3 — Coding Exercise

### Task c01 — Một vòng K-Means

Viết `kmeans_step(X, centers)` → `(labels, new_centers)`: gán mỗi điểm vào tâm gần nhất (Euclid), rồi cập nhật tâm = trung bình
các điểm của cụm (cụm rỗng giữ tâm cũ).

```python
def kmeans_step(X, centers):
    ### START CODE HERE ### (≈ 6 dòng)
    X, centers = np.asarray(X, float), np.asarray(centers, float)
    d = ((X[:, None, :] - centers[None, :, :]) ** 2).sum(-1)
    labels = d.argmin(axis=1)
    new = centers.copy()
    for k in range(len(centers)):
        if np.any(labels == k):
            new[k] = X[labels == k].mean(axis=0)
    return labels, new
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    X = np.array([[0.0, 0], [0, 1], [10, 10], [10, 11]])
    labels, centers = kmeans_step(X, np.array([[0.0, 0], [10, 10]]))
    check_equal(list(labels), [0, 0, 1, 1], "gán cụm")
    check_close(centers, [[0, 0.5], [10, 10.5]], message="cập nhật tâm")
    _, kept = kmeans_step(X, np.array([[0.0, 0], [10, 10], [100, 100]]))
    check_close(kept[2], [100, 100], message="cụm rỗng giữ tâm cũ")
```

### Task c02 — PCA bằng SVD

Viết `pca(X, k)` → `(Z, explained_ratio)`: trừ trung bình theo cột, SVD, chiếu lên k thành phần đầu (Z = X_c · V_k),
explained_ratio = σᵢ² / Σσ² của k thành phần đầu.

```python
def pca(X, k):
    ### START CODE HERE ### (≈ 4 dòng)
    Xc = np.asarray(X, float) - np.mean(X, axis=0)
    U, s, Vt = np.linalg.svd(Xc, full_matrices=False)
    ratio = s ** 2 / np.sum(s ** 2)
    return Xc @ Vt[:k].T, ratio[:k]
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    from sklearn.decomposition import PCA
    rng = np.random.default_rng(0)
    X = rng.normal(size=(100, 4)) @ rng.normal(size=(4, 4))
    Z, ratio = pca(X, 2)
    ref = PCA(2).fit(X)
    check_close(ratio, ref.explained_variance_ratio_, message="tỉ lệ phương sai giải thích")
    check_close(np.abs(Z), np.abs(ref.transform(X)), tol=1e-6, message="toạ độ chiếu (sai khác dấu cho phép)")
```

ARI > 0.85.

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score, silhouette_score
from sklearn.preprocessing import StandardScaler

from ml_course.utils import repo_path

customers = pd.read_csv(repo_path("data", "sample", "customer_segments.csv"))
FEATURES = ["annual_income_musd", "spending_score", "age", "visits_per_month"]
### START CODE HERE ### (≈ 5 dòng)
X = StandardScaler().fit_transform(customers[FEATURES])
scores = {k: silhouette_score(X, KMeans(k, n_init=10, random_state=0).fit_predict(X)) for k in range(2, 9)}
best_k = max(scores, key=scores.get)
labels = KMeans(best_k, n_init=10, random_state=0).fit_predict(X)
ari = adjusted_rand_score(customers.true_segment, labels)
### END CODE HERE ###
print(f"k = {best_k} | ARI = {ari:.3f}")
```

```python
with cp.task("project"):
    check_equal(best_k, 5, "silhouette chọn k = 5")
    check_greater(ari, 0.85, "ARI > 0.85")
    check_equal(len(labels), 600, "mọi khách được gán cụm")
```
