## Phần 3 — Coding Exercise

### Task c01 — Multinomial NB fit

Viết `mnb_fit(X, y, alpha)` (X là ma trận đếm dày `(n, V)`, y nhãn) trả về `(classes, log_theta, log_prior)` với
$\log\theta_{cw} = \log\frac{N_{cw} + \alpha}{\sum_{w'} N_{cw'} + \alpha V}$.

```python
def mnb_fit(X, y, alpha=1.0):
    classes = np.unique(y)
    ### START CODE HERE ### (≈ 3 dòng)
    counts = np.array([X[y == c].sum(axis=0) for c in classes]) + alpha
    log_theta = np.log(counts / counts.sum(axis=1, keepdims=True))
    log_prior = np.log([np.mean(y == c) for c in classes])
    ### END CODE HERE ###
    return classes, log_theta, log_prior
```

```python
with cp.task("c01"):
    from sklearn.naive_bayes import MultinomialNB
    rng = np.random.default_rng(3)
    X = rng.poisson(1.0, size=(60, 8)).astype(float)
    y = rng.integers(0, 3, 60)
    classes, lt, lp = mnb_fit(X, y, alpha=0.5)
    ref = MultinomialNB(alpha=0.5).fit(X, y)
    check_close(lt, ref.feature_log_prob_, message="log θ khớp sklearn")
    check_close(lp, ref.class_log_prior_, message="log prior khớp sklearn")
```

### Task c02 — Posterior ổn định số học

Viết `posterior_from_scores(S)` — đổi ma trận log-score `(n, C)` thành xác suất posterior (mỗi hàng tổng 1)
**ổn định số học** (log-sum-exp hoặc trừ max).

```python
def posterior_from_scores(S):
    ### START CODE HERE ### (≈ 2 dòng)
    E = np.exp(S - S.max(axis=1, keepdims=True))
    P = E / E.sum(axis=1, keepdims=True)
    ### END CODE HERE ###
    return P
```

```python
with cp.task("c02"):
    P = posterior_from_scores(np.array([[-2300.0, -2300.0 - np.log(3)], [0.0, 0.0]]))
    check_close(P[0], [0.75, 0.25], message="ổn định với log-score rất âm")
    check_close(P[1], [0.5, 0.5], message="hai lớp bằng nhau")
```

## Phần 5 — Mini Project

**Mini project — Bộ phân loại cảm xúc 3 lớp.** Đọc `reviews_vi.csv` (`repo_path("data", "sample", "reviews_vi.csv")`), target `sentiment`
(pos/neg/neu). Chia train/test 70/30 (`random_state=1`, `stratify`). Xây Pipeline `CountVectorizer` + Naive Bayes,
chọn hyperparameter bằng CV trên train. Đạt **macro-F1 test ≥ 0.97** (biến `test_f1`).

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import f1_score
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.naive_bayes import ComplementNB, MultinomialNB
from sklearn.pipeline import make_pipeline
from ml_course.utils import repo_path

df = pd.read_csv(repo_path("data", "sample", "reviews_vi.csv"))
### START CODE HERE ### (≈ 6 dòng)
tr, te = train_test_split(df, test_size=0.3, random_state=1, stratify=df["sentiment"])
grid = GridSearchCV(make_pipeline(CountVectorizer(), ComplementNB()),
                    {"countvectorizer__ngram_range": [(1, 1), (1, 2)], "complementnb__alpha": [0.01, 0.1, 1.0]},
                    cv=5, scoring="f1_macro").fit(tr["text"], tr["sentiment"])
test_f1 = f1_score(te["sentiment"], grid.predict(te["text"]), average="macro")
### END CODE HERE ###
print(grid.best_params_, "macro-F1 test:", round(test_f1, 4))
```

```python
with cp.task("project"):
    check_greater(test_f1, 0.97, "macro-F1 test ≥ 0.97")
```
