## Phần 3 — Coding Exercise

### Task c01 — Hàm kích hoạt và đạo hàm

Viết `leaky_relu(x, alpha)` và `leaky_relu_grad(x, alpha)` (vector hoá, không vòng lặp).

```python
def leaky_relu(x, alpha=0.01):
    ### START CODE HERE ### (≈ 1 dòng)
    return np.where(x > 0, x, alpha * x)
    ### END CODE HERE ###

def leaky_relu_grad(x, alpha=0.01):
    ### START CODE HERE ### (≈ 1 dòng)
    return np.where(x > 0, 1.0, alpha)
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    x = np.array([-2.0, -0.5, 0.5, 3.0])
    check_close(leaky_relu(x, 0.1), [-0.2, -0.05, 0.5, 3.0], message="leaky ReLU")
    num = (leaky_relu(x + 1e-6, 0.1) - leaky_relu(x - 1e-6, 0.1)) / 2e-6
    check_close(leaky_relu_grad(x, 0.1), num, tol=1e-6, message="đạo hàm khớp sai phân hữu hạn")
```

### Task c02 — Một bước Adam

Viết `adam_step(theta, g, m, s, t, lr, b1=0.9, b2=0.999, eps=1e-8)` trả về `(theta_new, m_new, s_new)` (có hiệu chỉnh bias, t bắt đầu từ 1).

```python
def adam_step(theta, g, m, s, t, lr, b1=0.9, b2=0.999, eps=1e-8):
    ### START CODE HERE ### (≈ 5 dòng)
    m = b1 * m + (1 - b1) * g
    s = b2 * s + (1 - b2) * g**2
    m_hat, s_hat = m / (1 - b1**t), s / (1 - b2**t)
    theta = theta - lr * m_hat / (np.sqrt(s_hat) + eps)
    ### END CODE HERE ###
    return theta, m, s
```

```python
with cp.task("c02"):
    th, m, s = adam_step(np.array([1.0]), np.array([2.0]), np.zeros(1), np.zeros(1), 1, 0.1, eps=0.0)
    check_close(th, [0.9], message="bước đầu có độ lớn η")
    th2, _, _ = adam_step(th, np.array([-1.0]), m, s, 2, 0.1, eps=0.0)
    check_close(th2, [0.8734], tol=1e-4, message="bước thứ hai")
```

## Phần 5 — Mini Project

**Mini project — Huấn luyện mạng SÂU cho chữ số.** `load_digits()`, chia train/test 75/25 (`random_state=0`, stratify), chuẩn hoá theo train.
Mạng phải có **ít nhất 6 lớp ẩn** (`n_hidden_layers`) và đạt accuracy test `test_acc` ≥ 0.95. Dùng những gì đã học (khởi tạo, chuẩn hoá,
optimizer phù hợp).

```python
import torch
from torch import nn
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X, y = load_digits(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, random_state=0, stratify=y)
sc = StandardScaler().fit(X_tr)
Xt, Xe = (torch.tensor(sc.transform(a), dtype=torch.float32) for a in (X_tr, X_te))
yt, ye = torch.tensor(y_tr), torch.tensor(y_te)
torch.manual_seed(0)
### START CODE HERE ### (≈ 20 dòng)
layers, d = [], 64
n_hidden_layers = 6
for _ in range(n_hidden_layers):
    lin = nn.Linear(d, 64)
    nn.init.kaiming_normal_(lin.weight, nonlinearity="relu")
    layers += [lin, nn.BatchNorm1d(64), nn.ReLU()]
    d = 64
model = nn.Sequential(*layers, nn.Linear(d, 10))
opt = torch.optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-4)
for epoch in range(30):
    model.train()
    perm = torch.randperm(len(Xt))
    for s in range(0, len(perm), 32):
        idx = perm[s:s + 32]
        opt.zero_grad()
        nn.functional.cross_entropy(model(Xt[idx]), yt[idx]).backward()
        opt.step()
model.eval()
with torch.no_grad():
    test_acc = (model(Xe).argmax(1) == ye).float().mean().item()
### END CODE HERE ###
print(f"{n_hidden_layers} lớp ẩn | accuracy test = {test_acc:.4f}")
```

```python
with cp.task("project"):
    check_true(n_hidden_layers >= 6, "≥ 6 lớp ẩn")
    check_greater(test_acc, 0.95, "accuracy test ≥ 0.95")
```
