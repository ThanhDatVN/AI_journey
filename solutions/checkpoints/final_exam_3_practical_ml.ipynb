### Task c01 — Precision, recall tại một ngưỡng

Viết `precision_recall_at(y_true, scores, threshold)` → `(precision, recall)` với dự đoán dương khi score ≥ threshold;
không có dự đoán dương thì precision = 0.

```python
def precision_recall_at(y_true, scores, threshold):
    ### START CODE HERE ### (≈ 5 dòng)
    y, pred = np.asarray(y_true), np.asarray(scores) >= threshold
    tp = int(np.sum(pred & (y == 1)))
    fp = int(np.sum(pred & (y == 0)))
    fn = int(np.sum(~pred & (y == 1)))
    return (tp / (tp + fp) if tp + fp else 0.0), (tp / (tp + fn) if tp + fn else 0.0)
    ### END CODE HERE ###
```

```python
with cp.task("c01"):
    y = np.array([1, 1, 0, 0, 1])
    s = np.array([0.9, 0.4, 0.6, 0.1, 0.8])
    check_close(precision_recall_at(y, s, 0.5), (2 / 3, 2 / 3), message="ngưỡng 0.5")
    check_close(precision_recall_at(y, s, 0.95), (0.0, 0.0), message="không có dự đoán dương")
    check_close(precision_recall_at(y, s, 0.0), (0.6, 1.0), message="mọi dự đoán dương")
```

### Task c02 — Ngưỡng tối ưu theo chi phí

Viết `best_threshold(y_true, proba, cost_fn, cost_fp, grid)`: với mỗi ngưỡng t trong `grid`, chi phí = cost_fn·(số bỏ sót) + cost_fp·(số báo động giả)
(dương khi proba ≥ t); trả về `(t_tốt_nhất, chi_phí_nhỏ_nhất)` (hoà → ngưỡng nhỏ hơn).

```python
def best_threshold(y_true, proba, cost_fn, cost_fp, grid):
    ### START CODE HERE ### (≈ 6 dòng)
    y, proba = np.asarray(y_true), np.asarray(proba)
    best = (None, np.inf)
    for t in sorted(grid):
        pred = proba >= t
        cost = cost_fn * np.sum(~pred & (y == 1)) + cost_fp * np.sum(pred & (y == 0))
        if cost < best[1]:
            best = (t, float(cost))
    return best
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    y = np.array([1, 0, 0, 1, 0])
    p = np.array([0.3, 0.2, 0.6, 0.8, 0.1])
    check_close(best_threshold(y, p, cost_fn=10, cost_fp=1, grid=[0.15, 0.25, 0.5, 0.7]), (0.25, 1.0), message="chi phí bỏ sót cao → ngưỡng thấp")
    check_close(best_threshold(y, p, cost_fn=1, cost_fp=10, grid=[0.15, 0.25, 0.5, 0.7]), (0.7, 1.0), message="chi phí báo động giả cao → ngưỡng cao")
```

động gì). Yêu cầu: PR-AUC validation > 0.06; ngưỡng < 0.5; chi phí test < 80% baseline.

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import average_precision_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from ml_course.utils import repo_path

tx = pd.read_csv(repo_path("data", "sample", "fraud_transactions.csv")).assign(log_amount=lambda d: np.log(d.amount_vnd))
NUMERIC = ["log_amount", "hour", "distance_from_home_km", "n_tx_last_24h", "card_age_days"]
CATEGORICAL = ["merchant_category", "is_foreign"]
GRID = [0.005, 0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.5]
rest, test = train_test_split(tx, test_size=0.3, random_state=0, stratify=tx.is_fraud)
train, valid = train_test_split(rest, test_size=0.3, random_state=0, stratify=rest.is_fraud)
### START CODE HERE ### (≈ 12 dòng)
model = make_pipeline(ColumnTransformer([("num", StandardScaler(), NUMERIC), ("cat", OneHotEncoder(handle_unknown="ignore"), CATEGORICAL)]),
                      LogisticRegression(max_iter=2000)).fit(train, train.is_fraud)
p_val, p_test = model.predict_proba(valid)[:, 1], model.predict_proba(test)[:, 1]
val_pr_auc = average_precision_score(valid.is_fraud, p_val)


def cost(y, proba, t):
    pred = proba >= t
    return 50 * np.sum(~pred & (y == 1)) + 1 * np.sum(pred & (y == 0))


threshold = min(GRID, key=lambda t: cost(valid.is_fraud.values, p_val, t))
test_cost = cost(test.is_fraud.values, p_test, threshold)
baseline_cost = 50 * int(test.is_fraud.sum())
### END CODE HERE ###
print(f"PR-AUC valid {val_pr_auc:.3f} | ngưỡng {threshold} | chi phí test {test_cost} vs không báo động {baseline_cost}")
```
