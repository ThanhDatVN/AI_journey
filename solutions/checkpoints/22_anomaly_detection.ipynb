## Phần 3 — Coding Exercise

### Task c01 — Chọn ε bằng F1

Viết `select_threshold(y_val, p_val, n_steps=1000)` trả về `(best_eps, best_f1)`: thử `n_steps` giá trị ε đều nhau giữa min và max
của `p_val`, dự đoán bất thường khi `p_val < eps`, giữ ε có F1 lớn nhất.

```python
from sklearn.metrics import f1_score

def select_threshold(y_val, p_val, n_steps=1000):
    best_eps, best_f1 = 0.0, 0.0
    ### START CODE HERE ### (≈ 5 dòng)
    for eps in np.linspace(p_val.min(), p_val.max(), n_steps):
        f1 = f1_score(y_val, p_val < eps, zero_division=0)
        if f1 > best_f1:
            best_eps, best_f1 = eps, f1
    ### END CODE HERE ###
    return best_eps, best_f1
```

```python
with cp.task("c01"):
    y = np.array([0, 0, 0, 0, 1, 1])
    p = np.array([0.9, 0.8, 0.85, 0.7, 0.1, 0.2])
    eps, f1 = select_threshold(y, p)
    check_close(f1, 1.0, message="tách hoàn hảo")
    check_true(0.2 < eps <= 0.7, "ε giữa hai nhóm")
    y2 = np.array([0, 0, 1, 0, 1])
    p2 = np.array([0.5, 0.05, 0.1, 0.9, 0.2])
    eps2, f12 = select_threshold(y2, p2)
    check_close(f12, 0.8, tol=1e-9, message="F1 tốt nhất khi một điểm bình thường lẫn vào")
```

### Task c02 — Robust z-score

Viết `robust_zscore(x)` = 0.6745·(x − median)/MAD với MAD = median(|x − median|).

```python
def robust_zscore(x):
    x = np.asarray(x, float)
    ### START CODE HERE ### (≈ 3 dòng)
    med = np.median(x)
    mad = np.median(np.abs(x - med))
    return 0.6745 * (x - med) / mad
    ### END CODE HERE ###
```

```python
with cp.task("c02"):
    check_close(robust_zscore([10, 11, 12, 13, 40])[-1], 0.6745 * 28, tol=1e-9, message="robust z của 40")
    rng = np.random.default_rng(0)
    x = np.r_[rng.normal(100, 10, 900), rng.normal(300, 30, 100)]
    rz = robust_zscore(x)
    check_greater((np.abs(rz[900:]) > 3.5).mean(), 0.95, "bắt hầu hết 10% outlier (không bị masking)")
    check_less((np.abs(rz[:900]) > 3.5).mean(), 0.01, "ít báo nhầm")
```

```python
import pandas as pd
from sklearn.metrics import average_precision_score
from sklearn.neighbors import LocalOutlierFactor
from sklearn.preprocessing import StandardScaler


def make_tx(n_normal, n_per, patterns, seed):
    r = np.random.default_rng(seed)
    def normal(n):
        return np.c_[r.normal(12.7, 0.9, n), np.log1p(r.exponential(8, n)), r.poisson(2, n),
                     np.log(r.uniform(30, 3650, n)), r.random(n) < 0.04].astype(float)
    X, kind = [normal(n_normal)], [np.zeros(n_normal, int)]
    for k in patterns:
        d = normal(n_per)
        if k == 1:
            d[:, 0], d[:, 1], d[:, 4] = r.normal(15, 0.6, n_per), np.log1p(r.exponential(300, n_per)), r.random(n_per) < 0.6
        if k == 2:
            d[:, 0], d[:, 2] = r.normal(9.5, 0.5, n_per), r.poisson(15, n_per)
        if k == 3:
            d[:, 3], d[:, 0], d[:, 2] = np.log(r.uniform(1, 7, n_per)), r.normal(14, 0.5, n_per), r.poisson(5, n_per)
        X.append(d)
        kind.append(np.full(n_per, k))
    return np.vstack(X), np.concatenate(kind)


X_tr, kind_tr = make_tx(20000, 60, [1, 2], seed=0)        # kind 0 = bình thường, 1 = A, 2 = B
X_te, kind_te = make_tx(20000, 40, [1, 2, 3], seed=1)     # 3 = C (chỉ có ở test)
y_te = (kind_te > 0).astype(int)
### START CODE HERE ### (≈ 4 dòng)
scaler = StandardScaler().fit(X_tr[kind_tr == 0])
lof = LocalOutlierFactor(n_neighbors=35, novelty=True).fit(scaler.transform(X_tr[kind_tr == 0]))
score = -lof.score_samples(scaler.transform(X_te))
### END CODE HERE ###
top = np.argsort(-score)[:200]
pr_auc = average_precision_score(y_te, score)
recall_c = (kind_te[top] == 3).sum() / (kind_te == 3).sum()
print(f"PR-AUC = {pr_auc:.3f} | recall@200 kiểu C = {recall_c:.2f}")
```
