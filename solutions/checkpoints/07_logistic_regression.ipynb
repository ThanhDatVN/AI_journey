## Phần 3 — Coding Exercise

### Task c01 — Cost và gradient của Logistic Regression

Viết `logistic_cost_grad(theta, X, y)` (X đã có cột bias) trả về `(J, grad)` với
$J = -\frac1m[y^\top\log h + (1-y)^\top\log(1-h)]$ (clip $h$ vào $[10^{-15}, 1 - 10^{-15}]$), $\nabla J = \frac1m X^\top(h - y)$.

```python
def logistic_cost_grad(theta, X, y):
    ### START CODE HERE ### (≈ 3 dòng)
    h = np.clip(1 / (1 + np.exp(-(X @ theta))), 1e-15, 1 - 1e-15)
    J = -np.mean(y * np.log(h) + (1 - y) * np.log(1 - h))
    grad = X.T @ (h - y) / len(y)
    ### END CODE HERE ###
    return J, grad
```

```python
with cp.task("c01"):
    from ml_course.utils import numerical_gradient, relative_error
    rng = np.random.default_rng(4)
    X = np.c_[np.ones(40), rng.normal(size=(40, 3))]
    y = rng.integers(0, 2, 40).astype(float)
    J0, _ = logistic_cost_grad(np.zeros(4), X, y)
    check_close(J0, np.log(2), message="J(0) = ln 2")
    t = rng.normal(size=4)
    num = numerical_gradient(lambda th: logistic_cost_grad(th, X, y)[0], t.copy())
    check_less(relative_error(logistic_cost_grad(t, X, y)[1], num), 1e-6, "gradient check")
```

### Task c02 — Softmax ổn định số học

Viết `softmax_rows(Z)` — softmax theo từng hàng, trừ max mỗi hàng trước khi `exp`.

```python
def softmax_rows(Z):
    ### START CODE HERE ### (≈ 2 dòng)
    E = np.exp(Z - Z.max(axis=1, keepdims=True))
    P = E / E.sum(axis=1, keepdims=True)
    ### END CODE HERE ###
    return P
```

```python
with cp.task("c02"):
    P = softmax_rows(np.array([[1.0, 2.0, 3.0], [1000.0, 1000.0, 1000.0], [0.0, np.log(3), -50.0]]))
    check_close(P.sum(axis=1), [1, 1, 1], message="mỗi hàng tổng = 1")
    check_close(P[1], [1/3, 1/3, 1/3], message="ổn định với giá trị lớn")
    check_close(P[2, :2], [0.25, 0.75], tol=1e-9, message="softmax([0, ln3, ...])")
```

## Phần 5 — Mini Project

**Mini project — Phân loại 10 chữ số với Softmax Regression from scratch.** Dùng `load_digits_data()` (X đã chia 16),
chia train/test 75/25 (`random_state=0`, `stratify=y`). Cài đặt và train Softmax Regression (có L2) bằng gradient descent
**không dùng sklearn cho model**. Đạt **accuracy test ≥ 0.95** (biến `test_acc`).

```python
from sklearn.model_selection import train_test_split
from ml_course.datasets import load_digits_data

X, y, _ = load_digits_data()
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, random_state=0, stratify=y)
### START CODE HERE ### (≈ 12 dòng)
C = 10
Y = np.eye(C)[y_tr]
W, b = np.zeros((X.shape[1], C)), np.zeros(C)
lr, lam, m = 0.5, 1.0, len(X_tr)
for _ in range(500):
    Z = X_tr @ W + b
    E = np.exp(Z - Z.max(axis=1, keepdims=True))
    A = E / E.sum(axis=1, keepdims=True)
    dZ = (A - Y) / m
    W -= lr * (X_tr.T @ dZ + lam / m * W)
    b -= lr * dZ.sum(axis=0)
test_acc = np.mean(np.argmax(X_te @ W + b, axis=1) == y_te)
### END CODE HERE ###
print("accuracy test:", test_acc)
```

```python
with cp.task("project"):
    check_greater(test_acc, 0.949, "accuracy test ≥ 0.95")
```
