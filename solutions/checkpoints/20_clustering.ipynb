## Phần 3 — Coding Exercise

### Task c01 — Bước gán và cập nhật

Viết `lloyd_step(X, centers)` trả về `(labels, new_centers)`: gán mỗi điểm về tâm gần nhất (Euclid), tâm mới = trung bình các điểm
của cụm (giữ tâm cũ nếu cụm rỗng).

```python
def lloyd_step(X, centers):
    ### START CODE HERE ### (≈ 3 dòng)
    labels = ((X[:, None, :] - centers[None]) ** 2).sum(-1).argmin(1)
    new_centers = np.array([X[labels == k].mean(0) if np.any(labels == k) else centers[k] for k in range(len(centers))])
    ### END CODE HERE ###
    return labels, new_centers
```

```python
with cp.task("c01"):
    X = np.array([[0, 0], [0, 1], [10, 10], [10, 11.0]])
    labels, c = lloyd_step(X, np.array([[0, 0], [10, 10.0]]))
    check_equal(labels.tolist(), [0, 0, 1, 1], "gán")
    check_close(c, [[0, 0.5], [10, 10.5]], message="tâm mới")
    _, c2 = lloyd_step(X, np.array([[0, 0], [10, 10.0], [100, 100.0]]))
    check_close(c2[2], [100, 100], message="cụm rỗng giữ tâm cũ")
```

### Task c02 — Silhouette một điểm

Viết `silhouette_point(i, X, labels)` (khoảng cách Euclid) theo định nghĩa s(i) = (b − a)/max(a, b).

```python
def silhouette_point(i, X, labels):
    ### START CODE HERE ### (≈ 5 dòng)
    d = np.linalg.norm(X - X[i], axis=1)
    same = (labels == labels[i]) & (np.arange(len(X)) != i)
    a = d[same].mean()
    b = min(d[labels == c].mean() for c in np.unique(labels) if c != labels[i])
    s = (b - a) / max(a, b)
    ### END CODE HERE ###
    return s
```

```python
with cp.task("c02"):
    from sklearn.metrics import silhouette_samples
    rng = np.random.default_rng(3)
    X = rng.normal(size=(40, 2))
    labels = rng.integers(0, 3, 40)
    ref = silhouette_samples(X, labels)
    check_close([silhouette_point(i, X, labels) for i in range(5)], ref[:5], tol=1e-9, message="khớp sklearn")
```

## Phần 5 — Mini Project

**Mini project — Phát hiện nhóm và điểm bất thường.** Dữ liệu tổng hợp: hai cụm hình mặt trăng (`make_moons(n_samples=800, noise=0.05, random_state=11)`)
cộng 40 điểm nhiễu đều trong hộp [−1.5, 2.5] × [−1, 1.5] (`np.random.default_rng(11)`). Dùng DBSCAN hoặc HDBSCAN (chọn tham số hợp lý, có thể dùng k-distance)
để tìm cụm. Yêu cầu: ARI trên 800 điểm gốc **≥ 0.95** (biến `ari`) và **≥ 60%** điểm nhiễu thêm vào được gán −1 (biến `noise_recall`).

```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
from sklearn.metrics import adjusted_rand_score

X0, y0 = make_moons(n_samples=800, noise=0.05, random_state=11)
noise_pts = np.random.default_rng(11).uniform([-1.5, -1], [2.5, 1.5], (40, 2))
X = np.vstack([X0, noise_pts])
### START CODE HERE ### (≈ 4 dòng)
labels = DBSCAN(eps=0.12, min_samples=6).fit_predict(X)
ari = adjusted_rand_score(y0, labels[:800])
noise_recall = np.mean(labels[800:] == -1)
### END CODE HERE ###
print(f"số cụm = {labels.max() + 1} | ARI = {ari:.3f} | tỉ lệ điểm nhiễu bị gán −1 = {noise_recall:.0%}")
```

```python
with cp.task("project"):
    check_greater(ari, 0.95, "ARI ≥ 0.95")
    check_greater(noise_recall, 0.6, "≥ 60% điểm nhiễu được phát hiện")
```
