### 1. Hàm Sigmoid

Các bạn có thể bỏ qua phần này vì đã có ví dụ trong lý thuyết.

<img src="images/activation-sigmoid.png" style="width:30%;height:30%;">

````python
def sigmoid_function(data):
    ### START CODE HERE ### (≈ 1 dòng)
    result = 1 / (1 + np.exp(-data))
    ### END CODE HERE ###
    return result


sigmoid_function(data)
````

**Đầu ra kỳ vọng:**

```
array([0.73105858, 0.99330715, 0.01798621, 0.95257413, 0.11920292])
```

### 2. Hàm tanh

<img src="images/activation_tanh.png" style="width:30%;height:30%;">

*Gợi ý:* tự viết theo công thức $\frac{e^x - e^{-x}}{e^x + e^{-x}}$ (đừng dùng `np.tanh` — dùng nó để kiểm tra).

````python
def tanh_function(data):
    ### START CODE HERE ### (≈ 1 dòng)
    result = (np.exp(data) - np.exp(-data)) / (np.exp(data) + np.exp(-data))
    ### END CODE HERE ###
    return result


tanh_function(data)
````

**Đầu ra kỳ vọng:**

```
array([ 0.76159416,  0.9999092 , -0.9993293 ,  0.99505475, -0.96402758])
```

````python
check_close(tanh_function(data), [0.76159416, 0.9999092, -0.9993293, 0.99505475, -0.96402758], tol=1e-7, message="tanh")
check_close(tanh_function(np.linspace(-3, 3, 7)), np.tanh(np.linspace(-3, 3, 7)), message="khớp np.tanh")
````

### 3. Hàm ReLU

<img src="images/activation_relu.png" style="width:30%;height:30%;">

````python
def ReLU_function(data):
    ### START CODE HERE ### (≈ 1 dòng)
    result = np.maximum(0, data)
    ### END CODE HERE ###
    return result


ReLU_function(data)
````

**Đầu ra kỳ vọng:**

```
array([1, 5, 0, 3, 0])
```

````python
check_close(ReLU_function(data), [1, 5, 0, 3, 0], message="ReLU")
````

### 4. Hàm PReLU

<img src="images/activation_prelu.png" style="width:30%;height:30%;">

````python
def PReLU_function(data, alpha=0.1):
    ### START CODE HERE ### (≈ 1 dòng)
    result = np.where(data > 0, data, alpha * data)
    ### END CODE HERE ###
    return result


PReLU_function(data)
````

**Đầu ra kỳ vọng:**

```
[1, 5, -0.4, 3, -0.2]
```

````python
check_close(PReLU_function(data), [1, 5, -0.4, 3, -0.2], message="PReLU")
check_close(PReLU_function(data, alpha=0.0), ReLU_function(data), message="alpha = 0 → ReLU")
````

### 5. Hàm ELU

<img src="images/activation_elu.png" style="width:30%;height:30%;">

````python
def ELU_function(data, alpha=0.1):
    ### START CODE HERE ### (≈ 1 dòng)
    result = np.where(data > 0, data, alpha * (np.exp(np.minimum(data, 0)) - 1))
    ### END CODE HERE ###
    return result


ELU_function(data)
````

**Đầu ra kỳ vọng:**

```
[1, 5, -0.09816843611112658, 3, -0.08646647167633874]
```

*Lưu ý:* `np.where` tính **cả hai** nhánh với mọi phần tử; `np.exp(1000)` ở nhánh không dùng vẫn gây cảnh báo tràn số → kẹp đầu vào của `exp` bằng `np.minimum(data, 0)`.

````python
check_close(ELU_function(data), [1, 5, -0.09816843611112658, 3, -0.08646647167633874], tol=1e-10, message="ELU")
import warnings
with warnings.catch_warnings():
    warnings.simplefilter("error")
    check_close(ELU_function(np.array([1000.0, -1000.0])), [1000.0, -0.1], message="không tràn số với đầu vào lớn")
````

### 6. Hàm Softplus

<img src="images/activation_softplus.png" style="width:30%;height:30%;">

$\text{softplus}(x) = \ln(1 + e^x)$. *Bổ sung:* với $x$ lớn, $e^x$ tràn số — dùng dạng ổn định $\max(x, 0) + \ln(1 + e^{-|x|})$ (`np.log1p`).

````python
def softplus_function(data):
    ### START CODE HERE ### (≈ 1 dòng)
    result = np.maximum(data, 0) + np.log1p(np.exp(-np.abs(data)))
    ### END CODE HERE ###
    return result


softplus_function(data)
````

**Đầu ra kỳ vọng:**

```
array([1.31326169, 5.00671535, 0.01814993, 3.04858735, 0.12692801])
```

````python
check_close(softplus_function(data), [1.31326169, 5.00671535, 0.01814993, 3.04858735, 0.12692801], tol=1e-7, message="softplus")
check_close(softplus_function(np.array([1000.0])), [1000.0], message="ổn định số với x lớn")
````

### 7. *(Bổ sung)* Đạo hàm

Backprop cần đạo hàm của hàm kích hoạt. Cài đặt (vector hoá):
- `sigmoid_derivative(x)` $= \sigma(x)(1 - \sigma(x))$
- `tanh_derivative(x)` $= 1 - \tanh^2(x)$
- `ReLU_derivative(x)` $= \mathbb 1[x > 0]$
- `ELU_derivative(x, alpha)` $= 1$ nếu $x > 0$, $\alpha e^x$ nếu $x \le 0$

Kiểm tra bằng sai phân hữu hạn.

````python
def sigmoid_derivative(x):
    ### START CODE HERE ### (≈ 2 dòng)
    # STARTER: return None
    s = sigmoid_function(x)
    return s * (1 - s)
    ### END CODE HERE ###


def tanh_derivative(x):
    ### START CODE HERE ### (≈ 1 dòng)
    # STARTER: return None
    return 1 - tanh_function(x) ** 2
    ### END CODE HERE ###


def ReLU_derivative(x):
    ### START CODE HERE ### (≈ 1 dòng)
    # STARTER: return None
    return (x > 0).astype(float)
    ### END CODE HERE ###


def ELU_derivative(x, alpha=0.1):
    ### START CODE HERE ### (≈ 1 dòng)
    # STARTER: return None
    return np.where(x > 0, 1.0, alpha * np.exp(np.minimum(x, 0)))
    ### END CODE HERE ###
````

````python
xs = np.array([-3.0, -0.7, 0.4, 2.5])                                  # tránh đúng x = 0 (ReLU, ELU không khả vi tại đó)
h = 1e-6
for name, f, df in [("sigmoid", sigmoid_function, sigmoid_derivative), ("tanh", tanh_function, tanh_derivative),
                    ("ReLU", ReLU_function, ReLU_derivative), ("ELU", ELU_function, ELU_derivative)]:
    numeric = (f(xs + h) - f(xs - h)) / (2 * h)
    check_close(df(xs), numeric, tol=1e-5, message=f"đạo hàm {name}")
check_close(sigmoid_derivative(np.array([0.0])), [0.25], message="σ'(0) = 0.25 — giá trị lớn nhất")
````

### 8. *(Bổ sung)* Softmax ổn định số

$\text{softmax}(\mathbf z)_k = \frac{e^{z_k}}{\sum_j e^{z_j}}$. Với $\mathbf z$ lớn, $e^{z}$ tràn số. Mẹo: trừ $\max_j z_j$ trước (kết quả không đổi vì nhân cả tử và mẫu với $e^{-\max}$). Hàm nhận ma trận, tính softmax theo **từng hàng**.

````python
def softmax_function(Z):
    ### START CODE HERE ### (≈ 2 dòng)
    # STARTER: return None
    E = np.exp(Z - Z.max(axis=1, keepdims=True))
    return E / E.sum(axis=1, keepdims=True)
    ### END CODE HERE ###
````

````python
P = softmax_function(np.array([[1.0, 2.0, 3.0], [1000.0, 1001.0, 1002.0]]))
check_close(P[0], np.exp([1, 2, 3]) / np.exp([1, 2, 3]).sum(), message="softmax đúng")
check_close(P[1], P[0], message="dịch toàn bộ logit không đổi kết quả, không tràn số")
check_close(P.sum(axis=1), [1.0, 1.0], message="tổng mỗi hàng = 1")
````

### 9. *(Bổ sung)* GELU

GELU — hàm kích hoạt mặc định của Transformer — có dạng xấp xỉ (Hendrycks & Gimpel, 2016):
$$\text{GELU}(x) \approx 0.5\,x\left(1 + \tanh\left[\sqrt{2/\pi}\,(x + 0.044715\,x^3)\right]\right)$$

````python
def GELU_function(data):
    ### START CODE HERE ### (≈ 1 dòng)
    result = 0.5 * data * (1 + np.tanh(np.sqrt(2 / np.pi) * (data + 0.044715 * data**3)))
    ### END CODE HERE ###
    return result
````

````python
from scipy.stats import norm

xg = np.linspace(-4, 4, 81)
check_close(GELU_function(xg), xg * norm.cdf(xg), tol=2e-3, message="xấp xỉ sát x·Φ(x)")
check_true(GELU_function(np.array([-0.5]))[0] < 0, "GELU hơi âm quanh 0 (khác ReLU)")
check_close(GELU_function(np.array([6.0])), [6.0], tol=1e-6, message="≈ x khi x lớn")
````
