```python
import matplotlib.pyplot as plt
import numpy as np

from ml_course.checks import check_close, check_equal, check_true
```

**Bài tập 1:** Vẽ đồ thị đường thẳng `y = 4*x + 3` với:
- tiêu đề hình: `'Linear Function'`
- tiêu đề trục x: `'x axis'`, trục y: `'y axis'`

```python
x = np.arange(3, 10)
y = 4 * x + 3

fig, ax = plt.subplots(figsize=(6, 4))
### START CODE HERE ### (≈ 4 dòng)
ax.set_title('Linear Function')
ax.set_xlabel('x axis')
ax.set_ylabel('y axis')
ax.plot(x, y)
### END CODE HERE ###
plt.show()
```

**Bài tập 2:** Vẽ đồ thị hàm `tanh` và đạo hàm của nó, với trục toạ độ ở giữa.

$$\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}, \qquad \tanh'(x) = 1 - \tanh^2(x)$$

```python
def tanh(x):
    ex = np.exp(x)
    emx = np.exp(-x)
    return (ex - emx) / (ex + emx)


x = np.arange(-5., 5., 0.1)
a = tanh(x)
dx = 1 - a**2

# Tạo figure + axes với figsize = (10, 5)
### START CODE HERE ### (≈ 1 dòng)
# STARTER: fig, ax = None, None
fig, ax = plt.subplots(figsize=(10, 5))
### END CODE HERE ###

ax.spines['left'].set_position('center')
ax.spines['bottom'].set_position('center')
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')
ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')

# Vẽ hàm tanh: color='#d35400', linewidth=3, label='tanh'
### START CODE HERE ### (≈ 1 dòng)
ax.plot(x, a, color='#d35400', linewidth=3, label='tanh')
### END CODE HERE ###

# Vẽ đạo hàm: color='#1abd15', linewidth=3, label='derivative'
### START CODE HERE ### (≈ 1 dòng)
ax.plot(x, dx, color='#1abd15', linewidth=3, label='derivative')
### END CODE HERE ###

# Hiển thị chú thích: loc='upper left', frameon=False
### START CODE HERE ### (≈ 1 dòng)
ax.legend(loc='upper left', frameon=False)
### END CODE HERE ###
plt.show()
```

```python
check_close(fig.get_size_inches(), [10, 5], message="figsize = (10, 5)")
check_equal(len(ax.lines), 2, "vẽ 2 đường")
check_equal([line.get_label() for line in ax.lines], ['tanh', 'derivative'], "nhãn các đường")
check_close(ax.lines[1].get_ydata(), 1 - np.tanh(np.arange(-5., 5., 0.1)) ** 2, message="dữ liệu đạo hàm")
check_true(ax.get_legend() is not None, "có chú thích (legend)")
```

**Bài tập 3**: Tương tự bài tập 2, hãy **tự xây dựng từ đầu** đồ thị hàm `ReLU` và đạo hàm của nó (trục ở giữa, 2 đường có label `'ReLU'` và `'derivative'`, có legend):

$$f(x) = \begin{cases} 0 & x < 0 \\ x & x \ge 0 \end{cases} \qquad f'(x) = \begin{cases} 0 & x < 0 \\ 1 & x \ge 0 \end{cases}$$

```python
x = np.arange(-5., 5., 0.1)
### START CODE HERE ### (≈ 12 dòng)
# STARTER: relu = ...
# STARTER: d_relu = ...
# STARTER: fig, ax = plt.subplots(figsize=(10, 5))
# STARTER: ...
relu = np.maximum(0, x)
d_relu = (x >= 0).astype(float)
fig, ax = plt.subplots(figsize=(10, 5))
ax.spines['left'].set_position('center')
ax.spines['bottom'].set_position('center')
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')
ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')
ax.plot(x, relu, color='#d35400', linewidth=3, label='ReLU')
ax.plot(x, d_relu, color='#1abd15', linewidth=3, label='derivative')
ax.legend(loc='upper left', frameon=False)
### END CODE HERE ###
plt.show()
```

```python
check_equal([line.get_label() for line in ax.lines], ['ReLU', 'derivative'], "nhãn các đường")
check_close(ax.lines[0].get_ydata(), np.maximum(0, np.arange(-5., 5., 0.1)), message="dữ liệu ReLU")
check_close(ax.lines[1].get_ydata(), (np.arange(-5., 5., 0.1) >= 0).astype(float), message="dữ liệu đạo hàm ReLU")
```

**Bài tập 4**: Vẽ dữ liệu kiểu điểm (scatter) với các thông số: `marker='x'`, `c='r'`, `s=20`; tiêu đề `'Bài tập 4'`.

```python
x = [1, 2, 5, 6, 8, 10, 12, 3.4, 3.6, 8]
y = [-1, 3, 5, 9, 8, 0, -3, 10, 9, 8]

fig, ax = plt.subplots()
### START CODE HERE ### (≈ 1 dòng)
ax.scatter(x, y, marker='x', c='r', s=20)
### END CODE HERE ###
ax.set_title('Bài tập 4')
ax.set_xlabel("Trục x")
ax.set_ylabel("Trục y")
plt.show()
```

```python
check_equal(len(ax.collections), 1, "có 1 scatter")
check_close(ax.collections[0].get_offsets(), np.column_stack([x, y]), message="toạ độ các điểm")
check_close(ax.collections[0].get_sizes(), [20], message="kích thước điểm s=20")
```

**Bài tập 5 (bổ sung — ML):** Dữ liệu 2 lớp `X, y` bên dưới. Vẽ **scatter tô màu theo lớp** (lớp 0 màu xanh `'tab:blue'`, lớp 1 màu đỏ `'tab:red'`, mỗi lớp một lời gọi `ax.scatter` với `label='class 0'` / `'class 1'`) và **histogram** của feature thứ nhất cho mỗi lớp trên một axes thứ hai. Dùng `fig, axes = plt.subplots(1, 2, figsize=(12, 4))`.

```python
from ml_course.datasets import make_classification_2d

X, y = make_classification_2d("blobs", n=300, seed=1)
### START CODE HERE ### (≈ 8 dòng)
# STARTER: fig, axes = None, None
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for label, color in [(0, 'tab:blue'), (1, 'tab:red')]:
    axes[0].scatter(X[y == label, 0], X[y == label, 1], c=color, s=12, label=f'class {label}')
    axes[1].hist(X[y == label, 0], bins=20, alpha=0.6, color=color, label=f'class {label}')
axes[0].legend()
axes[1].legend()
axes[0].set_title("Scatter theo lớp")
axes[1].set_title("Histogram feature 1")
### END CODE HERE ###
plt.show()
```

```python
check_equal(len(axes[0].collections), 2, "2 scatter (mỗi lớp một)")
check_equal(sorted(c.get_label() for c in axes[0].collections), ['class 0', 'class 1'], "nhãn scatter")
check_true(len(axes[1].patches) > 0, "có histogram ở axes thứ hai")
```
