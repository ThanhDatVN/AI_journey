### 1.1 — Hàm sigmoid, `np.exp()`

Trước khi dùng `np.exp()`, bạn sẽ dùng `math.exp()` để cài đặt hàm sigmoid, và sẽ thấy lý do tại sao chúng ta dùng `np.exp()` thay cho `math.exp()`.

$$\text{sigmoid}(x) = \frac{1}{1+e^{-x}}$$

Đây là hàm phi tuyến được dùng nhiều trong cả Machine Learning (Logistic Regression — chương 07) lẫn Deep Learning.

<img src="images/Sigmoid.png" style="width:400px">

**Bài tập:** cài đặt `basic_sigmoid(x)` với $x$ là **số thực**, dùng `math.exp()`.

````python
def basic_sigmoid(x):
    """
    Tính sigmoid của số thực x.

    Arguments:
    x -- một số thực (scalar)

    Return:
    s -- sigmoid(x)
    """
    ### START CODE HERE ### (≈ 1 dòng)
    s = 1 / (1 + math.exp(-x))
    ### END CODE HERE ###
    return s


basic_sigmoid(10)
````

````python
check_close(basic_sigmoid(10), 0.9999546021312976, message="basic_sigmoid(10)")
````

**Đầu ra kỳ vọng**: `0.9999546021312976`

Trong thực tế, ta hầu như không dùng thư viện `math` trong ML/DL vì đầu vào của nó chỉ là số thực, trong khi đầu vào của ta là **ma trận và vector**:

````python
x = [1, 2, 3]
try:
    basic_sigmoid(x)       # lỗi vì x là vector
except TypeError as e:
    print("TypeError:", e)
````

Nếu $x = (x_1, x_2, ..., x_n)$ là một vector, `np.exp(x)` tính trên **từng phần tử**: $np.exp(x) = (e^{x_1}, e^{x_2}, ..., e^{x_n})$. Tương tự, `x + 3` hay `1 / x` cho đầu ra cùng kích thước với $x$:

````python
x = np.array([1, 2, 3])
print(np.exp(x))
print("x + 3:", x + 3)
print("1 / x:", 1 / x)
````

**Bài tập:** cài đặt `sigmoid(x)` bằng NumPy — $x$ có thể là số, vector hay ma trận:

$$\text{với } x \in \mathbb{R}^n,\quad \text{sigmoid}\begin{pmatrix} x_1 \\ \vdots \\ x_n \end{pmatrix} = \begin{pmatrix} \frac{1}{1+e^{-x_1}} \\ \vdots \\ \frac{1}{1+e^{-x_n}} \end{pmatrix}$$

````python
def sigmoid(x):
    """
    Arguments:
    x -- số thực hoặc numpy array kích thước bất kỳ

    Return:
    s -- sigmoid(x), cùng shape với x
    """
    ### START CODE HERE ### (≈ 1 dòng)
    s = 1 / (1 + np.exp(-x))
    ### END CODE HERE ###
    return s


sigmoid(np.array([1, 2, 3]))
````

````python
check_close(sigmoid(np.array([1, 2, 3])), [0.73105858, 0.88079708, 0.95257413], tol=1e-7)
check_shape(sigmoid(np.zeros((4, 5))), (4, 5))
````

**Đầu ra kỳ vọng**: `array([0.73105858, 0.88079708, 0.95257413])`

### 1.2 — Đạo hàm của hàm sigmoid

Đạo hàm cần cho thuật toán **gradient descent** và **backpropagation**:

$$\sigma'(x) = \sigma(x)\,(1 - \sigma(x))$$

**Bài tập:** cài đặt `sigmoid_derivative(x)` theo 2 bước: (1) tính `s = sigmoid(x)` bằng hàm bạn vừa viết; (2) tính `ds = s * (1 - s)`.

````python
def sigmoid_derivative(x):
    """
    Tính đạo hàm của sigmoid theo x.

    Arguments:
    x -- số thực hoặc numpy array

    Return:
    ds -- gradient đã tính
    """
    ### START CODE HERE ### (≈ 2 dòng)
    s = sigmoid(x)
    ds = s * (1 - s)
    ### END CODE HERE ###
    return ds


print("sigmoid_derivative(x) =", sigmoid_derivative(np.array([1, 2, 3])))
````

````python
check_close(sigmoid_derivative(np.array([1, 2, 3])), [0.19661193, 0.10499359, 0.04517666], tol=1e-7)
# Kiểm tra bằng đạo hàm số: (σ(x+h) − σ(x−h)) / 2h
h = 1e-5
x = np.linspace(-4, 4, 9)
check_close(sigmoid_derivative(x), (sigmoid(x + h) - sigmoid(x - h)) / (2 * h), tol=1e-6,
            message="khớp với đạo hàm số (gradient check)")
````

**Đầu ra kỳ vọng**: `[0.19661193 0.10499359 0.04517666]`

### 1.3 — Reshaping arrays

Hai thao tác dùng nhiều trong ML/DL: `X.shape` (lấy kích thước) và `X.reshape(...)` (định hình lại).

**Câu hỏi phụ:** hãy phân biệt `Transpose` và `Reshape` qua ví dụ sau:

````python
A = np.array([[1, 2], [3, 4], [5, 6]])
print('Matrix A:\n', A)
print('Transpose:\n', A.T)
print('Reshape (2, 3):\n', A.reshape(2, 3))
````

Một ảnh màu được biểu diễn bởi mảng 3 chiều shape $(length, height, depth=3)$. Khi đưa ảnh vào nhiều thuật toán, ta cần "duỗi" (unroll) nó thành vector shape $(length \times height \times 3, 1)$.

<img src="images/image2vector_kiank.png" style="width:450px">

**Bài tập:** cài đặt `image2vector(image)` nhận mảng shape `(length, height, depth)` và trả về vector shape `(length*height*depth, 1)`.

````python
def image2vector(image):
    """
    Argument:
    image -- numpy array shape (length, height, depth)

    Returns:
    v -- vector shape (length*height*depth, 1)
    """
    ### START CODE HERE ### (≈ 1 dòng)
    v = image.reshape(-1, 1)
    ### END CODE HERE ###
    return v


# Mảng 3 × 3 × 2; ảnh thật thường là (num_px_x, num_px_y, 3) với 3 kênh RGB
image = np.array([[[0.67826139, 0.29380381], [0.90714982, 0.52835647], [0.4215251, 0.45017551]],
                  [[0.92814219, 0.96677647], [0.85304703, 0.52351845], [0.19981397, 0.27417313]],
                  [[0.60659855, 0.00533165], [0.10820313, 0.49978937], [0.34144279, 0.94630077]]])
print("image2vector(image) =\n", image2vector(image))
````

````python
v = image2vector(image)
check_shape(v, (18, 1))
check_close(v[:3, 0], [0.67826139, 0.29380381, 0.90714982])
````

**Đầu ra kỳ vọng**: vector cột 18 phần tử, bắt đầu bằng `[[0.67826139] [0.29380381] [0.90714982] ...`

### 1.4 — Chuẩn hoá theo hàng (normalizing rows)

Chuẩn hoá dữ liệu giúp gradient descent hội tụ nhanh hơn. Ở đây ta chuẩn hoá mỗi hàng của $x$ thành vector độ dài 1: chia mỗi hàng cho **chuẩn (norm)** của nó.

Ví dụ, với
$$x = \begin{bmatrix} 0 & 3 & 4 \\ 2 & 6 & 4 \end{bmatrix},\quad \lVert x \rVert = \texttt{np.linalg.norm(x, axis=1, keepdims=True)} = \begin{bmatrix} 5 \\ \sqrt{56} \end{bmatrix},\quad \frac{x}{\lVert x\rVert} = \begin{bmatrix} 0 & \frac{3}{5} & \frac{4}{5} \\ \frac{2}{\sqrt{56}} & \frac{6}{\sqrt{56}} & \frac{4}{\sqrt{56}} \end{bmatrix}$$

**Bài tập:** cài đặt `normalize_rows(x)`.

````python
def normalize_rows(x):
    """
    Chuẩn hoá mỗi hàng của ma trận x để có độ dài 1.

    Argument:
    x -- numpy matrix shape (n, m)

    Returns:
    x -- ma trận đã chuẩn hoá theo hàng
    """
    ### START CODE HERE ### (≈ 2 dòng)
    x_norm = np.linalg.norm(x, ord=2, axis=1, keepdims=True)
    x = x / x_norm
    ### END CODE HERE ###
    return x


x = np.array([[0, 3, 4], [1, 6, 4]])
print("normalize_rows(x) =\n", normalize_rows(x))
````

````python
out = normalize_rows(np.array([[0, 3, 4], [1, 6, 4]]))
check_close(out, [[0, 0.6, 0.8], [0.13736056, 0.82416338, 0.54944226]], tol=1e-7)
check_close(np.linalg.norm(out, axis=1), [1, 1], message="mỗi hàng có độ dài 1")
````

**Đầu ra kỳ vọng**:
```
[[0.         0.6        0.8       ]
 [0.13736056 0.82416338 0.54944226]]
```

**Chú ý:** `x_norm` có shape `(2, 1)` còn `x` có shape `(2, 3)` — phép chia vẫn hoạt động nhờ **broadcasting**. Thử bỏ `keepdims=True` và quan sát lỗi.

### 1.5 — Broadcasting và hàm softmax

<img src="images/broadcasting.png" style="width:650px">

**Softmax** biến một vector điểm số thành **phân phối xác suất** — dùng cho phân loại nhiều lớp (chương 07, 28). Với ma trận $x \in \mathbb{R}^{m \times n}$, softmax áp dụng trên **từng hàng**:

$$\text{softmax}(x)_{ij} = \frac{e^{x_{ij}}}{\sum_k e^{x_{ik}}}$$

**Bài tập:** cài đặt `softmax(x)` theo từng hàng. Thêm bước **trừ max của mỗi hàng** trước khi lấy `exp` để tránh tràn số (kết quả không đổi vì $\frac{e^{x-c}}{\sum e^{x-c}} = \frac{e^{x}}{\sum e^{x}}$).

````python
def softmax(x):
    """
    Softmax theo từng hàng của x.

    Argument:
    x -- numpy matrix shape (m, n)

    Returns:
    s -- numpy matrix shape (m, n), mỗi hàng là một phân phối xác suất
    """
    ### START CODE HERE ### (≈ 4 dòng)
    x = x - x.max(axis=1, keepdims=True)
    x_exp = np.exp(x)
    x_sum = x_exp.sum(axis=1, keepdims=True)
    s = x_exp / x_sum
    ### END CODE HERE ###
    return s


x = np.array([[9, 2, 5, 0, 0], [7, 5, 0, 0, 0]])
print("softmax(x) =\n", softmax(x))
````

````python
s = softmax(np.array([[9, 2, 5, 0, 0], [7, 5, 0, 0, 0]]))
check_close(s[0, :3], [9.80897665e-01, 8.94462891e-04, 1.79657674e-02], tol=1e-7)
check_close(s.sum(axis=1), [1, 1], message="mỗi hàng có tổng bằng 1")
check_close(softmax(np.array([[1000.0, 1000.0]])), [[0.5, 0.5]], message="ổn định số học với giá trị lớn")
````

**Đầu ra kỳ vọng**:
```
[[9.80897665e-01 8.94462891e-04 1.79657674e-02 1.21052389e-04 1.21052389e-04]
 [8.78679856e-01 1.18916387e-01 8.01252314e-04 8.01252314e-04 8.01252314e-04]]
```

**Những điều bạn nên nhớ:**

- Hàm sigmoid và đạo hàm của nó; kiểm tra đạo hàm bằng sai phân số.

## 2 — Vectorization

Trong ML/DL, bạn làm việc với bộ dữ liệu lớn, vì vậy code phải tính toán hiệu quả. Hãy so sánh phép toán dot / outer / element-wise bằng vòng lặp và bằng vectorization (dùng vector 100 000 phần tử để thấy rõ khác biệt):

````python
rng = np.random.default_rng(0)
x1 = rng.integers(0, 10, 100_000)
x2 = rng.integers(0, 10, 100_000)

tic = time.perf_counter()
dot = 0
for i in range(len(x1)):
    dot += x1[i] * x2[i]
t_loop = time.perf_counter() - tic

tic = time.perf_counter()
dot_vec = np.dot(x1, x2)
t_vec = time.perf_counter() - tic

print(f"dot (vòng lặp)   = {dot} — {1000 * t_loop:.1f} ms")
print(f"dot (vectorized) = {dot_vec} — {1000 * t_vec:.3f} ms")
````

````python
small1, small2 = x1[:300], x2[:300]
tic = time.perf_counter()
outer = np.zeros((len(small1), len(small2)))
for i in range(len(small1)):
    for j in range(len(small2)):
        outer[i, j] = small1[i] * small2[j]
t_loop = time.perf_counter() - tic
tic = time.perf_counter()
outer_vec = np.outer(small1, small2)
t_vec = time.perf_counter() - tic
print(f"outer: vòng lặp {1000 * t_loop:.1f} ms vs vectorized {1000 * t_vec:.3f} ms — giống nhau: {np.allclose(outer, outer_vec)}")
````

**Lưu ý:** `np.dot()` / `@` thực hiện phép nhân ma trận–ma trận hoặc ma trận–vector. Nó **khác** với `np.multiply()` và `*` (nhân từng phần tử).

### 2.1 — Hàm loss L1 và L2

Hàm loss đánh giá hiệu suất của model: loss lớn nghĩa là giá trị dự đoán $\hat{y}$ sai khác nhiều so với giá trị thực $y$.

**Bài tập:** cài đặt L1 loss bằng NumPy (vectorized):
$$L_1(\hat{y}, y) = \sum_{i=1}^m \left|y^{(i)} - \hat{y}^{(i)}\right|$$

````python
def L1(yhat, y):
    """
    Arguments:
    yhat -- vector kích thước m (giá trị dự đoán)
    y -- vector kích thước m (giá trị thật)

    Returns:
    loss -- giá trị L1 loss
    """
    ### START CODE HERE ### (≈ 1 dòng)
    loss = np.sum(np.abs(y - yhat))
    ### END CODE HERE ###
    return loss


yhat = np.array([.9, 0.2, 0.1, .4, .9])
y = np.array([1, 0, 0, 1, 1])
print("L1 =", L1(yhat, y))
````

````python
check_close(L1(np.array([.9, 0.2, 0.1, .4, .9]), np.array([1, 0, 0, 1, 1])), 1.1)
````

**Đầu ra kỳ vọng**: `L1 = 1.1`

**Bài tập:** cài đặt L2 loss. Gợi ý: với $x = [x_1, ..., x_n]$, `np.dot(x, x)` $= \sum_j x_j^2$.

$$L_2(\hat{y}, y) = \sum_{i=1}^m \left(y^{(i)} - \hat{y}^{(i)}\right)^2$$

````python
def L2(yhat, y):
    """
    Arguments:
    yhat -- vector kích thước m (giá trị dự đoán)
    y -- vector kích thước m (giá trị thật)

    Returns:
    loss -- giá trị L2 loss
    """
    ### START CODE HERE ### (≈ 1 dòng)
    loss = np.dot(y - yhat, y - yhat)
    ### END CODE HERE ###
    return loss


print("L2 =", L2(yhat, y))
````

````python
check_close(L2(np.array([.9, 0.2, 0.1, .4, .9]), np.array([1, 0, 0, 1, 1])), 0.43)
````

**Đầu ra kỳ vọng**: `L2 = 0.43`
