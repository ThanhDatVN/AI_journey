**Bài tập 1:** Chuyển góc từ độ sang radian. Công thức chuyển từ góc đơn vị độ $A_d$ sang góc đơn vị radian $A_r$:

$$A_r = A_d \times \frac{\pi}{180}$$

*Gợi ý:* lấy số `pi` bằng `math.pi`.

```python
def degree_to_radian(degree):
    ### START CODE HERE ### (≈ 1 dòng)
    radian = degree * math.pi / 180
    ### END CODE HERE ###
    return radian


print("40 độ sang radian:", degree_to_radian(40))
```

```python
check_close(degree_to_radian(40), 0.6981317007977318, message="40 độ → radian")
check_close(degree_to_radian(180), math.pi, message="180 độ → π")
```

**Đầu ra kỳ vọng**: `40 độ sang radian: 0.6981317007977318`

**Bài tập 2:** Kiểm tra có phải tam giác vuông dựa vào định lý Pythagoras $a^2 + b^2 = c^2$ (với $c$ là cạnh lớn nhất).

<img src="images/tam-giac-vuong.png" style="width:30%">

Viết hàm `is_right_triangle(a, b, c)`:
1. Nếu có cạnh nào không dương → trả về `False`.
2. Sắp xếp 3 cạnh để cạnh lớn nhất ở cuối, rồi kiểm tra định lý (dùng `math.isclose` vì có thể là số thực).

```python
def is_right_triangle(a, b, c):
    ### START CODE HERE ### (≈ 4 dòng)
    if min(a, b, c) <= 0:
        return False
    x, y, z = sorted([a, b, c])
    answer = math.isclose(x**2 + y**2, z**2)
    ### END CODE HERE ###
    return answer


print("(3, 4, 5):", is_right_triangle(3, 4, 5))
print("(3, 4, 6):", is_right_triangle(3, 4, 6))
```

```python
check_equal(is_right_triangle(3, 4, 5), True, "(3, 4, 5) là tam giác vuông")
check_equal(is_right_triangle(3, 4, 6), False, "(3, 4, 6) không phải tam giác vuông")
check_equal(is_right_triangle(5, 3, 4), True, "thứ tự cạnh không quan trọng")
check_equal(is_right_triangle(-3, 4, 5), False, "cạnh âm → False")
```

**Đầu ra kỳ vọng**: `(3, 4, 5): True` và `(3, 4, 6): False`

**Bài tập 3:** Tính chu vi và diện tích cho các hình cơ bản.

<img src="images/toan.png" style="width:50%">

Viết các hàm sau (mỗi hàm trả về **tuple** `(chu_vi, diện_tích)`):
- `triangle(a, b, c)`: tam giác 3 cạnh — diện tích theo **công thức Heron**: $p = \frac{a+b+c}{2}$, $S = \sqrt{p(p-a)(p-b)(p-c)}$
- `rectangle(w, h)`: hình chữ nhật
- `circle(r)`: hình tròn

```python
def triangle(a, b, c):
    ### START CODE HERE ### (≈ 3 dòng)
    perimeter = a + b + c
    p = perimeter / 2
    area = math.sqrt(p * (p - a) * (p - b) * (p - c))
    ### END CODE HERE ###
    return perimeter, area


def rectangle(w, h):
    ### START CODE HERE ### (≈ 2 dòng)
    perimeter = 2 * (w + h)
    area = w * h
    ### END CODE HERE ###
    return perimeter, area


def circle(r):
    ### START CODE HERE ### (≈ 2 dòng)
    perimeter = 2 * math.pi * r
    area = math.pi * r**2
    ### END CODE HERE ###
    return perimeter, area


print("Tam giác (3, 4, 5):", triangle(3, 4, 5))
print("Chữ nhật (2, 3):   ", rectangle(2, 3))
print("Hình tròn r=1:     ", circle(1))
```

```python
check_close(triangle(3, 4, 5), (12, 6), message="tam giác 3-4-5")
check_close(rectangle(2, 3), (10, 6), message="chữ nhật 2×3")
check_close(circle(1), (2 * math.pi, math.pi), message="hình tròn r=1")
```

**Đầu ra kỳ vọng**: `Tam giác (3, 4, 5): (12, 6.0)`, `Chữ nhật (2, 3): (10, 6)`, `Hình tròn r=1: (6.283..., 3.1415...)`

**Bài tập 4:** Ước lượng số $\pi$ bằng các số ngẫu nhiên (phương pháp **Monte Carlo**).

1. Cho hình vuông $V$ cạnh $s = 2$ và hình tròn $T$ bán kính $r = 1$, cùng tâm $(0, 0)$.
2. Sinh $N$ điểm $p = (x, y)$ với $x, y$ có phân phối đều trên $[-1, 1]$.
3. Đếm số điểm thuộc $T$ (tức $x^2 + y^2 \le 1$), gọi là $N_T$.
4. Vì $\frac{\text{diện tích } T}{\text{diện tích } V} = \frac{\pi}{4}$, ta có $\pi \approx 4 \times \frac{N_T}{N}$.

<img src="images/pi_estimation.png" style="width:30%">

*Gợi ý:* `random.random()` sinh số trong $[0, 1)$ → biến đổi thành $[-1, 1)$ bằng `2 * random.random() - 1`.

```python
def estimate_pi(N, seed=0):
    random.seed(seed)
    N_T = 0
    for _ in range(N):
        ### START CODE HERE ### (≈ 4 dòng)
        x = 2 * random.random() - 1
        y = 2 * random.random() - 1
        if x**2 + y**2 <= 1:
            N_T += 1
        ### END CODE HERE ###
    ### START CODE HERE ### (≈ 1 dòng)
    pi = 4 * N_T / N
    ### END CODE HERE ###
    return pi


print(estimate_pi(100_000))
```

```python
check_close(estimate_pi(100_000), math.pi, tol=0.02, message="π ≈ 3.14 (sai số ≤ 0.02)")
errors = [abs(estimate_pi(n) - math.pi) for n in (100, 10_000, 200_000)]
print("Sai số theo N = 100, 10 000, 200 000:", [round(e, 4) for e in errors])
check_true(errors[2] < errors[0], "N càng lớn, ước lượng càng chính xác")
```

**Đầu ra kỳ vọng**: một số gần `3.14` (ví dụ `3.14552`, tuỳ seed).

> 💡 **Liên hệ ML:** Monte Carlo — ước lượng một đại lượng bằng cách lấy mẫu ngẫu nhiên — xuất hiện ở khắp nơi: bootstrap (chương 02), Random Forest (chương 11), dropout (chương 28), Reinforcement Learning (chương 35). Sai số giảm theo $1/\sqrt{N}$.

**Bài tập 5:** Số $e$ được ước lượng dựa vào chuỗi Taylor:

$$e \approx 1 + \frac{1}{1!} + \frac{1}{2!} + \dots + \frac{1}{n!}$$

$n$ càng lớn thì độ chính xác càng cao. Viết hàm `factorial(n)` (dùng vòng lặp `for`, không dùng `math.factorial`) và `estimate_e(n)`.

```python
def factorial(n):
    ### START CODE HERE ### (≈ 3 dòng)
    result = 1
    for k in range(2, n + 1):
        result *= k
    ### END CODE HERE ###
    return result


def estimate_e(n):
    ### START CODE HERE ### (≈ 1 dòng)
    result = sum(1 / factorial(k) for k in range(n + 1))
    ### END CODE HERE ###
    return result


print(estimate_e(10))
```

```python
check_equal(factorial(0), 1, "0! = 1")
check_equal(factorial(5), 120, "5! = 120")
check_close(estimate_e(10), 2.7182818011463845, message="estimate_e(10)")
```

**Đầu ra kỳ vọng**: `2.7182818011463845`
